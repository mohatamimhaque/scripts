````python
# ==============================================================================
# Cell 2: AutoCrat Pro Production Core Engine (with QR & Image Support)
# ==============================================================================
import os
import re
import sys
import copy
import time
import hashlib
import threading
import tempfile
import urllib.request
import concurrent.futures
from typing import List, Dict, Optional, Tuple, Any
import pandas as pd
from pptx import Presentation
from pptx.util import Inches, Pt
import pypdf
import pypdf
from pypdf import PdfWriter, PdfReader, PdfWriter
from PIL import Image, ImageDraw, ImageColor, ImageOps, ImageTk
import tkinter as tk
from tkinter import ttk, colorchooser
import zxingcpp
try:
    from qrcode.image.styledpil import StyledPilImage
    from qrcode.image.styles.moduledrawers.pil import (
        SquareModuleDrawer, CircleModuleDrawer, RoundedModuleDrawer, GappedSquareModuleDrawer
    )
    from qrcode.image.styles.colormasks import SolidFillColorMask
except ImportError:
    pass


try:
    from tqdm import tqdm
except ImportError:
    def tqdm(iterable, **kwargs):
        return iterable


# ------------------------------------------------------------------------------
# 1. PowerPoint to PDF COM Converter
# ------------------------------------------------------------------------------
class PPTXToPDFConverter:
    """
    Manages a persistent Microsoft PowerPoint COM application instance for 
    ultra-fast batch PDF rendering on Windows.
    """
    def __init__(self):
        self.powerpoint = None
        self.com_initialized = False
        self.backend = None

    def start(self) -> bool:
        if self.com_initialized and self.powerpoint is not None:
            return True

        # Attempt 1: win32com.client
        try:
            import win32com.client
            self.powerpoint = win32com.client.Dispatch("Powerpoint.Application")
            self.com_initialized = True
            self.backend = "win32com"
            print("  [OK] PowerPoint COM engine initialized successfully (win32com).")
            return True
        except Exception:
            pass

        # Attempt 2: comtypes.client
        try:
            import comtypes.client
            self.powerpoint = comtypes.client.CreateObject("Powerpoint.Application")
            self.com_initialized = True
            self.backend = "comtypes"
            print("  [OK] PowerPoint COM engine initialized successfully (comtypes).")
            return True
        except Exception:
            pass

        print("  [WARN] Could not initialize PowerPoint COM. Native PDF conversion disabled.")
        print("         Ensure Microsoft Office PowerPoint is installed on this Windows machine.")
        return False

    def convert(self, pptx_path: str, pdf_path: str) -> bool:
        if not self.com_initialized or self.powerpoint is None:
            return False
        
        abs_pptx = os.path.abspath(pptx_path)
        abs_pdf = os.path.abspath(pdf_path)

        deck = None
        try:
            deck = self.powerpoint.Presentations.Open(abs_pptx, WithWindow=False)
            deck.SaveAs(abs_pdf, 32)
            deck.Close()
            deck = None
            return True
        except Exception as err:
            print(f"    [ERR] COM conversion error for {os.path.basename(pptx_path)}: {err}")
            if deck is not None:
                try: deck.Close()
                except Exception: pass
            return False

    def stop(self):
        if self.powerpoint is not None:
            try:
                import pythoncom
                pythoncom.CoUninitialize()
            except Exception:
                pass
            try: self.powerpoint.Quit()
            except Exception: pass
            self.powerpoint = None
            self.com_initialized = False
            self.backend = None
            print("  [OK] PowerPoint COM engine released.")


# ------------------------------------------------------------------------------
# 2. Intelligent Placeholder Replacement Engine
# ------------------------------------------------------------------------------
class PlaceholderEngine:
    """
    Advanced placeholder replacer supporting:
    - Fuzzy space and case normalization
    - Regular text frames
    - Table cells (rows & columns)
    - Grouped shapes (recursive traversal)
    - Multi-run formatting preservation
    """
    @staticmethod
    def normalize_key(key: Any) -> str:
        return re.sub(r"[\s_]+", "", str(key)).lower()

    @classmethod
    def build_mapping(cls, row: pd.Series, custom_mapping: Optional[Dict[str, str]] = None) -> Dict[str, str]:
        mapping = {}
        for col_name, val in row.items():
            str_val = "" if pd.isna(val) else str(val).strip()
            if re.match(r"^\d+\.0$", str_val):
                str_val = str_val[:-2]
            raw_key = str(col_name).strip()
            norm_key = cls.normalize_key(raw_key)
            mapping[raw_key] = str_val
            mapping[norm_key] = str_val

        # Apply explicit user placeholder mappings: {template_placeholder: excel_column_name}
        if custom_mapping:
            for ph_tag, col_name in custom_mapping.items():
                if col_name and col_name in row:
                    val = row[col_name]
                    str_val = "" if pd.isna(val) else str(val).strip()
                    if re.match(r"^\d+\.0$", str_val):
                        str_val = str_val[:-2]
                    raw_tag = str(ph_tag).strip()
                    norm_tag = cls.normalize_key(raw_tag)
                    mapping[raw_tag] = str_val
                    mapping[norm_tag] = str_val
        return mapping

    @classmethod
    def replace_text(cls, text: str, mapping: Dict[str, str]) -> str:
        if not text or "{{" not in text:
            return text

        def replacer(match):
            raw_token = match.group(1).strip()
            norm_token = cls.normalize_key(raw_token)
            if raw_token in mapping:
                return mapping[raw_token]
            if norm_token in mapping:
                return mapping[norm_token]
            return match.group(0)

        return re.sub(r"\{\{\s*([^}]+)\s*\}\}", replacer, text)

    @classmethod
    def process_text_frame(cls, text_frame, mapping: Dict[str, str]):
        for para in text_frame.paragraphs:
            full_text = "".join(run.text for run in para.runs)
            # Skip dedicated QR and Image placeholder shapes; handled by their respective engines
            if re.search(r"^\{\{\s*(qr|image|photo|picture)\s*\}\}$", full_text.strip(), re.IGNORECASE):
                continue
            if "{{" in full_text:
                new_text = cls.replace_text(full_text, mapping)
                for i in range(len(para.runs) - 1, 0, -1):
                    try: para.runs[i].text = ""
                    except Exception: pass
                if len(para.runs) > 0:
                    para.runs[0].text = new_text
                else:
                    para.add_run().text = new_text

    @classmethod
    def process_shape(cls, shape, mapping: Dict[str, str]):
        if shape.has_text_frame:
            cls.process_text_frame(shape.text_frame, mapping)
        if shape.has_table:
            for row in shape.table.rows:
                for cell in row.cells:
                    cls.process_text_frame(cell.text_frame, mapping)
        if hasattr(shape, "shapes"):
            for child_shape in shape.shapes:
                cls.process_shape(child_shape, mapping)

    @classmethod
    def process_slide(cls, slide, mapping: Dict[str, str]):
        for shape in slide.shapes:
            cls.process_shape(shape, mapping)

    @classmethod
    def extract_placeholders(cls, prs: Presentation) -> List[str]:
        found = set()

        def scan_text(txt):
            if txt and "{{" in txt:
                for m in re.finditer(r"\{\{\s*([^}]+)\s*\}\}", txt):
                    found.add(m.group(1).strip())

        def scan_shape(sh):
            if sh.has_text_frame:
                for p in sh.text_frame.paragraphs:
                    scan_text(p.text)
            if sh.has_table:
                for r in sh.table.rows:
                    for c in r.cells:
                        for p in c.text_frame.paragraphs:
                            scan_text(p.text)
            if hasattr(sh, "shapes"):
                for child in sh.shapes:
                    scan_shape(child)

        for s in prs.slides:
            for sh in s.shapes:
                scan_shape(sh)

        return sorted(list(found))


# ------------------------------------------------------------------------------
# 3. Slide Duplication Engine
# ------------------------------------------------------------------------------
class SlideDuplicator:
    @staticmethod
    def duplicate_slide(prs: Presentation, source_slide):
        layout = source_slide.slide_layout
        new_slide = prs.slides.add_slide(layout)

        source_cSld = source_slide.element.cSld
        new_cSld = new_slide.element.cSld
        bg_element = None
        for child in source_cSld:
            if child.tag.endswith("}bg"):
                bg_element = child
                break

        if bg_element is not None:
            for child in list(new_cSld):
                if child.tag.endswith("}bg"):
                    new_cSld.remove(child)
            cloned_bg = copy.deepcopy(bg_element)
            new_cSld.insert(0, cloned_bg)

        for shape in list(new_slide.shapes):
            new_slide.shapes.element.remove(shape.element)

        for shape in source_slide.shapes:
            cloned_el = copy.deepcopy(shape.element)
            new_slide.shapes._spTree.insert_element_before(cloned_el, "p:extLst")

        for rel_id, rel in source_slide.part.rels.items():
            if "notesSlide" not in rel.reltype and "slideLayout" not in rel.reltype:
                try:
                    new_slide.part.rels.add_relationship(rel.reltype, rel._target, rel.rId)
                except Exception:
                    pass

        return new_slide


# ------------------------------------------------------------------------------
# 4. Data Loader & File Utilities
# ------------------------------------------------------------------------------
class DataLoader:
    @staticmethod
    def load(data_path: str, sheet_name: Optional[Any] = None) -> pd.DataFrame:
        clean_path = str(data_path).strip().strip("'\"")
        if not os.path.exists(clean_path):
            raise FileNotFoundError(f"Data file does not exist: {clean_path}")

        ext = os.path.splitext(clean_path)[1].lower()
        if ext == ".csv":
            df = pd.read_csv(clean_path)
        elif ext in [".xlsx", ".xls"]:
            target_sheet = 0 if sheet_name is None else sheet_name
            df = pd.read_excel(clean_path, sheet_name=target_sheet)
            if isinstance(df, dict):
                df = list(df.values())[0]
        else:
            raise ValueError(f"Unsupported file format '{ext}'. Please provide a .csv or .xlsx file.")

        df.columns = [str(c).strip() for c in df.columns]
        return df

    @staticmethod
    def get_sheet_names(excel_path: str) -> List[str]:
        clean_path = str(excel_path).strip().strip("'\"")
        if clean_path.lower().endswith((".xlsx", ".xls")):
            try:
                xl = pd.ExcelFile(clean_path)
                return xl.sheet_names
            except Exception:
                return []
        return []


class FileHelper:
    @staticmethod
    def sanitize_filename(name: str) -> str:
        clean = re.sub(r'[\\/*?:"<>|]', "_", str(name)).strip()
        return clean or "document"

    @staticmethod
    def safe_delete(path: str, retries: int = 5, delay: float = 0.25) -> bool:
        for _ in range(retries):
            try:
                if os.path.exists(path):
                    os.remove(path)
                return True
            except Exception:
                time.sleep(delay)
        return not (path and os.path.exists(path))

    @staticmethod
    def get_export_dir(template_path: str, subfolder: str = "output") -> str:
        parent = os.path.dirname(os.path.abspath(template_path)) if template_path else os.getcwd()
        export_dir = os.path.join(parent, subfolder)
        os.makedirs(export_dir, exist_ok=True)
        return export_dir


# ------------------------------------------------------------------------------
# 5. Interactive GUI & Console Dynamic Pickers
# ------------------------------------------------------------------------------
class DynamicPicker:
    @staticmethod
    def select_file(title: str = "Select File", filetypes: Optional[List[Tuple[str, str]]] = None) -> Optional[str]:
        try:
            import tkinter as tk
            from tkinter import filedialog
            root = tk.Tk()
            root.withdraw()
            root.attributes("-topmost", True)
            selected = filedialog.askopenfilename(title=title, filetypes=filetypes or [("All Files", "*.*")])
            root.destroy()
            if selected:
                return os.path.abspath(selected)
        except Exception:
            pass

        val = input(f"\n{title} [Path]: ").strip().strip("'\"")
        return os.path.abspath(val) if val else None

    @staticmethod
    def select_directory(title: str = "Select Folder") -> Optional[str]:
        try:
            import tkinter as tk
            from tkinter import filedialog
            root = tk.Tk()
            root.withdraw()
            root.attributes("-topmost", True)
            selected = filedialog.askdirectory(title=title)
            root.destroy()
            if selected:
                return os.path.abspath(selected)
        except Exception:
            pass

        val = input(f"\n{title} [Folder Path]: ").strip().strip("'\"")
        return os.path.abspath(val) if val else None


# ==============================================================================
# AutoCrat Pro Advanced Multi-Asset & QR/Image Styling Engine
# Created by mohatamim
# ==============================================================================
import os
import re
import json
import time
import urllib.request
import tempfile
import threading
from typing import List, Dict, Optional, Tuple, Any
import pandas as pd
from PIL import Image, ImageDraw, ImageColor, ImageOps, ImageTk
import tkinter as tk
from tkinter import ttk, colorchooser, filedialog, messagebox
from pptx import Presentation
from pptx.util import Inches, Pt
import qrcode
from qrcode.image.styledpil import StyledPilImage
from qrcode.image.styles.moduledrawers.pil import (
    SquareModuleDrawer, CircleModuleDrawer, RoundedModuleDrawer, GappedSquareModuleDrawer,
    VerticalBarsDrawer, HorizontalBarsDrawer
)
from qrcode.image.styles.colormasks import (
    SolidFillColorMask, RadialGradiantColorMask,
    HorizontalGradiantColorMask, VerticalGradiantColorMask
)
import zxingcpp

# ------------------------------------------------------------------------------
# ------------------------------------------------------------------------------
# 1. Row Selection Parser
# ------------------------------------------------------------------------------
def parse_row_selection(range_str: str, total_rows: int) -> List[int]:
    """
    Parses user-specified 1-based row numbers or ranges (e.g. '1-5, 8, 10-12')
    into a sorted list of zero-based DataFrame indices.
    """
    if not range_str or not range_str.strip() or range_str.strip().lower() == "all":
        return list(range(total_rows))
    
    selected_indices = set()
    parts = [p.strip() for p in range_str.replace(";", ",").split(",") if p.strip()]
    for part in parts:
        if "-" in part:
            sub = part.split("-")
            if len(sub) == 2:
                s_str, e_str = sub[0].strip(), sub[1].strip()
                if s_str and not e_str:
                    try:
                        start_1 = int(s_str)
                        for r in range(max(1, start_1), total_rows + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
                elif not s_str and e_str:
                    try:
                        end_1 = int(e_str)
                        for r in range(1, min(total_rows, end_1) + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
                elif s_str and e_str:
                    try:
                        start_1, end_1 = int(s_str), int(e_str)
                        if start_1 > end_1: start_1, end_1 = end_1, start_1
                        for r in range(max(1, start_1), min(total_rows, end_1) + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
        else:
            try:
                r_1 = int(part)
                if 1 <= r_1 <= total_rows:
                    selected_indices.add(r_1 - 1)
            except ValueError: pass
            
    res = sorted(list(selected_indices))
    return res if res else list(range(total_rows))


# ------------------------------------------------------------------------------
# 2. Advanced Image & Photo Element Configuration
# ------------------------------------------------------------------------------
class ImageElementConfig:
    """
    Configurable image / photo placeholder definition:
    - Independent styling and placement per placeholder (e.g. {{photo}}, {{signature}}, {{logo}}).
    - Shape: 'rect', 'square', 'rounded', 'circle'.
    - Fitting: 'fill' (crop), 'contain' (fit box), 'fit', 'stretch'.
    - Full border stroke styling, corner radius, grayscale, opacity, and rotation.
    """
    def __init__(
        self,
        tag: str = "{{photo}}",
        enabled: bool = True,
        source_mode: str = "folder",   # 'folder', 'column', 'static'
        data_column: Optional[str] = None,
        image_folder: Optional[str] = None,
        id_column: Optional[str] = None,
        static_path: Optional[str] = None,
        shape_mode: str = "rounded",   # 'rect', 'square', 'rounded', 'circle'
        corner_radius: int = 25,
        border_width: int = 3,
        border_color: str = "#1E3A8A",
        fitting_mode: str = "fill",     # 'fill', 'contain', 'fit', 'stretch'
        alignment: str = "center",      # 'center', 'top', 'bottom', 'left', 'right'
        opacity: float = 1.0,           # 0.1 to 1.0
        grayscale: bool = False,
        rotation: int = 0,              # 0, 90, 180, 270
        position_mode: str = "placeholder", # 'placeholder', 'custom'
        custom_left_in: float = 1.0,
        custom_top_in: float = 1.0,
        custom_width_in: float = 2.0,
        custom_height_in: float = 2.0,
        **kwargs
    ):
        raw_tag = tag.strip()
        self.tag = raw_tag if raw_tag.startswith("{{") else f"{{{{{raw_tag}}}}}"
        self.enabled = bool(enabled)
        self.source_mode = str(source_mode).lower()
        self.data_column = data_column
        self.image_folder = image_folder
        self.id_column = id_column
        self.static_path = static_path
        self.static_image_path = static_path
        self.shape_mode = str(shape_mode).lower()
        self.corner_radius = int(corner_radius)
        self.border_width = int(border_width)
        self.border_color = str(border_color).strip()
        self.fitting_mode = str(fitting_mode).lower()
        self.alignment = str(alignment).lower()
        self.opacity = max(0.05, min(1.0, float(opacity)))
        self.grayscale = bool(grayscale)
        self.rotation = int(rotation)
        self.position_mode = str(position_mode).lower()
        self.custom_left_in = float(custom_left_in)
        self.custom_top_in = float(custom_top_in)
        self.custom_width_in = float(custom_width_in)
        self.custom_height_in = float(custom_height_in)
        # Compatibility properties
        self.custom_left_inches = self.custom_left_in
        self.custom_top_inches = self.custom_top_in
        self.custom_size_inches = self.custom_width_in

    def to_dict(self) -> Dict[str, Any]:
        return {
            "tag": self.tag,
            "enabled": self.enabled,
            "source_mode": self.source_mode,
            "data_column": self.data_column,
            "image_folder": self.image_folder,
            "id_column": self.id_column,
            "static_path": self.static_path,
            "static_image_path": self.static_image_path,
            "shape_mode": self.shape_mode,
            "corner_radius": self.corner_radius,
            "border_width": self.border_width,
            "border_color": self.border_color,
            "fitting_mode": self.fitting_mode,
            "alignment": self.alignment,
            "opacity": self.opacity,
            "grayscale": self.grayscale,
            "rotation": self.rotation,
            "position_mode": self.position_mode,
            "custom_left_in": self.custom_left_in,
            "custom_top_in": self.custom_top_in,
            "custom_width_in": self.custom_width_in,
            "custom_height_in": self.custom_height_in
        }

# Alias for backward compatibility
ImageConfig = ImageElementConfig


# ------------------------------------------------------------------------------
# 3. Advanced QR Code Element Configuration
# ------------------------------------------------------------------------------
class QRElementConfig:
    """
    Configurable dynamic QR code placeholder definition:
    - Multiple QR codes with different data and styling on the same slide.
    - Payload Modes:
        1. Single Column
        2. URL + Specific Column Template
        3. Static URL / Info across all PDFs
        4. Multiple Columns (Name: Value formatted multiline text)
    - Appearance:
        - Module Styles: 'square', 'rounded', 'circle', 'gapped'
        - Preset Themes: 'classic_bw', 'corporate_blue', 'dark_slate', 'premium_gold', or Custom HEX
        - Center Branding Logo with auto-sized white backing box
        - Validation with zxing-cpp to guarantee scan reliability
    """
    def __init__(
        self,
        tag: str = "{{qr}}",
        enabled: bool = True,
        payload_mode: str = "column",  # 'column', 'template', 'static', 'multi_column'
        data_column: Optional[str] = None,
        data_template: Optional[str] = None,
        static_payload: Optional[str] = None,
        multi_columns: Optional[List[str]] = None,
        module_drawer: str = "rounded", # 'square', 'rounded', 'circle', 'gapped'
        fg_color: str = "#0F172A",
        bg_color: str = "#FFFFFF",
        preset_theme: str = "corporate_blue", # 'classic_bw', 'corporate_blue', 'dark_slate', 'premium_gold'
        error_correction: str = "H",   # 'L', 'M', 'Q', 'H'
        box_size: int = 10,
        border: int = 2,
        logo_path: Optional[str] = None,
        logo_size_ratio: float = 0.22,
        validate_decoding: bool = True,
        position_mode: str = "placeholder", # 'placeholder', 'custom'
        custom_left_in: float = 1.0,
        custom_top_in: float = 1.0,
        custom_size_in: float = 1.6,
        margin_inches: float = 0.5,
        **kwargs
    ):
        raw_tag = tag.strip()
        self.tag = raw_tag if raw_tag.startswith("{{") else f"{{{{{raw_tag}}}}}"
        self.enabled = bool(enabled)
        self.payload_mode = str(payload_mode).lower()
        self.data_column = data_column
        self.data_template = data_template
        self.static_payload = str(static_payload or "")
        self.multi_columns = list(multi_columns or [])
        self.module_drawer = str(module_drawer).lower()
        self.fg_color = str(fg_color).strip()
        self.bg_color = str(bg_color).strip()
        self.preset_theme = str(preset_theme).lower()
        self.error_correction = str(error_correction).upper()
        self.box_size = int(box_size)
        self.border = int(border)
        self.logo_path = logo_path
        self.logo_size_ratio = float(logo_size_ratio)
        self.validate_decoding = bool(validate_decoding)
        self.position_mode = str(position_mode).lower()
        self.custom_left_in = float(custom_left_in)
        self.custom_top_in = float(custom_top_in)
        self.custom_size_in = float(custom_size_in)
        self.margin_inches = float(margin_inches)
        # Compatibility properties
        self.custom_left_inches = self.custom_left_in
        self.custom_top_inches = self.custom_top_in
        self.custom_size_inches = self.custom_size_in

    def resolve_payload(self, row_map: Dict[str, str], raw_row: pd.Series) -> str:
        if not self.enabled:
            return ""
        
        mode = getattr(self, "payload_mode", "column")
        if mode == "static" or (getattr(self, "static_payload", None) and not self.data_column and not self.data_template and not getattr(self, "multi_columns", None)):
            return str(getattr(self, "static_payload", "")).strip()

        if mode == "multi_column" or (getattr(self, "multi_columns", None) and len(self.multi_columns) > 0):
            lines = []
            for col in (getattr(self, "multi_columns", []) or []):
                if col in raw_row:
                    v = raw_row[col]
                    v_str = "" if pd.isna(v) else str(v).strip()
                    lines.append(f"{col}: {v_str}")
                elif col in row_map:
                    lines.append(f"{col}: {row_map[col]}")
            if lines:
                return "\n".join(lines)

        if self.data_template:
            return PlaceholderEngine.replace_text(self.data_template, row_map).strip()

        if self.data_column and self.data_column in raw_row:
            val = raw_row[self.data_column]
            return "" if pd.isna(val) else str(val).strip()
        
        for key in ["serial", "id", "roll", "registration_id", "cert_id", "url", "qr"]:
            for col, val in row_map.items():
                if PlaceholderEngine.normalize_key(col) == key and val:
                    return str(val).strip()
        return ""

    def to_dict(self) -> Dict[str, Any]:
        return {
            "tag": self.tag,
            "enabled": self.enabled,
            "payload_mode": self.payload_mode,
            "data_column": self.data_column,
            "data_template": self.data_template,
            "static_payload": self.static_payload,
            "multi_columns": self.multi_columns,
            "module_drawer": self.module_drawer,
            "fg_color": self.fg_color,
            "bg_color": self.bg_color,
            "preset_theme": self.preset_theme,
            "error_correction": self.error_correction,
            "box_size": self.box_size,
            "border": self.border,
            "logo_path": self.logo_path,
            "logo_size_ratio": self.logo_size_ratio,
            "validate_decoding": self.validate_decoding,
            "position_mode": self.position_mode,
            "custom_left_in": self.custom_left_in,
            "custom_top_in": self.custom_top_in,
            "custom_size_in": self.custom_size_in
        }

# Alias for backward compatibility
QRCodeConfig = QRElementConfig


# ------------------------------------------------------------------------------
# 4. Advanced Image Processing Engine
# ------------------------------------------------------------------------------
class AdvancedImageProcessor:
    @staticmethod
    def resolve_image_path(cfg: ImageElementConfig, row: pd.Series, row_map: Dict[str, str], temp_dir: str) -> Tuple[Optional[str], bool]:
        """
        Resolves the image file path based on folder ID matching, column URL/file, or static asset.
        Supports automatic Google Drive link conversion and local file scanning.
        """
        if not cfg.enabled:
            return None, False

        # 1. Static asset (e.g. institutional logo, seal, or signature)
        static_p = cfg.static_path or getattr(cfg, "static_image_path", None)
        if static_p and os.path.isfile(static_p):
            return os.path.abspath(static_p), False

        # 2. Check folder matching
        if cfg.image_folder and os.path.isdir(cfg.image_folder):
            id_val = ""
            if cfg.id_column and cfg.id_column in row:
                id_val = str(row[cfg.id_column]).strip()
            if not id_val:
                for c in ["serial", "id", "roll", "reg_no", "certificate_id"]:
                    if c in row_map and row_map[c]:
                        id_val = row_map[c]
                        break

            if id_val:
                for ext in ["", ".jpg", ".jpeg", ".png", ".webp", ".bmp", ".JPG", ".PNG", ".JPEG"]:
                    cand = os.path.join(cfg.image_folder, f"{id_val}{ext}")
                    if os.path.isfile(cand):
                        return cand, False
                # Case-insensitive directory scan
                try:
                    for f_name in os.listdir(cfg.image_folder):
                        base, ext = os.path.splitext(f_name)
                        if base.strip().lower() == id_val.lower() and ext.lower() in [".jpg", ".jpeg", ".png", ".webp", ".bmp"]:
                            return os.path.join(cfg.image_folder, f_name), False
                except Exception:
                    pass

        # 3. Check column value (path or web URL)
        target_val = None
        if cfg.data_column and cfg.data_column in row:
            v = row[cfg.data_column]
            if pd.notna(v) and str(v).strip():
                target_val = str(v).strip()

        if not target_val:
            for c, v in row_map.items():
                if PlaceholderEngine.normalize_key(c) in ["photo", "image", "picture", "avatar", "photourl", "imageurl", "upload your photos"] and v:
                    target_val = str(v).strip()
                    break

        if target_val:
            # Check if web URL
            if target_val.startswith(("http://", "https://")):
                try:
                    dl_url = target_val
                    # Convert Google Drive view URL to direct export download
                    if "drive.google.com" in target_val:
                        gd_match = re.search(r"id=([a-zA-Z0-9_-]+)", target_val) or re.search(r"/d/([a-zA-Z0-9_-]+)", target_val)
                        if gd_match:
                            dl_url = f"https://drive.google.com/uc?export=download&id={gd_match.group(1)}"

                    ext = os.path.splitext(dl_url.split("?")[0])[1] or ".jpg"
                    dl_path = os.path.join(temp_dir, f"_temp_img_{int(time.time() * 1000)}{ext}")
                    urllib.request.urlretrieve(dl_url, dl_path)
                    return dl_path, True
                except Exception as e:
                    print(f"    [WARN] Failed to download image from {target_val}: {e}")
                    return None, False

            # Local file path
            if os.path.isfile(target_val):
                return os.path.abspath(target_val), False

        return None, False

    @staticmethod
    def process_image(src_path: str, target_w: int, target_h: int, cfg: ImageElementConfig, dest_path: str) -> str:
        """
        Applies image preprocessing: fitting mode (fill/contain/fit/stretch), alignment,
        shape (rect, square, circle, rounded), corner radius, outside border stroke,
        grayscale, opacity, and rotation.
        """
        if not os.path.isfile(src_path):
            return src_path

        try:
            with Image.open(src_path) as orig:
                im = ImageOps.exif_transpose(orig).convert("RGBA")

            # 1. Rotation
            if cfg.rotation in [90, 180, 270]:
                im = im.rotate(-cfg.rotation, expand=True)

            # 2. Grayscale
            if cfg.grayscale:
                alpha = im.split()[-1]
                im_gray = ImageOps.grayscale(im.convert("RGB")).convert("RGBA")
                im_gray.putalpha(alpha)
                im = im_gray

            # 3. Fitting
            tw, th = max(10, target_w), max(10, target_h)
            fit_mode = cfg.fitting_mode
            if fit_mode == "stretch":
                im_fitted = im.resize((tw, th), Image.Resampling.LANCZOS)
            elif fit_mode in ["contain", "fit"]:
                im_fitted = ImageOps.contain(im, (tw, th), Image.Resampling.LANCZOS)
                canvas = Image.new("RGBA", (tw, th), (0, 0, 0, 0))
                px = (tw - im_fitted.width) // 2
                py = (th - im_fitted.height) // 2
                canvas.paste(im_fitted, (px, py), mask=im_fitted)
                im_fitted = canvas
            else: # 'fill' (crop to fill target aspect ratio)
                aspect_im = im.width / im.height
                aspect_t = tw / th
                if aspect_im > aspect_t:
                    crop_w = int(im.height * aspect_t)
                    if cfg.alignment == "left":
                        x0 = 0
                    elif cfg.alignment == "right":
                        x0 = im.width - crop_w
                    else: # center
                        x0 = (im.width - crop_w) // 2
                    box = (x0, 0, x0 + crop_w, im.height)
                else:
                    crop_h = int(im.width / aspect_t)
                    if cfg.alignment == "top":
                        y0 = 0
                    elif cfg.alignment == "bottom":
                        y0 = im.height - crop_h
                    else: # center
                        y0 = (im.height - crop_h) // 2
                    box = (0, y0, im.width, y0 + crop_h)
                im_cropped = im.crop(box)
                im_fitted = im_cropped.resize((tw, th), Image.Resampling.LANCZOS)

            # 4. Shape & Outside Border Stroke
            bw = max(0, int(cfg.border_width))
            scale = 2
            w_s, h_s = tw * scale, th * scale
            bw_s = bw * scale
            im_s = im_fitted.resize((w_s, h_s), Image.Resampling.LANCZOS)

            mask = Image.new("L", (w_s, h_s), 0)
            d_mask = ImageDraw.Draw(mask)

            tot_w_s = w_s + 2 * bw_s
            tot_h_s = h_s + 2 * bw_s
            out = Image.new("RGBA", (tot_w_s, tot_h_s), (0, 0, 0, 0))

            try:
                c_rgb = ImageColor.getrgb(cfg.border_color)
            except Exception:
                c_rgb = (30, 58, 138)
            c_rgba = (c_rgb[0], c_rgb[1], c_rgb[2], 255)

            if cfg.shape_mode == "circle":
                if bw_s > 0:
                    d_out = ImageDraw.Draw(out)
                    d_out.ellipse((0, 0, tot_w_s - 1, tot_h_s - 1), fill=c_rgba)
                d_mask.ellipse((0, 0, w_s - 1, h_s - 1), fill=255)
            elif cfg.shape_mode == "rounded":
                r = min(int(cfg.corner_radius * scale), min(w_s, h_s) // 2)
                if bw_s > 0:
                    d_out = ImageDraw.Draw(out)
                    d_out.rounded_rectangle((0, 0, tot_w_s - 1, tot_h_s - 1), radius=r + bw_s, fill=c_rgba)
                d_mask.rounded_rectangle((0, 0, w_s - 1, h_s - 1), radius=r, fill=255)
            else: # square or rect
                if bw_s > 0:
                    d_out = ImageDraw.Draw(out)
                    d_out.rectangle((0, 0, tot_w_s - 1, tot_h_s - 1), fill=c_rgba)
                d_mask.rectangle((0, 0, w_s - 1, h_s - 1), fill=255)

            out.paste(im_s, (bw_s, bw_s), mask=mask)

            # 5. Opacity
            if cfg.opacity < 1.0:
                alpha = out.split()[-1]
                alpha = alpha.point(lambda p: int(p * max(0.1, cfg.opacity)))
                out.putalpha(alpha)

            final_img = out.resize((tw + 2 * bw, th + 2 * bw), Image.Resampling.LANCZOS)
            final_img.save(dest_path, "PNG")
            return dest_path
        except Exception as e:
            print(f"    [WARN] Image styling failed: {e}")
            return src_path


# ------------------------------------------------------------------------------
# 5. Advanced QR Code Generation Engine
# ------------------------------------------------------------------------------
class AdvancedQRProcessor:
    @staticmethod
    def generate_qr(payload: str, cfg: QRElementConfig, out_path: str, target_size_px: int = 400) -> bool:
        """
        Generates customizable QR code: styled modules (square, rounded, circle, gapped),
        preset themes or custom HEX colors, center branding logo with white backing box,
        and automatic decoding validation with zxing-cpp to guarantee readability.
        """
        if not payload:
            return False

        try:
            ec_map = {
                "L": qrcode.constants.ERROR_CORRECT_L,
                "M": qrcode.constants.ERROR_CORRECT_M,
                "Q": qrcode.constants.ERROR_CORRECT_Q,
                "H": qrcode.constants.ERROR_CORRECT_H,
            }
            ec = ec_map.get(cfg.error_correction.upper(), qrcode.constants.ERROR_CORRECT_H)
            if cfg.logo_path and os.path.isfile(cfg.logo_path):
                ec = qrcode.constants.ERROR_CORRECT_H

            qr = qrcode.QRCode(
                version=None,
                error_correction=ec,
                box_size=max(2, cfg.box_size),
                border=max(1, cfg.border)
            )
            qr.add_data(payload)
            qr.make(fit=True)

            drawers = {
                "rounded": RoundedModuleDrawer(),
                "circle": CircleModuleDrawer(),
                "square": SquareModuleDrawer(),
                "gapped": GappedSquareModuleDrawer(),
                "vertical_bars": VerticalBarsDrawer(),
                "horizontal_bars": HorizontalBarsDrawer()
            }
            drawer = drawers.get(cfg.module_drawer, RoundedModuleDrawer())

            # Themes and Color Masks
            th = cfg.preset_theme
            if th == "classic_bw":
                mask = SolidFillColorMask(back_color=(255, 255, 255), front_color=(0, 0, 0))
            elif th == "corporate_blue":
                mask = SolidFillColorMask(back_color=(255, 255, 255), front_color=(30, 58, 138))
            elif th == "dark_slate":
                mask = SolidFillColorMask(back_color=(248, 250, 252), front_color=(15, 23, 42))
            elif th == "premium_gold":
                mask = SolidFillColorMask(back_color=(255, 251, 235), front_color=(180, 83, 9))
            elif th == "emerald_green":
                mask = SolidFillColorMask(back_color=(255, 255, 255), front_color=(5, 150, 105))
            elif th == "crimson_red":
                mask = SolidFillColorMask(back_color=(255, 255, 255), front_color=(185, 28, 28))
            elif th == "royal_purple":
                mask = SolidFillColorMask(back_color=(255, 255, 255), front_color=(109, 40, 217))
            elif th == "cyber_cyan":
                mask = SolidFillColorMask(back_color=(11, 15, 25), front_color=(2, 132, 199))
            elif th == "gradient_sunset":
                mask = RadialGradiantColorMask(back_color=(255, 255, 255), center_color=(234, 88, 12), edge_color=(159, 18, 57))
            elif th == "gradient_ocean":
                mask = HorizontalGradiantColorMask(back_color=(255, 255, 255), left_color=(15, 23, 42), right_color=(13, 148, 136))
            elif th == "gradient_neon":
                mask = VerticalGradiantColorMask(back_color=(255, 255, 255), top_color=(124, 58, 237), bottom_color=(6, 182, 212))
            else: # custom
                try: fg_rgb = ImageColor.getrgb(cfg.fg_color)
                except Exception: fg_rgb = (0, 0, 0)
                try: bg_rgb = ImageColor.getrgb(cfg.bg_color)
                except Exception: bg_rgb = (255, 255, 255)
                mask = SolidFillColorMask(back_color=bg_rgb, front_color=fg_rgb)

            img = qr.make_image(
                image_factory=StyledPilImage,
                module_drawer=drawer,
                color_mask=mask
            )

            pil_im = img.get_image() if hasattr(img, "get_image") else img
            if hasattr(pil_im, "_img"): pil_im = pil_im._img
            pil_im = pil_im.convert("RGBA")

            # Center Logo Branding
            clean_logo = str(cfg.logo_path).strip().strip("'\"") if cfg.logo_path else ""
            if clean_logo and os.path.isfile(clean_logo):
                try:
                    with Image.open(clean_logo) as logo_orig:
                        logo = logo_orig.convert("RGBA")
                    ratio = min(0.25, max(0.12, cfg.logo_size_ratio))
                    l_sz = int(pil_im.width * ratio)
                    logo_fit = logo.resize((l_sz, l_sz), Image.Resampling.LANCZOS)
                    pad = max(2, int(l_sz * 0.12))
                    pad_box = Image.new("RGBA", (l_sz + 2 * pad, l_sz + 2 * pad), (255, 255, 255, 255))
                    pos_x = (pil_im.width - (l_sz + 2 * pad)) // 2
                    pos_y = (pil_im.height - (l_sz + 2 * pad)) // 2
                    pil_im.paste(pad_box, (pos_x, pos_y))
                    pil_im.paste(logo_fit, (pos_x + pad, pos_y + pad), mask=logo_fit)
                except Exception as e:
                    print(f"    [WARN] Failed to embed QR center logo: {e}")

            # Validation using zxingcpp
            if cfg.validate_decoding and zxingcpp is not None:
                try:
                    res = zxingcpp.read_barcode(pil_im)
                    if not res or res.text != payload:
                        print("    [WARN] Styled QR code failed decoding verification. Retrying with high-contrast format and logo preservation.")
                        qr_fallback = qrcode.QRCode(version=None, error_correction=qrcode.constants.ERROR_CORRECT_H, box_size=max(2, cfg.box_size), border=max(1, cfg.border))
                        qr_fallback.add_data(payload)
                        qr_fallback.make(fit=True)
                        pil_im_fallback = qr_fallback.make_image(fill_color=fg_rgb, back_color=bg_rgb).convert("RGBA")
                        if clean_logo and os.path.isfile(clean_logo):
                            try:
                                with Image.open(clean_logo) as logo_orig:
                                    logo = logo_orig.convert("RGBA")
                                ratio = min(0.22, max(0.12, cfg.logo_size_ratio))
                                l_sz = int(pil_im_fallback.width * ratio)
                                logo_fit = logo.resize((l_sz, l_sz), Image.Resampling.LANCZOS)
                                pad = max(2, int(l_sz * 0.12))
                                pad_box = Image.new("RGBA", (l_sz + 2 * pad, l_sz + 2 * pad), (255, 255, 255, 255))
                                pos_x = (pil_im_fallback.width - (l_sz + 2 * pad)) // 2
                                pos_y = (pil_im_fallback.height - (l_sz + 2 * pad)) // 2
                                pil_im_fallback.paste(pad_box, (pos_x, pos_y))
                                pil_im_fallback.paste(logo_fit, (pos_x + pad, pos_y + pad), mask=logo_fit)
                            except Exception:
                                pass
                        pil_im = pil_im_fallback
                except Exception:
                    pass

            pil_im.save(out_path, "PNG")
            return True
        except Exception as e:
            print(f"    [WARN] QR generation error: {e}")
            return False


# ------------------------------------------------------------------------------
# 6. Multi-Asset Slide Engine
# ------------------------------------------------------------------------------
class MultiAssetEngine:
    """
    Manages multiple images and multiple QR codes on the same slide:
    - Independent detection and mapping per placeholder tag.
    - Preserves slide layout, shape dimensions, and unrelated formatting.
    - Never overwrites or conflicts between multiple elements.
    """
    @staticmethod
    def scan_placeholders(prs: Optional[Presentation]) -> Dict[str, List[str]]:
        """
        Scans presentation shapes and discovers all image and QR placeholder tags.
        """
        found = {"images": [], "qrs": [], "text": []}
        if prs is None:
            return found

        tag_set = set()
        for slide in prs.slides:
            for shape in slide.shapes:
                if shape.has_text_frame and shape.text_frame.text:
                    matches = re.findall(r"\{\{([^{}]+)\}\}", shape.text_frame.text)
                    for m in matches:
                        clean = f"{{{{{m.strip()}}}}}"
                        tag_set.add(clean)

        for tag in tag_set:
            norm = tag.lower().replace("{", "").replace("}", "").strip()
            if "qr" in norm:
                found["qrs"].append(tag)
            elif any(k in norm for k in ["photo", "image", "picture", "avatar", "signature", "sign", "logo", "stamp", "seal"]):
                found["images"].append(tag)
            else:
                found["text"].append(tag)

        return found

    @staticmethod
    def embed_all_assets(slide, prs: Presentation, row: pd.Series, row_map: Dict[str, str],
                          img_config: Any, qr_config: Any, export_dir: str):
        """
        Embeds all configured images and QR codes onto the given slide independently.
        """
        # 1. Process Images
        img_dict = {}
        if isinstance(img_config, dict):
            if "multi_images" in img_config:
                img_dict = img_config["multi_images"]
            elif img_config.get("enabled"):
                tag = img_config.get("tag") or "{{photo}}"
                img_dict[tag] = ImageElementConfig(**img_config)
        elif hasattr(img_config, "multi_images"):
            img_dict = getattr(img_config, "multi_images")
        elif isinstance(img_config, ImageElementConfig) and img_config.enabled:
            img_dict[img_config.tag] = img_config

        for tag, cfg in img_dict.items():
            if not cfg.enabled:
                continue

            # Find matching shape on slide
            target_shape = None
            tag_clean = tag.lower().replace("{", "").replace("}", "").strip()
            for shape in slide.shapes:
                if shape.has_text_frame and re.search(rf"\{{\{{\s*{re.escape(tag_clean)}\s*\}}\}}", shape.text_frame.text, re.IGNORECASE):
                    target_shape = shape
                    break

            # Resolve image path
            img_path, is_temp = AdvancedImageProcessor.resolve_image_path(cfg, row, row_map, export_dir)
            if not img_path or not os.path.isfile(img_path):
                continue

            if target_shape is not None and cfg.position_mode in ["placeholder", "auto"]:
                left, top, w, h = target_shape.left, target_shape.top, target_shape.width, target_shape.height
                tw = max(50, int((w / 914400.0) * 120))
                th = max(50, int((h / 914400.0) * 120))
                proc_img = os.path.join(export_dir, f"_proc_{tag_clean}_{int(time.time()*1000)}.png")
                AdvancedImageProcessor.process_image(img_path, tw, th, cfg, proc_img)
                slide.shapes.add_picture(proc_img, left, top, w, h)
                try:
                    elem = target_shape._element
                    elem.getparent().remove(elem)
                except Exception:
                    try: target_shape.text_frame.text = ""
                    except Exception: pass
                FileHelper.safe_delete(proc_img)
            else:
                # Custom coordinates (handles newly added images or dragged boxes)
                left = int(cfg.custom_left_in * 914400)
                top = int(cfg.custom_top_in * 914400)
                w = int(cfg.custom_width_in * 914400)
                h = int(cfg.custom_height_in * 914400)
                tw = max(50, int(cfg.custom_width_in * 120))
                th = max(50, int(cfg.custom_height_in * 120))
                proc_img = os.path.join(export_dir, f"_proc_{tag_clean}_{int(time.time()*1000)}.png")
                AdvancedImageProcessor.process_image(img_path, tw, th, cfg, proc_img)
                slide.shapes.add_picture(proc_img, left, top, w, h)
                if target_shape is not None:
                    try:
                        elem = target_shape._element
                        elem.getparent().remove(elem)
                    except Exception:
                        try: target_shape.text_frame.text = ""
                        except Exception: pass
                FileHelper.safe_delete(proc_img)

            if is_temp and img_path:
                FileHelper.safe_delete(img_path)

        # 2. Process QR Codes
        qr_dict = {}
        if isinstance(qr_config, dict):
            if "multi_qrs" in qr_config:
                qr_dict = qr_config["multi_qrs"]
            elif qr_config.get("enabled"):
                tag = qr_config.get("tag") or "{{qr}}"
                qr_dict[tag] = QRElementConfig(**qr_config)
        elif hasattr(qr_config, "multi_qrs"):
            qr_dict = getattr(qr_config, "multi_qrs")
        elif isinstance(qr_config, QRElementConfig) and qr_config.enabled:
            qr_dict[qr_config.tag] = qr_config

        for tag, cfg in qr_dict.items():
            if not cfg.enabled:
                continue

            target_shape = None
            tag_clean = tag.lower().replace("{", "").replace("}", "").strip()
            for shape in slide.shapes:
                if shape.has_text_frame and re.search(rf"\{{\{{\s*{re.escape(tag_clean)}\s*\}}\}}", shape.text_frame.text, re.IGNORECASE):
                    target_shape = shape
                    break

            payload = cfg.resolve_payload(row_map, row)
            if not payload:
                continue

            temp_qr = os.path.join(export_dir, f"_qr_{tag_clean}_{int(time.time()*1000)}.png")

            if target_shape is not None and cfg.position_mode in ["placeholder", "auto"]:
                left, top, w, h = target_shape.left, target_shape.top, target_shape.width, target_shape.height
                sz = min(w, h)
                sz_px = max(50, int((sz / 914400.0) * 120))
                if AdvancedQRProcessor.generate_qr(payload, cfg, temp_qr, sz_px):
                    slide.shapes.add_picture(temp_qr, left, top, sz, sz)
                    try:
                        elem = target_shape._element
                        elem.getparent().remove(elem)
                    except Exception:
                        try: target_shape.text_frame.text = ""
                        except Exception: pass
                FileHelper.safe_delete(temp_qr)
            else:
                # Custom coordinates (handles newly added QR codes or dragged positions)
                left = int(cfg.custom_left_in * 914400)
                top = int(cfg.custom_top_in * 914400)
                sz = int(cfg.custom_size_in * 914400)
                sz_px = max(50, int(cfg.custom_size_in * 120))
                if AdvancedQRProcessor.generate_qr(payload, cfg, temp_qr, sz_px):
                    slide.shapes.add_picture(temp_qr, left, top, sz, sz)
                if target_shape is not None:
                    try:
                        elem = target_shape._element
                        elem.getparent().remove(elem)
                    except Exception:
                        try: target_shape.text_frame.text = ""
                        except Exception: pass
                FileHelper.safe_delete(temp_qr)


# ------------------------------------------------------------------------------
# 7. Backward-Compatible Handlers
# ------------------------------------------------------------------------------
class QRCodeHandler:
    @staticmethod
    def generate_image(payload: str, out_path: str, box_size: int = 10, border: int = 1) -> bool:
        dummy_cfg = QRElementConfig(box_size=box_size, border=border)
        return AdvancedQRProcessor.generate_qr(payload, dummy_cfg, out_path)

    @classmethod
    def embed_in_slide(cls, slide, prs: Presentation, qr_path: str, config: Any):
        if not config:
            return
        dummy_row = pd.Series({})
        MultiAssetEngine.embed_all_assets(slide, prs, dummy_row, {}, None, config, os.path.dirname(qr_path) or ".")

class ImageHandler:
    @staticmethod
    def resolve_image_path(config: Any, row: pd.Series, row_map: Dict[str, str], temp_dir: str) -> Tuple[Optional[str], bool]:
        if not config:
            return None, False
        if not isinstance(config, ImageElementConfig):
            config = ImageElementConfig(**(config if isinstance(config, dict) else config.__dict__))
        return AdvancedImageProcessor.resolve_image_path(config, row, row_map, temp_dir)

    @staticmethod
    def apply_styling(src_path: str, config: Any, dest_path: str) -> str:
        if not isinstance(config, ImageElementConfig):
            config = ImageElementConfig(**(config if isinstance(config, dict) else config.__dict__))
        return AdvancedImageProcessor.process_image(src_path, 400, 400, config, dest_path)

    @classmethod
    def embed_in_slide(cls, slide, prs: Presentation, image_path: str, config: Any):
        if not config:
            return
        dummy_row = pd.Series({})
        MultiAssetEngine.embed_all_assets(slide, prs, dummy_row, {}, config, None, os.path.dirname(image_path) or ".")


# ------------------------------------------------------------------------------

# ------------------------------------------------------------------------------

# ------------------------------------------------------------------------------
# 8. Interactive Multi-Asset Visual Canvas Designer
# ------------------------------------------------------------------------------
# ------------------------------------------------------------------------------
# 8. Interactive Multi-Asset Visual Canvas Designer
# ------------------------------------------------------------------------------
# ------------------------------------------------------------------------------
# 8. Interactive Multi-Asset Visual Canvas Designer
# ------------------------------------------------------------------------------
class VisualDragDropCanvas:
    """
    Interactive Multi-Asset Visual Drag-and-Drop Layout Designer:
    - Multiple images (photos, logos, signatures) and multiple QR codes on the same slide.
    - Actual PowerPoint template background rendering with persistent disk-caching and background pre-caching.
    - Mouse expandable / resizable cursors (size_nw_se, size_ne_sw, fleur) on hover and drag.
    - High-fidelity enterprise typography, professional inputs, labels, and action buttons.
    - Diverse QR code styles: rounded, circle, square, gapped, vertical/horizontal bars, gradients, and custom colors.
    - Sub-50ms instant modal opening via optimized disk preview caching.
    Created by mohatamim.
    """

    THEME_DISPLAY_MAP = {
        "Corporate Deep Navy": "corporate_blue",
        "Classic Monochrome (B&W)": "classic_bw",
        "Executive Slate": "dark_slate",
        "Royal Gold & Black": "premium_gold",
        "Emerald Forest Green": "emerald_green",
        "Crimson Ruby Red": "crimson_red",
        "Royal Violet Purple": "royal_purple",
        "Cyberpunk Neon Cyan": "cyber_cyan",
        "Sunset Glow (Radial)": "gradient_sunset",
        "Ocean Breeze (Linear)": "gradient_ocean",
        "Neon Horizon (Linear)": "gradient_neon",
        "Custom Scheme...": "custom"
    }
    THEME_KEY_MAP = {v: k for k, v in THEME_DISPLAY_MAP.items()}

    DRAWER_DISPLAY_MAP = {
        "Smooth Rounded": "rounded",
        "Circular Dots": "circle",
        "Classic Square": "square",
        "Gapped Grid": "gapped",
        "Vertical Bars": "vertical_bars",
        "Horizontal Scanlines": "horizontal_bars"
    }
    DRAWER_KEY_MAP = {v: k for k, v in DRAWER_DISPLAY_MAP.items()}

    FIT_DISPLAY_MAP = {
        "Cover / Crop (Fill)": "fill",
        "Contain (Fit Proportionally)": "contain",
        "Stretch to Boundary": "stretch"
    }
    FIT_KEY_MAP = {v: k for k, v in FIT_DISPLAY_MAP.items()}

    SHAPE_DISPLAY_MAP = {
        "Rectangle": "rect",
        "Square": "square",
        "Rounded Rectangle": "rounded",
        "Circular (Avatar)": "circle"
    }
    SHAPE_KEY_MAP = {v: k for k, v in SHAPE_DISPLAY_MAP.items()}

    @classmethod
    def get_cache_dir(cls) -> str:
        cache_d = os.path.join(tempfile.gettempdir(), "autocrat_preview_cache")
        os.makedirs(cache_d, exist_ok=True)
        return cache_d

    @classmethod
    def get_cached_preview_path(cls, pptx_path: str) -> Optional[str]:
        if not pptx_path or not os.path.isfile(pptx_path):
            return None
        try:
            abs_p = os.path.abspath(pptx_path)
            mtime = os.path.getmtime(abs_p)
            cache_key = hashlib.md5(f"{abs_p}_{mtime}".encode("utf-8")).hexdigest()
            cached_p = os.path.join(cls.get_cache_dir(), f"slide1_{cache_key}.png")
            if os.path.isfile(cached_p) and os.path.getsize(cached_p) > 100:
                return cached_p
        except Exception:
            pass
        return None

    @classmethod
    def precache_slide_preview(cls, pptx_path: str):
        """
        Background worker that exports Slide 1 to disk cache ahead of time.
        Enables instant (<30ms) opening when the user launches the Layout Designer.
        """
        clean_path = str(pptx_path).strip().strip("'\"") if pptx_path else ""
        if not clean_path or not os.path.isfile(clean_path):
            return

        cached = cls.get_cached_preview_path(clean_path)
        if cached and os.path.isfile(cached):
            return

        try:
            import pythoncom
            pythoncom.CoInitialize()
            import win32com.client

            abs_p = os.path.abspath(clean_path)
            mtime = os.path.getmtime(abs_p)
            cache_key = hashlib.md5(f"{abs_p}_{mtime}".encode("utf-8")).hexdigest()
            dest_p = os.path.join(cls.get_cache_dir(), f"slide1_{cache_key}.png")
            temp_export = os.path.join(cls.get_cache_dir(), f"_tmp_{cache_key}.png")

            ppt_app = win32com.client.Dispatch("PowerPoint.Application")
            pres = ppt_app.Presentations.Open(abs_p, ReadOnly=True, Untitled=False, WithWindow=False)
            pres.Slides(1).Export(temp_export, "PNG", 1300, 900)
            pres.Close()

            if os.path.isfile(temp_export) and os.path.getsize(temp_export) > 100:
                if os.path.isfile(dest_p):
                    try: os.remove(dest_p)
                    except Exception: pass
                os.replace(temp_export, dest_p)
        except Exception:
            pass
        finally:
            try:
                pythoncom.CoUninitialize()
            except Exception:
                pass

    def __init__(self, df: pd.DataFrame, prs: Optional[Presentation], pptx_path: Optional[str] = None, parent=None):
        self.df = df
        self.prs = prs
        self.pptx_path = pptx_path
        self.parent = parent
        self.columns = list(df.columns) if df is not None else []
        self.temp_dir = tempfile.mkdtemp(prefix="autocrat_canvas_")
        self.cancelled = True
        self.rendered_preview_refs = []

        if prs is not None:
            self.slide_w_in = prs.slide_width / 914400.0
            self.slide_h_in = prs.slide_height / 914400.0
        else:
            self.slide_w_in, self.slide_h_in = 13.33, 7.5

        # Visual Canvas Scaling
        self.max_canvas_w, self.max_canvas_h = 650, 450
        aspect = self.slide_w_in / self.slide_h_in
        if aspect >= (self.max_canvas_w / self.max_canvas_h):
            self.canvas_w = self.max_canvas_w
            self.canvas_h = max(240, int(self.canvas_w / aspect))
        else:
            self.canvas_h = self.max_canvas_h
            self.canvas_w = max(320, int(self.canvas_h * aspect))
        self.px_per_inch = self.canvas_w / self.slide_w_in

        # Detect elements on slide
        discovered = MultiAssetEngine.scan_placeholders(prs)
        self.image_configs: Dict[str, ImageElementConfig] = {}
        self.qr_configs: Dict[str, QRElementConfig] = {}

        # Default image elements
        img_tags = discovered["images"] if discovered["images"] else ["{{photo}}"]
        default_folder = r"D:\New folder (3)\images" if os.path.isdir(r"D:\New folder (3)\images") else ""
        for idx, tag in enumerate(img_tags):
            self.image_configs[tag] = ImageElementConfig(
                tag=tag,
                enabled=True,
                source_mode="folder" if default_folder else "column",
                image_folder=default_folder,
                id_column="Serial" if "Serial" in self.columns else (self.columns[0] if self.columns else ""),
                data_column=self.columns[0] if self.columns else "",
                shape_mode="rounded",
                corner_radius=25,
                border_width=3,
                border_color="#1E3A8A",
                position_mode="custom",
                custom_left_in=max(0.5, self.slide_w_in - 2.8 - (idx * 0.4)),
                custom_top_in=max(0.5, 0.8 + (idx * 0.4)),
                custom_width_in=1.8,
                custom_height_in=2.2
            )

        # Default QR elements
        qr_tags = discovered["qrs"] if discovered["qrs"] else ["{{qr}}"]
        for idx, tag in enumerate(qr_tags):
            self.qr_configs[tag] = QRElementConfig(
                tag=tag,
                enabled=True,
                payload_mode="template" if "verif" in tag.lower() else "column",
                data_column="Serial" if "Serial" in self.columns else (self.columns[0] if self.columns else ""),
                data_template=f"https://carnival2026.org/verify?id={{{{Serial}}}}",
                module_drawer="rounded",
                preset_theme="corporate_blue" if idx == 0 else "classic_bw",
                position_mode="custom",
                custom_left_in=max(0.5, self.slide_w_in - 2.4 - (idx * 2.0)),
                custom_top_in=max(0.5, self.slide_h_in - 2.4),
                custom_size_in=1.6
            )

        self.current_tab_type = "image" if img_tags else "qr"
        all_tags = self._get_all_tags()
        self.active_tag = all_tags[0] if all_tags else ""

        # Render background (cached instant load or COM export)
        self.bg_image_pil = self._render_slide_background()

        # Window setup
        self.root = tk.Toplevel(parent) if parent else tk.Tk()
        self.root.title("AutoCrat Pro -- Layout & Style Designer (mohatamim)")
        self.root.geometry("1180x680")
        self.root.minsize(1100, 640)
        self.root.attributes("-topmost", True)
        self.root.configure(bg="#0F172A")
        self.root.protocol("WM_DELETE_WINDOW", self._cancel)

        # Element dragging/resizing state
        self.active_drag_elem = None
        self.drag_action = None       # 'move' or 'resize'
        self.drag_handle = None       # 'se', 'nw', 'ne', 'sw'
        self.drag_start_x = 0
        self.drag_start_y = 0
        self.elem_start_x = 0
        self.elem_start_y = 0
        self.elem_start_w = 0
        self.elem_start_h = 0

        self._build_ui()
        self._load_active_element_ui()
        self._redraw_canvas()

    def _render_slide_background(self) -> Image.Image:
        """
        Loads the template slide preview from fast disk cache if available.
        Otherwise exports via PowerPoint COM and caches it for subsequent instant openings.
        Falls back smoothly to vector rendering if COM is not available.
        """
        clean_path = str(self.pptx_path).strip().strip("'\"") if self.pptx_path else ""
        if clean_path and os.path.isfile(clean_path):
            # Check instant disk cache first
            cached = self.get_cached_preview_path(clean_path)
            if cached and os.path.isfile(cached):
                try:
                    with Image.open(cached) as im:
                        return im.convert("RGB").resize((self.canvas_w, self.canvas_h), Image.Resampling.LANCZOS)
                except Exception:
                    pass

            # Export via COM and persist to cache
            try:
                import win32com.client
                abs_p = os.path.abspath(clean_path)
                mtime = os.path.getmtime(abs_p)
                cache_key = hashlib.md5(f"{abs_p}_{mtime}".encode("utf-8")).hexdigest()
                dest_p = os.path.join(self.get_cache_dir(), f"slide1_{cache_key}.png")

                ppt_app = win32com.client.Dispatch("PowerPoint.Application")
                pres = ppt_app.Presentations.Open(abs_p, ReadOnly=True, Untitled=False, WithWindow=False)
                pres.Slides(1).Export(dest_p, "PNG", int(self.canvas_w * 2), int(self.canvas_h * 2))
                pres.Close()
                if os.path.isfile(dest_p):
                    with Image.open(dest_p) as im:
                        return im.convert("RGB").resize((self.canvas_w, self.canvas_h), Image.Resampling.LANCZOS)
            except Exception as e:
                print(f"[NOTE] Actual template COM preview unavailable ({e}). Using vector rendering fallback.")

        return self._render_vector_slide()

    def _render_vector_slide(self) -> Image.Image:
        im = Image.new("RGB", (self.canvas_w, self.canvas_h), "#FFFFFF")
        draw = ImageDraw.Draw(im)
        draw.rectangle((0, 0, self.canvas_w - 1, self.canvas_h - 1), outline="#CBD5E1", width=2)
        grid = max(20, int(self.px_per_inch))
        for x in range(grid, self.canvas_w, grid):
            draw.line((x, 0, x, self.canvas_h), fill="#F8FAFC", width=1)
        for y in range(grid, self.canvas_h, grid):
            draw.line((0, y, self.canvas_w, y), fill="#F8FAFC", width=1)
        if self.prs is not None:
            for s in self.prs.slides:
                for sh in s.shapes:
                    if sh.has_text_frame and sh.text_frame.text.strip():
                        t = sh.text_frame.text.strip()
                        if re.search(r"\{\{.*?\}\}", t):
                            x1 = int((sh.left / 914400.0) * self.px_per_inch)
                            y1 = int((sh.top / 914400.0) * self.px_per_inch)
                            x2 = int(((sh.left + sh.width) / 914400.0) * self.px_per_inch)
                            y2 = int(((sh.top + sh.height) / 914400.0) * self.px_per_inch)
                            draw.rectangle((x1, y1, x2, y2), outline="#E2E8F0", fill="#F1F5F9")
                            draw.text((x1 + 4, y1 + 3), t[:25], fill="#64748B")
        return im

    def _build_ui(self):
        # 1. Header Bar
        hdr = tk.Frame(self.root, bg="#1E293B", padx=16, pady=10, relief="solid", bd=1)
        hdr.pack(fill=tk.X)
        tk.Label(hdr, text="AutoCrat Pro | Visual Layout & Asset Designer", font=("Segoe UI", 11, "bold"), fg="#38BDF8", bg="#1E293B").pack(side=tk.LEFT)
        self.lbl_badge_status = tk.Label(hdr, text="Live Participant Preview (Record 1)", font=("Segoe UI", 8, "bold"), fg="#34D399", bg="#0F172A", padx=10, pady=3)
        self.lbl_badge_status.pack(side=tk.LEFT, padx=14)

        # 2. Footer Action Bar
        ftr = tk.Frame(self.root, bg="#1E293B", padx=16, pady=8, relief="solid", bd=1)
        ftr.pack(side=tk.BOTTOM, fill=tk.X)
        tk.Button(ftr, text="Apply & Save Layout", command=self._apply, font=("Segoe UI", 9, "bold"), bg="#0284C7", fg="#FFFFFF", activebackground="#0369A1", relief="flat", padx=22, pady=6, cursor="hand2").pack(side=tk.RIGHT)
        tk.Button(ftr, text="Discard Changes", command=self._cancel, font=("Segoe UI", 9), bg="#334155", fg="#F8FAFC", activebackground="#475569", relief="flat", padx=16, pady=6, cursor="hand2").pack(side=tk.RIGHT, padx=(0, 10))
        tk.Label(ftr, text="Interactive Canvas: Drag bounding box to reposition. Drag corner handles to scale. Click any asset to configure.", font=("Segoe UI", 8, "italic"), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)

        # 3. Central Working Area
        body = tk.Frame(self.root, bg="#0F172A", padx=10, pady=6)
        body.pack(fill=tk.BOTH, expand=True)

        # Visual Canvas Box (Left)
        canvas_box = tk.Frame(body, bg="#1E293B", padx=4, pady=4, relief="solid", bd=1)
        canvas_box.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

        self.canvas = tk.Canvas(canvas_box, width=self.canvas_w, height=self.canvas_h, bg="#FFFFFF", highlightthickness=0)
        self.canvas.pack(anchor="center", expand=True)
        self.canvas.bind("<ButtonPress-1>", self._on_canvas_press)
        self.canvas.bind("<B1-Motion>", self._on_canvas_motion)
        self.canvas.bind("<ButtonRelease-1>", self._on_canvas_release)
        self.canvas.bind("<Motion>", self._on_canvas_hover)
        self.canvas.bind("<Leave>", lambda e: self.canvas.configure(cursor="arrow"))

        # Inspector Sidebar (Right)
        sidebar = tk.Frame(body, bg="#1E293B", width=420, padx=12, pady=8, relief="solid", bd=1)
        sidebar.pack(side=tk.RIGHT, fill=tk.Y, padx=(10, 0))
        sidebar.pack_propagate(False)

        # Asset Selector Bar - Row 1 (Selection & Removal)
        elem_bar_top = tk.Frame(sidebar, bg="#1E293B")
        elem_bar_top.pack(fill=tk.X, pady=(0, 6))
        tk.Label(elem_bar_top, text="Active Asset:", font=("Segoe UI", 9, "bold"), fg="#F8FAFC", bg="#1E293B").pack(side=tk.LEFT)

        self.cb_active_elem = ttk.Combobox(elem_bar_top, values=self._get_all_tags(), font=("Segoe UI", 8))
        self.cb_active_elem.set(self.active_tag)
        self.cb_active_elem.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=(6, 6))
        self.cb_active_elem.bind("<<ComboboxSelected>>", self._on_element_combo_changed)

        self.btn_delete_asset = tk.Button(elem_bar_top, text="Remove Asset", command=self._delete_active_element, font=("Segoe UI", 8, "bold"), bg="#DC2626", fg="#FFFFFF", activebackground="#B91C1C", relief="flat", padx=10, pady=3, cursor="hand2")
        self.btn_delete_asset.pack(side=tk.RIGHT)

        # Asset Selector Bar - Row 2 (Addition Toolbar)
        elem_bar_actions = tk.Frame(sidebar, bg="#1E293B")
        elem_bar_actions.pack(fill=tk.X, pady=(0, 8))
        tk.Button(elem_bar_actions, text="+ Add Image Asset", command=self._add_image_element, font=("Segoe UI", 8, "bold"), bg="#0D9488", fg="#FFFFFF", activebackground="#0F766E", relief="flat", pady=4, cursor="hand2").pack(side=tk.LEFT, fill=tk.X, expand=True, padx=(0, 3))
        tk.Button(elem_bar_actions, text="+ Add QR Code", command=self._add_qr_element, font=("Segoe UI", 8, "bold"), bg="#7C3AED", fg="#FFFFFF", activebackground="#6D28D9", relief="flat", pady=4, cursor="hand2").pack(side=tk.LEFT, fill=tk.X, expand=True, padx=(3, 0))

        # Settings Container Frame
        self.frame_settings = tk.Frame(sidebar, bg="#1E293B")
        self.frame_settings.pack(fill=tk.BOTH, expand=True)

    def _get_all_tags(self) -> List[str]:
        return list(self.image_configs.keys()) + list(self.qr_configs.keys())

    def _add_image_element(self):
        idx = len(self.image_configs) + 1
        new_tag = f"{{{{image_{idx}}}}}"
        default_folder = r"D:\New folder (3)\images" if os.path.isdir(r"D:\New folder (3)\images") else ""
        self.image_configs[new_tag] = ImageElementConfig(
            tag=new_tag,
            enabled=True,
            source_mode="folder" if default_folder else "column",
            image_folder=default_folder,
            id_column="Serial" if "Serial" in self.columns else (self.columns[0] if self.columns else ""),
            shape_mode="rounded",
            corner_radius=25,
            border_width=3,
            border_color="#1E3A8A",
            position_mode="custom",
            custom_left_in=round(1.5 + (idx * 0.4), 2),
            custom_top_in=round(1.5 + (idx * 0.4), 2),
            custom_width_in=1.8,
            custom_height_in=2.2
        )
        self.cb_active_elem.configure(values=self._get_all_tags())
        self.cb_active_elem.set(new_tag)
        self.active_tag = new_tag
        self.current_tab_type = "image"
        self._load_active_element_ui()
        self._redraw_canvas()

    def _add_qr_element(self):
        idx = len(self.qr_configs) + 1
        new_tag = f"{{{{qr_{idx}}}}}"
        self.qr_configs[new_tag] = QRElementConfig(
            tag=new_tag,
            enabled=True,
            payload_mode="template",
            data_template=f"https://carnival2026.org/verify?id={{{{Serial}}}}",
            module_drawer="rounded",
            preset_theme="corporate_blue",
            position_mode="custom",
            custom_left_in=round(3.0 + (idx * 0.4), 2),
            custom_top_in=round(3.0 + (idx * 0.4), 2),
            custom_size_in=1.6
        )
        self.cb_active_elem.configure(values=self._get_all_tags())
        self.cb_active_elem.set(new_tag)
        self.active_tag = new_tag
        self.current_tab_type = "qr"
        self._load_active_element_ui()
        self._redraw_canvas()

    def _delete_active_element(self):
        """
        Removes the currently selected image or QR element from the configuration.
        """
        if not self.active_tag:
            return

        tag = self.active_tag
        if tag in self.image_configs:
            del self.image_configs[tag]
        elif tag in self.qr_configs:
            del self.qr_configs[tag]

        all_tags = self._get_all_tags()
        if all_tags:
            self.active_tag = all_tags[0]
            self.current_tab_type = "image" if self.active_tag in self.image_configs else "qr"
            self.cb_active_elem.configure(values=all_tags)
            self.cb_active_elem.set(self.active_tag)
            self._load_active_element_ui()
        else:
            self.active_tag = ""
            self.cb_active_elem.configure(values=[])
            self.cb_active_elem.set("")
            for w in self.frame_settings.winfo_children():
                w.destroy()
            tk.Label(self.frame_settings, text="No visual assets configured.\nClick + Image or + QR Code to add elements.", font=("Segoe UI", 9), fg="#94A3B8", bg="#1E293B").pack(pady=30)

        self._redraw_canvas()

    def _on_element_combo_changed(self, event=None):
        selected = self.cb_active_elem.get()
        if selected in self.image_configs:
            self.active_tag = selected
            self.current_tab_type = "image"
            self._load_active_element_ui()
            self._redraw_canvas()
        elif selected in self.qr_configs:
            self.active_tag = selected
            self.current_tab_type = "qr"
            self._load_active_element_ui()
            self._redraw_canvas()

    def _load_active_element_ui(self):
        for widget in self.frame_settings.winfo_children():
            widget.destroy()

        if self.current_tab_type == "image" and self.active_tag in self.image_configs:
            self._build_image_settings_ui(self.frame_settings, self.image_configs[self.active_tag])
        elif self.current_tab_type == "qr" and self.active_tag in self.qr_configs:
            self._build_qr_settings_ui(self.frame_settings, self.qr_configs[self.active_tag])

    def _build_image_settings_ui(self, parent, cfg: ImageElementConfig):
        # Enable Checkbox
        var_en = tk.BooleanVar(value=cfg.enabled)
        chk = tk.Checkbutton(parent, text=f"Enable Asset ({cfg.tag})", variable=var_en, font=("Segoe UI", 9, "bold"), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A", activebackground="#1E293B")
        chk.pack(anchor="w", pady=(0, 4))
        chk.configure(command=lambda: [setattr(cfg, "enabled", var_en.get()), self._redraw_canvas()])

        # Source & Mapping
        src_box = tk.LabelFrame(parent, text="Asset Source & Data Mapping", font=("Segoe UI", 8, "bold"), fg="#C084FC", bg="#1E293B", padx=8, pady=4)
        src_box.pack(fill=tk.X, pady=(0, 4))

        var_mode = tk.StringVar(value=cfg.source_mode)
        mode_options = [
            ("folder", "Directory Lookup (Match by Record ID)"),
            ("column", "Spreadsheet Column (Direct File Path / URL)"),
            ("static", "Fixed Static Asset (Uniform across records)")
        ]
        for m, l in mode_options:
            rb = tk.Radiobutton(src_box, text=l, variable=var_mode, value=m, font=("Segoe UI", 8), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A")
            rb.pack(anchor="w")
            rb.configure(command=lambda: [setattr(cfg, "source_mode", var_mode.get()), self._load_active_element_ui(), self._redraw_canvas()])

        if cfg.source_mode == "folder":
            f_row = tk.Frame(src_box, bg="#1E293B")
            f_row.pack(fill=tk.X, pady=(2, 2))
            ent_f = tk.Entry(f_row, font=("Segoe UI", 8))
            ent_f.insert(0, cfg.image_folder or "")
            ent_f.pack(side=tk.LEFT, fill=tk.X, expand=True)
            def _sync_fld(e=None):
                cfg.image_folder = ent_f.get().strip()
                self._redraw_canvas()
            ent_f.bind("<FocusOut>", _sync_fld)
            ent_f.bind("<KeyRelease>", _sync_fld)
            tk.Button(f_row, text="Browse...", command=lambda: self._browse_folder_for(cfg, ent_f), font=("Segoe UI", 8), bg="#334155", fg="#F8FAFC", relief="flat", padx=6, pady=1).pack(side=tk.LEFT, padx=3)

            id_row = tk.Frame(src_box, bg="#1E293B")
            id_row.pack(fill=tk.X, pady=(2, 2))
            tk.Label(id_row, text="Record Identifier Column:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
            cb_id = ttk.Combobox(id_row, values=self.columns, font=("Segoe UI", 8), width=16)
            cb_id.set(cfg.id_column or "")
            cb_id.pack(side=tk.LEFT, padx=4)
            cb_id.bind("<<ComboboxSelected>>", lambda e: [setattr(cfg, "id_column", cb_id.get()), self._redraw_canvas()])
        elif cfg.source_mode == "column":
            c_row = tk.Frame(src_box, bg="#1E293B")
            c_row.pack(fill=tk.X, pady=2)
            tk.Label(c_row, text="Select Image Column:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
            cb_col = ttk.Combobox(c_row, values=self.columns, font=("Segoe UI", 8))
            cb_col.set(cfg.data_column or "")
            cb_col.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=4)
            cb_col.bind("<<ComboboxSelected>>", lambda e: [setattr(cfg, "data_column", cb_col.get()), self._redraw_canvas()])
        else: # static
            s_row = tk.Frame(src_box, bg="#1E293B")
            s_row.pack(fill=tk.X, pady=2)
            ent_s = tk.Entry(s_row, font=("Segoe UI", 8))
            ent_s.insert(0, cfg.static_path or "")
            ent_s.pack(side=tk.LEFT, fill=tk.X, expand=True)
            def _sync_st(e=None):
                cfg.static_path = ent_s.get().strip()
                self._redraw_canvas()
            ent_s.bind("<FocusOut>", _sync_st)
            ent_s.bind("<KeyRelease>", _sync_st)
            tk.Button(s_row, text="Browse...", command=lambda: self._browse_static_for(cfg, ent_s), font=("Segoe UI", 8), bg="#334155", fg="#F8FAFC", relief="flat", padx=6, pady=1).pack(side=tk.LEFT, padx=3)

        # Styling Box
        sty_box = tk.LabelFrame(parent, text="Geometry, Border & Fit Styling", font=("Segoe UI", 8, "bold"), fg="#C084FC", bg="#1E293B", padx=8, pady=4)
        sty_box.pack(fill=tk.X, pady=(0, 4))

        # Shape Geometry
        sh_row = tk.Frame(sty_box, bg="#1E293B")
        sh_row.pack(fill=tk.X, pady=1)
        tk.Label(sh_row, text="Geometry:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        var_sh = tk.StringVar(value=self.SHAPE_KEY_MAP.get(cfg.shape_mode, "Rounded Rectangle"))
        for s_lbl in ["Rectangle", "Square", "Rounded Rectangle", "Circular (Avatar)"]:
            rb_s = tk.Radiobutton(sh_row, text=s_lbl, variable=var_sh, value=s_lbl, font=("Segoe UI", 8), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A")
            rb_s.pack(side=tk.LEFT, padx=1)
            rb_s.configure(command=lambda: [setattr(cfg, "shape_mode", self.SHAPE_DISPLAY_MAP.get(var_sh.get(), "rounded")), self._redraw_canvas()])

        # Corner Radius (User adjustable)
        rad_row = tk.Frame(sty_box, bg="#1E293B")
        rad_row.pack(fill=tk.X, pady=1)
        tk.Label(rad_row, text="Corner Radius (px):", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        var_rad = tk.IntVar(value=cfg.corner_radius)
        sc_rad = tk.Scale(rad_row, from_=0, to=100, variable=var_rad, orient=tk.HORIZONTAL, bg="#1E293B", fg="#F8FAFC", highlightthickness=0, troughcolor="#0F172A")
        sc_rad.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=4)
        sc_rad.configure(command=lambda v: [setattr(cfg, "corner_radius", var_rad.get()), self._redraw_canvas()])

        # Border Stroke & Color
        b_row = tk.Frame(sty_box, bg="#1E293B")
        b_row.pack(fill=tk.X, pady=1)
        tk.Label(b_row, text="Border Width (px):", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        var_bw = tk.IntVar(value=cfg.border_width)
        sc_bw = tk.Scale(b_row, from_=0, to=12, variable=var_bw, orient=tk.HORIZONTAL, bg="#1E293B", fg="#F8FAFC", highlightthickness=0, troughcolor="#0F172A")
        sc_bw.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=4)
        sc_bw.configure(command=lambda v: [setattr(cfg, "border_width", var_bw.get()), self._redraw_canvas()])

        c_row = tk.Frame(sty_box, bg="#1E293B")
        c_row.pack(fill=tk.X, pady=2)
        tk.Label(c_row, text="Border Stroke Color:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        swatch = tk.Label(c_row, text="      ", bg=cfg.border_color, relief="solid", bd=1)
        swatch.pack(side=tk.LEFT, padx=4)
        tk.Button(c_row, text="Select Color...", command=lambda: self._pick_color_for(cfg, swatch), font=("Segoe UI", 8), bg="#334155", fg="#F8FAFC", relief="flat", padx=6, pady=1).pack(side=tk.LEFT)

        # Fit Mode & Grayscale
        opt_row = tk.Frame(sty_box, bg="#1E293B")
        opt_row.pack(fill=tk.X, pady=2)
        tk.Label(opt_row, text="Sizing & Fit Mode:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        cb_fit = ttk.Combobox(opt_row, values=list(self.FIT_DISPLAY_MAP.keys()), font=("Segoe UI", 8), width=18)
        current_fit_display = self.FIT_KEY_MAP.get(cfg.fitting_mode, "Cover / Crop (Fill)")
        cb_fit.set(current_fit_display)
        cb_fit.pack(side=tk.LEFT, padx=3)
        cb_fit.bind("<<ComboboxSelected>>", lambda e: [setattr(cfg, "fitting_mode", self.FIT_DISPLAY_MAP.get(cb_fit.get(), "fill")), self._redraw_canvas()])

        var_gray = tk.BooleanVar(value=cfg.grayscale)
        chk_g = tk.Checkbutton(opt_row, text="Grayscale (B&W)", variable=var_gray, font=("Segoe UI", 8), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A")
        chk_g.pack(side=tk.LEFT, padx=6)
        chk_g.configure(command=lambda: [setattr(cfg, "grayscale", var_gray.get()), self._redraw_canvas()])

    def _build_qr_settings_ui(self, parent, cfg: QRElementConfig):
        # Enable Checkbox
        var_en = tk.BooleanVar(value=cfg.enabled)
        chk = tk.Checkbutton(parent, text=f"Enable QR Code ({cfg.tag})", variable=var_en, font=("Segoe UI", 9, "bold"), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A", activebackground="#1E293B")
        chk.pack(anchor="w", pady=(0, 4))
        chk.configure(command=lambda: [setattr(cfg, "enabled", var_en.get()), self._redraw_canvas()])

        # QR Payload Box
        src_box = tk.LabelFrame(parent, text="QR Payload & Data Mapping", font=("Segoe UI", 8, "bold"), fg="#38BDF8", bg="#1E293B", padx=8, pady=4)
        src_box.pack(fill=tk.X, pady=(0, 4))

        var_mode = tk.StringVar(value=cfg.payload_mode)
        mode_options = [
            ("column", "Single Spreadsheet Column"),
            ("template", "Dynamic URL Template (e.g., https://.../{{Serial}})"),
            ("static", "Static Payload / URL (Uniform across records)"),
            ("multi_column", "Multi-Column Manifest (Key: Value Payload)")
        ]
        for m, l in mode_options:
            rb = tk.Radiobutton(src_box, text=l, variable=var_mode, value=m, font=("Segoe UI", 8), fg="#F8FAFC", bg="#1E293B", selectcolor="#0F172A")
            rb.pack(anchor="w")
            rb.configure(command=lambda: [setattr(cfg, "payload_mode", var_mode.get()), self._load_active_element_ui(), self._redraw_canvas()])

        if cfg.payload_mode == "column":
            c_row = tk.Frame(src_box, bg="#1E293B")
            c_row.pack(fill=tk.X, pady=2)
            tk.Label(c_row, text="Select Data Column:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
            cb_c = ttk.Combobox(c_row, values=self.columns, font=("Segoe UI", 8))
            cb_c.set(cfg.data_column or "")
            cb_c.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=4)
            cb_c.bind("<<ComboboxSelected>>", lambda e: [setattr(cfg, "data_column", cb_c.get()), self._redraw_canvas()])
        elif cfg.payload_mode == "template":
            ent_t = tk.Entry(src_box, font=("Segoe UI", 8))
            ent_t.insert(0, cfg.data_template or "")
            ent_t.pack(fill=tk.X, pady=2)
            def _sync_tpl(e=None):
                cfg.data_template = ent_t.get().strip()
                self._redraw_canvas()
            ent_t.bind("<FocusOut>", _sync_tpl)
            ent_t.bind("<KeyRelease>", _sync_tpl)
        elif cfg.payload_mode == "static":
            ent_s = tk.Entry(src_box, font=("Segoe UI", 8))
            ent_s.insert(0, cfg.static_payload or "")
            ent_s.pack(fill=tk.X, pady=2)
            def _sync_stat(e=None):
                cfg.static_payload = ent_s.get().strip()
                self._redraw_canvas()
            ent_s.bind("<FocusOut>", _sync_stat)
            ent_s.bind("<KeyRelease>", _sync_stat)
        else: # multi_column
            lb_m = tk.Listbox(src_box, selectmode=tk.MULTIPLE, font=("Segoe UI", 8), height=3, bg="#0F172A", fg="#F8FAFC", selectbackground="#0284C7")
            lb_m.pack(fill=tk.X, pady=2)
            for c in self.columns:
                lb_m.insert(tk.END, c)
                if c in cfg.multi_columns:
                    lb_m.selection_set(tk.END)
            lb_m.bind("<<ListboxSelect>>", lambda e: [setattr(cfg, "multi_columns", [self.columns[i] for i in lb_m.curselection()]), self._redraw_canvas()])

        # Appearance & Themes
        app_box = tk.LabelFrame(parent, text="Visual Aesthetics & Theme Palette", font=("Segoe UI", 8, "bold"), fg="#38BDF8", bg="#1E293B", padx=8, pady=4)
        app_box.pack(fill=tk.X, pady=(0, 4))

        th_row = tk.Frame(app_box, bg="#1E293B")
        th_row.pack(fill=tk.X, pady=2)
        tk.Label(th_row, text="Color Scheme / Theme:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        theme_display_list = list(self.THEME_DISPLAY_MAP.keys())
        cb_th = ttk.Combobox(th_row, values=theme_display_list, font=("Segoe UI", 8), width=22)
        current_theme_display = self.THEME_KEY_MAP.get(cfg.preset_theme, "Corporate Deep Navy")
        cb_th.set(current_theme_display)
        cb_th.pack(side=tk.LEFT, padx=4)
        def _on_theme_select(e=None):
            cfg.preset_theme = self.THEME_DISPLAY_MAP.get(cb_th.get(), "corporate_blue")
            self._load_active_element_ui()
            self._redraw_canvas()
        cb_th.bind("<<ComboboxSelected>>", _on_theme_select)

        if cfg.preset_theme == "custom":
            cust_row = tk.Frame(app_box, bg="#1E293B")
            cust_row.pack(fill=tk.X, pady=2)
            tk.Label(cust_row, text="Foreground:", font=("Segoe UI", 7), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
            sw_fg = tk.Label(cust_row, text="    ", bg=cfg.fg_color, relief="solid", bd=1)
            sw_fg.pack(side=tk.LEFT, padx=2)
            tk.Button(cust_row, text="Pick...", command=lambda: self._pick_qr_color(cfg, "fg_color", sw_fg), font=("Segoe UI", 7), bg="#334155", fg="#F8FAFC", relief="flat", padx=3).pack(side=tk.LEFT)
            tk.Label(cust_row, text="Background:", font=("Segoe UI", 7), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT, padx=(6, 0))
            sw_bg = tk.Label(cust_row, text="    ", bg=cfg.bg_color, relief="solid", bd=1)
            sw_bg.pack(side=tk.LEFT, padx=2)
            tk.Button(cust_row, text="Pick...", command=lambda: self._pick_qr_color(cfg, "bg_color", sw_bg), font=("Segoe UI", 7), bg="#334155", fg="#F8FAFC", relief="flat", padx=3).pack(side=tk.LEFT)

        dr_row = tk.Frame(app_box, bg="#1E293B")
        dr_row.pack(fill=tk.X, pady=2)
        tk.Label(dr_row, text="Module Pattern Style:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        drawer_display_list = list(self.DRAWER_DISPLAY_MAP.keys())
        cb_dr = ttk.Combobox(dr_row, values=drawer_display_list, font=("Segoe UI", 8), width=18)
        current_dr_display = self.DRAWER_KEY_MAP.get(cfg.module_drawer, "Smooth Rounded")
        cb_dr.set(current_dr_display)
        cb_dr.pack(side=tk.LEFT, padx=4)
        cb_dr.bind("<<ComboboxSelected>>", lambda e: [setattr(cfg, "module_drawer", self.DRAWER_DISPLAY_MAP.get(cb_dr.get(), "rounded")), self._redraw_canvas()])

        # Center Branding Logo
        l_row = tk.Frame(app_box, bg="#1E293B")
        l_row.pack(fill=tk.X, pady=2)
        tk.Label(l_row, text="Center Branding Emblem:", font=("Segoe UI", 8), fg="#94A3B8", bg="#1E293B").pack(side=tk.LEFT)
        ent_l = tk.Entry(l_row, font=("Segoe UI", 8))
        ent_l.insert(0, cfg.logo_path or "")
        ent_l.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=2)
        def _sync_logo(e=None):
            cfg.logo_path = ent_l.get().strip().strip("'\"")
            self._redraw_canvas()
        ent_l.bind("<KeyRelease>", _sync_logo)
        ent_l.bind("<FocusOut>", _sync_logo)
        tk.Button(l_row, text="Browse...", command=lambda: self._browse_logo_for(cfg, ent_l), font=("Segoe UI", 8), bg="#334155", fg="#F8FAFC", relief="flat", padx=5, pady=1).pack(side=tk.LEFT)
        tk.Button(l_row, text="Clear", command=lambda: [ent_l.delete(0, tk.END), setattr(cfg, "logo_path", ""), self._redraw_canvas()], font=("Segoe UI", 8), bg="#475569", fg="#F8FAFC", relief="flat", padx=4, pady=1).pack(side=tk.LEFT, padx=2)

    def _browse_folder_for(self, cfg: ImageElementConfig, ent: tk.Entry):
        try:
            prev = bool(self.root.attributes("-topmost"))
            if prev: self.root.attributes("-topmost", False)
            fld = filedialog.askdirectory(parent=self.root, title="Select Participant Images Directory")
            if prev: self.root.attributes("-topmost", True)
            self.root.lift()
            self.root.focus_force()
            if fld:
                cfg.image_folder = os.path.abspath(fld)
                ent.delete(0, tk.END)
                ent.insert(0, cfg.image_folder)
                self._redraw_canvas()
        except Exception as e:
            messagebox.showerror("Directory Selection Error", str(e), parent=self.root)

    def _browse_static_for(self, cfg: ImageElementConfig, ent: tk.Entry):
        try:
            prev = bool(self.root.attributes("-topmost"))
            if prev: self.root.attributes("-topmost", False)
            fl = filedialog.askopenfilename(parent=self.root, title="Select Static Graphic Asset", filetypes=[("Image Files", "*.png;*.jpg;*.jpeg;*.webp")])
            if prev: self.root.attributes("-topmost", True)
            self.root.lift()
            self.root.focus_force()
            if fl:
                cfg.static_path = os.path.abspath(fl)
                ent.delete(0, tk.END)
                ent.insert(0, cfg.static_path)
                self._redraw_canvas()
        except Exception as e:
            messagebox.showerror("File Selection Error", str(e), parent=self.root)

    def _browse_logo_for(self, cfg: QRElementConfig, ent: tk.Entry):
        try:
            prev = bool(self.root.attributes("-topmost"))
            if prev: self.root.attributes("-topmost", False)
            fl = filedialog.askopenfilename(parent=self.root, title="Select Center Branding Emblem", filetypes=[("Image Files", "*.png;*.jpg;*.jpeg;*.webp")])
            if prev: self.root.attributes("-topmost", True)
            self.root.lift()
            self.root.focus_force()
            if fl:
                cfg.logo_path = os.path.abspath(fl)
                ent.delete(0, tk.END)
                ent.insert(0, cfg.logo_path)
                self._redraw_canvas()
        except Exception as e:
            messagebox.showerror("Emblem Selection Error", str(e), parent=self.root)

    def _pick_color_for(self, cfg: ImageElementConfig, swatch: tk.Label):
        try:
            self.root.update_idletasks()
            prev = bool(self.root.attributes("-topmost"))
            if prev: self.root.attributes("-topmost", False)
            c = colorchooser.askcolor(color=cfg.border_color, title="Select Border Stroke Color", parent=self.root)
            if prev: self.root.attributes("-topmost", True)
            self.root.lift()
            self.root.focus_force()
            if c and c[1]:
                hex_c = c[1].upper()
                cfg.border_color = hex_c
                swatch.configure(bg=hex_c)
                self._redraw_canvas()
        except Exception as e:
            messagebox.showerror("Color Picker Error", str(e), parent=self.root)

    def _pick_qr_color(self, cfg: QRElementConfig, attr_name: str, swatch: tk.Label):
        try:
            self.root.update_idletasks()
            prev = bool(self.root.attributes("-topmost"))
            if prev: self.root.attributes("-topmost", False)
            cur = getattr(cfg, attr_name, "#000000")
            c = colorchooser.askcolor(color=cur, title="Select Palette Color", parent=self.root)
            if prev: self.root.attributes("-topmost", True)
            self.root.lift()
            self.root.focus_force()
            if c and c[1]:
                hex_c = c[1].upper()
                setattr(cfg, attr_name, hex_c)
                swatch.configure(bg=hex_c)
                self._redraw_canvas()
        except Exception as e:
            messagebox.showerror("Color Picker Error", str(e), parent=self.root)

    def _redraw_canvas(self):
        self.canvas.delete("all")
        self.rendered_preview_refs = []

        # 1. Slide Actual Background / Vector Fallback
        self.bg_tk = ImageTk.PhotoImage(self.bg_image_pil.resize((self.canvas_w, self.canvas_h), Image.Resampling.LANCZOS))
        self.canvas.create_image(0, 0, image=self.bg_tk, anchor="nw")

        row0 = self.df.iloc[0] if self.df is not None and len(self.df) > 0 else pd.Series({})
        row0_map = {str(k): str(v) for k, v in row0.to_dict().items()}

        # 2. Render all Image Elements
        for tag, cfg in self.image_configs.items():
            if not cfg.enabled:
                continue

            ix = int(cfg.custom_left_in * self.px_per_inch)
            iy = int(cfg.custom_top_in * self.px_per_inch)
            iw = int(cfg.custom_width_in * self.px_per_inch)
            ih = int(cfg.custom_height_in * self.px_per_inch)

            img_path, _ = AdvancedImageProcessor.resolve_image_path(cfg, row0, row0_map, self.temp_dir)
            rendered_live = False

            if img_path and os.path.isfile(img_path) and iw > 20 and ih > 20:
                try:
                    temp_out = os.path.join(self.temp_dir, f"_canvas_{tag.strip('{}')}.png")
                    AdvancedImageProcessor.process_image(img_path, iw, ih, cfg, temp_out)
                    if os.path.isfile(temp_out):
                        with Image.open(temp_out) as p_img:
                            p_tk = ImageTk.PhotoImage(p_img)
                        self.rendered_preview_refs.append(p_tk)
                        self.canvas.create_image(ix, iy, image=p_tk, anchor="nw", tags=(tag, "box"))
                        rendered_live = True
                except Exception:
                    pass

            if not rendered_live:
                if cfg.shape_mode == "circle":
                    self.canvas.create_oval(ix, iy, ix + iw, iy + ih, fill="#F3E8FF", outline=cfg.border_color, width=max(1, cfg.border_width), tags=(tag, "box"))
                else:
                    self.canvas.create_rectangle(ix, iy, ix + iw, iy + ih, fill="#F3E8FF", outline=cfg.border_color, width=max(1, cfg.border_width), tags=(tag, "box"))
                cx, cy = ix + iw // 2, iy + ih // 2
                self.canvas.create_text(cx, cy, text=tag, font=("Segoe UI", 9, "bold"), fill="#6B21A8", tags=(tag, "text"))

            # Active indicator with 4 corner expandable handles
            if tag == self.active_tag:
                self.canvas.create_rectangle(ix - 2, iy - 2, ix + iw + 2, iy + ih + 2, outline="#A855F7", width=2, dash=(4, 2))
                hr = 7
                self.canvas.create_rectangle(ix + iw - hr, iy + ih - hr, ix + iw + hr, iy + ih + hr, fill="#A855F7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_se", "handle"))
                self.canvas.create_rectangle(ix - hr, iy - hr, ix + hr, iy + hr, fill="#A855F7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_nw", "handle"))
                self.canvas.create_rectangle(ix + iw - hr, iy - hr, ix + iw + hr, iy + hr, fill="#A855F7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_ne", "handle"))
                self.canvas.create_rectangle(ix - hr, iy + ih - hr, ix + hr, iy + ih + hr, fill="#A855F7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_sw", "handle"))

        # 3. Render all QR Elements
        for tag, cfg in self.qr_configs.items():
            if not cfg.enabled:
                continue

            qx = int(cfg.custom_left_in * self.px_per_inch)
            qy = int(cfg.custom_top_in * self.px_per_inch)
            qw = int(cfg.custom_size_in * self.px_per_inch)
            qh = qw

            payload = cfg.resolve_payload(row0_map, row0)
            rendered_qr = False

            if payload and qw > 20 and qh > 20:
                try:
                    temp_qr = os.path.join(self.temp_dir, f"_canvas_qr_{tag.strip('{}')}.png")
                    if AdvancedQRProcessor.generate_qr(payload, cfg, temp_qr, qw):
                        with Image.open(temp_qr) as q_img:
                            q_fit = q_img.resize((qw, qh), Image.Resampling.LANCZOS)
                            q_tk = ImageTk.PhotoImage(q_fit)
                        self.rendered_preview_refs.append(q_tk)
                        self.canvas.create_image(qx, qy, image=q_tk, anchor="nw", tags=(tag, "box"))
                        self.canvas.create_rectangle(qx, qy, qx + qw, qy + qh, outline="#0284C7", width=2, tags=(tag, "outline"))
                        rendered_qr = True
                except Exception:
                    pass

            if not rendered_qr:
                self.canvas.create_rectangle(qx, qy, qx + qw, qy + qh, fill="#E0F2FE", outline="#0284C7", width=2, tags=(tag, "box"))
                cx, cy = qx + qw // 2, qy + qh // 2
                self.canvas.create_text(cx, cy, text=tag, font=("Segoe UI", 9, "bold"), fill="#0369A1", tags=(tag, "text"))

            if tag == self.active_tag:
                self.canvas.create_rectangle(qx - 2, qy - 2, qx + qw + 2, qy + qh + 2, outline="#0284C7", width=2, dash=(4, 2))
                hr = 7
                self.canvas.create_rectangle(qx + qw - hr, qy + qh - hr, qx + qw + hr, qy + qh + hr, fill="#0284C7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_se", "handle"))
                self.canvas.create_rectangle(qx - hr, qy - hr, qx + hr, qy + hr, fill="#0284C7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_nw", "handle"))
                self.canvas.create_rectangle(qx + qw - hr, qy - hr, qx + qw + hr, qy + hr, fill="#0284C7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_ne", "handle"))
                self.canvas.create_rectangle(qx - hr, qy + qh - hr, qx + hr, qy + qh + hr, fill="#0284C7", outline="#FFFFFF", width=1, tags=(f"{tag}_handle_sw", "handle"))

    def _get_hit_detail(self, x: int, y: int) -> Tuple[Optional[str], Optional[str], Optional[str]]:
        hr = 10
        # 1. Check handles of active element first
        for tag, cfg in list(self.image_configs.items()) + list(self.qr_configs.items()):
            if not cfg.enabled or tag != self.active_tag:
                continue
            ix = int(cfg.custom_left_in * self.px_per_inch)
            iy = int(cfg.custom_top_in * self.px_per_inch)
            iw = int((cfg.custom_width_in if hasattr(cfg, "custom_width_in") else cfg.custom_size_in) * self.px_per_inch)
            ih = int((cfg.custom_height_in if hasattr(cfg, "custom_height_in") else cfg.custom_size_in) * self.px_per_inch)

            if (ix + iw - hr <= x <= ix + iw + hr) and (iy + ih - hr <= y <= iy + ih + hr):
                return tag, "resize", "se"
            if (ix - hr <= x <= ix + hr) and (iy - hr <= y <= iy + hr):
                return tag, "resize", "nw"
            if (ix + iw - hr <= x <= ix + iw + hr) and (iy - hr <= y <= iy + hr):
                return tag, "resize", "ne"
            if (ix - hr <= x <= ix + hr) and (iy + ih - hr <= y <= iy + ih + hr):
                return tag, "resize", "sw"

        # 2. Check element body
        for tag, cfg in list(self.image_configs.items()) + list(self.qr_configs.items()):
            if not cfg.enabled:
                continue
            ix = int(cfg.custom_left_in * self.px_per_inch)
            iy = int(cfg.custom_top_in * self.px_per_inch)
            iw = int((cfg.custom_width_in if hasattr(cfg, "custom_width_in") else cfg.custom_size_in) * self.px_per_inch)
            ih = int((cfg.custom_height_in if hasattr(cfg, "custom_height_in") else cfg.custom_size_in) * self.px_per_inch)
            if ix <= x <= ix + iw and iy <= y <= iy + ih:
                return tag, "move", None

        return None, None, None

    def _on_canvas_hover(self, event):
        if self.active_drag_elem:
            return

        tag, action, handle_type = self._get_hit_detail(event.x, event.y)
        if action == "resize":
            if handle_type in ["se", "nw"]:
                self.canvas.configure(cursor="size_nw_se")
            elif handle_type in ["ne", "sw"]:
                self.canvas.configure(cursor="size_ne_sw")
            else:
                self.canvas.configure(cursor="size_nw_se")
        elif action == "move":
            self.canvas.configure(cursor="fleur")
        else:
            self.canvas.configure(cursor="arrow")

    def _on_canvas_press(self, event):
        tag, action, handle_type = self._get_hit_detail(event.x, event.y)
        if not tag:
            return

        self.active_tag = tag
        self.current_tab_type = "image" if tag in self.image_configs else "qr"
        self.cb_active_elem.set(tag)
        self._load_active_element_ui()

        self.active_drag_elem = tag
        self.drag_action = action
        self.drag_handle = handle_type or "se"
        self.drag_start_x = event.x
        self.drag_start_y = event.y

        cfg = self.image_configs[tag] if tag in self.image_configs else self.qr_configs[tag]
        self.elem_start_x = int(cfg.custom_left_in * self.px_per_inch)
        self.elem_start_y = int(cfg.custom_top_in * self.px_per_inch)
        self.elem_start_w = int((cfg.custom_width_in if hasattr(cfg, "custom_width_in") else cfg.custom_size_in) * self.px_per_inch)
        self.elem_start_h = int((cfg.custom_height_in if hasattr(cfg, "custom_height_in") else cfg.custom_size_in) * self.px_per_inch)

        if action == "move":
            self.canvas.configure(cursor="fleur")
        elif action == "resize":
            self.canvas.configure(cursor="size_nw_se" if self.drag_handle in ["se", "nw"] else "size_ne_sw")

        self._redraw_canvas()

    def _on_canvas_motion(self, event):
        if not self.active_drag_elem or not self.drag_action:
            return

        tag = self.active_drag_elem
        cfg = self.image_configs[tag] if tag in self.image_configs else self.qr_configs[tag]
        dx = event.x - self.drag_start_x
        dy = event.y - self.drag_start_y

        cfg.position_mode = "custom"

        if self.drag_action == "move":
            self.canvas.configure(cursor="fleur")
            new_x = max(0, min(self.canvas_w - self.elem_start_w, self.elem_start_x + dx))
            new_y = max(0, min(self.canvas_h - self.elem_start_h, self.elem_start_y + dy))
            cfg.custom_left_in = round(new_x / self.px_per_inch, 2)
            cfg.custom_top_in = round(new_y / self.px_per_inch, 2)
        elif self.drag_action == "resize":
            self.canvas.configure(cursor="size_nw_se" if self.drag_handle in ["se", "nw"] else "size_ne_sw")
            min_dim = int(0.5 * self.px_per_inch)
            if tag in self.image_configs:
                if self.drag_handle == "se":
                    new_w = max(min_dim, self.elem_start_w + dx)
                    new_h = max(min_dim, self.elem_start_h + dy)
                    cfg.custom_width_in = round(new_w / self.px_per_inch, 2)
                    cfg.custom_height_in = round(new_h / self.px_per_inch, 2)
                elif self.drag_handle == "nw":
                    new_w = max(min_dim, self.elem_start_w - dx)
                    new_h = max(min_dim, self.elem_start_h - dy)
                    cfg.custom_left_in = round(max(0, self.elem_start_x + dx) / self.px_per_inch, 2)
                    cfg.custom_top_in = round(max(0, self.elem_start_y + dy) / self.px_per_inch, 2)
                    cfg.custom_width_in = round(new_w / self.px_per_inch, 2)
                    cfg.custom_height_in = round(new_h / self.px_per_inch, 2)
                elif self.drag_handle == "ne":
                    new_w = max(min_dim, self.elem_start_w + dx)
                    new_h = max(min_dim, self.elem_start_h - dy)
                    cfg.custom_top_in = round(max(0, self.elem_start_y + dy) / self.px_per_inch, 2)
                    cfg.custom_width_in = round(new_w / self.px_per_inch, 2)
                    cfg.custom_height_in = round(new_h / self.px_per_inch, 2)
                elif self.drag_handle == "sw":
                    new_w = max(min_dim, self.elem_start_w - dx)
                    new_h = max(min_dim, self.elem_start_h + dy)
                    cfg.custom_left_in = round(max(0, self.elem_start_x + dx) / self.px_per_inch, 2)
                    cfg.custom_width_in = round(new_w / self.px_per_inch, 2)
                    cfg.custom_height_in = round(new_h / self.px_per_inch, 2)
            else:
                # QR code maintains 1:1 aspect ratio
                delta = max(dx, dy) if self.drag_handle in ["se", "sw"] else min(dx, dy)
                if self.drag_handle == "se":
                    new_sz = max(min_dim, self.elem_start_w + delta)
                else:
                    new_sz = max(min_dim, self.elem_start_w + dx)
                cfg.custom_size_in = round(new_sz / self.px_per_inch, 2)

        self._redraw_canvas()

    def _on_canvas_release(self, event):
        self.active_drag_elem = None
        self.drag_action = None
        self.drag_handle = None
        self._on_canvas_hover(event)

    def _apply(self):
        self.cancelled = False
        self.root.destroy()

    def _cancel(self):
        self.cancelled = True
        self.root.destroy()

    def show(self):
        if self.parent:
            self.root.transient(self.parent)
            self.root.grab_set()
            self.parent.wait_window(self.root)
        else:
            self.root.mainloop()
        return self.get_configs()

    def get_configs(self):
        if self.cancelled:
            return None, None

        primary_qr = list(self.qr_configs.values())[0] if self.qr_configs else None
        primary_img = list(self.image_configs.values())[0] if self.image_configs else None

        qr_out = primary_qr.to_dict() if primary_qr else {"enabled": False}
        img_out = primary_img.to_dict() if primary_img else {"enabled": False}

        has_any_qr = any(cfg.enabled for cfg in self.qr_configs.values()) if self.qr_configs else False
        has_any_img = any(cfg.enabled for cfg in self.image_configs.values()) if self.image_configs else False
        qr_out["enabled"] = has_any_qr
        img_out["enabled"] = has_any_img

        qr_out["multi_qrs"] = self.qr_configs
        img_out["multi_images"] = self.image_configs

        return qr_out, img_out


def prompt_qr_and_image_config(df: pd.DataFrame, prs: Optional[Presentation] = None, pptx_path: Optional[str] = None):
    try:
        editor = VisualDragDropCanvas(df, prs, pptx_path)
        raw_qr, raw_img = editor.show()
        if raw_qr is not None and raw_img is not None:
            return raw_qr, raw_img
    except Exception as e:
        print(f"[NOTE] Canvas not opened ({e}). Falling back.")
    return {"enabled": False}, {"enabled": False}


print('[+] AutoCrat Pro Core & Advanced Multi-Asset Engine loaded successfully!')
````

````python
# ==============================================================================
# CELL 3: AUTOCRAT PRO ENTERPRISE DESKTOP GUI DASHBOARD
# ==============================================================================
import os
import re
import sys
import time
import threading
import subprocess
from typing import List, Dict, Optional, Tuple, Any
import pandas as pd
from pptx import Presentation
import pypdf
from pypdf import PdfWriter, PdfReader
import tkinter as tk
from tkinter import ttk, filedialog, messagebox, colorchooser
from PIL import Image, ImageTk

def parse_row_selection(range_str: str, total_rows: int) -> List[int]:
    """
    Parses user-specified 1-based row numbers or ranges (e.g. '1-5, 8, 10-12')
    into a sorted list of zero-based DataFrame indices.
    """
    if not range_str or not range_str.strip() or range_str.strip().lower() == "all":
        return list(range(total_rows))
    selected_indices = set()
    parts = [p.strip() for p in range_str.replace(";", ",").split(",") if p.strip()]
    for part in parts:
        if "-" in part:
            sub = part.split("-")
            if len(sub) == 2:
                s_str, e_str = sub[0].strip(), sub[1].strip()
                if s_str and not e_str:
                    try:
                        start_1 = int(s_str)
                        for r in range(max(1, start_1), total_rows + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
                elif not s_str and e_str:
                    try:
                        end_1 = int(e_str)
                        for r in range(1, min(total_rows, end_1) + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
                elif s_str and e_str:
                    try:
                        start_1, end_1 = int(s_str), int(e_str)
                        if start_1 > end_1: start_1, end_1 = end_1, start_1
                        for r in range(max(1, start_1), min(total_rows, end_1) + 1):
                            selected_indices.add(r - 1)
                    except ValueError: pass
        else:
            try:
                r_1 = int(part)
                if 1 <= r_1 <= total_rows:
                    selected_indices.add(r_1 - 1)
            except ValueError: pass
    res = sorted(list(selected_indices))
    return res if res else list(range(total_rows))

# ==============================================================================
# Cell 3: Interactive Universal Dashboard (All-in-One Console)
# ==============================================================================
# Run this cell to launch the interactive AutoCrat Pro menu.
# You can choose any mode, browse files via Windows GUI dialogs, and monitor execution.


# ------------------------------------------------------------------------------
# Concurrency Worker Engine for Individual Document Processing (Mode 1)
# ------------------------------------------------------------------------------
def process_single_individual_pptx(item: dict):
    """
    Worker task: Concurrently generates 1 individual PPTX file (Text + QR + Image).
    Executed across concurrent worker threads without process spawn overhead or COM crashes.
    Returns (success, idx, temp_pptx, pdf_path, error_message).
    """
    idx = item["idx"]
    row_dict = item["row_dict"]
    pptx_path = item["pptx_path"]
    export_dir = item["export_dir"]
    id_cols = item.get("id_cols", [])
    qr_config = item.get("qr_config")
    img_config = item.get("img_config")

    temp_pptx = ""
    temp_qr_img = ""
    try:
        row = pd.Series(row_dict)
        custom_mapping = item.get("custom_mapping")
        mapping = PlaceholderEngine.build_mapping(row, custom_mapping)

        if id_cols:
            parts = [str(row[c]).strip() for c in id_cols if c in row and str(row[c]).strip()]
            base_name = FileHelper.sanitize_filename("-".join(parts) if parts else f"doc_{idx+1}")
        else:
            base_name = f"doc_{idx+1}"

        pdf_path = os.path.join(export_dir, f"{base_name}.pdf")
        temp_pptx = os.path.join(export_dir, f"{base_name}.pptx")
        temp_qr_img = os.path.join(export_dir, f"_temp_qr_{idx+1}.png")

        prs = Presentation(pptx_path)

        # 1. Fill text placeholders
        for slide in prs.slides:
            PlaceholderEngine.process_slide(slide, mapping)

        # 2. Embed QR codes and Images (Supports multiple elements per slide)
        for slide in prs.slides:
            MultiAssetEngine.embed_all_assets(slide, prs, row, mapping, img_config, qr_config, export_dir)

        prs.save(temp_pptx)
        return (True, idx, temp_pptx, pdf_path, "")
    except Exception as err:
        if temp_pptx:
            FileHelper.safe_delete(temp_pptx)
        if temp_qr_img:
            FileHelper.safe_delete(temp_qr_img)
        return (False, idx, "", "", str(err))

def run_mode_1_individual():
    print("\n" + "=" * 60)
    print("  MODE 1: Individual Document Generator (1 PDF per Row)")
    print("=" * 60)
    
    # 1. Data File
    data_path = DynamicPicker.select_file("Select Excel (.xlsx) or CSV file", [("Spreadsheets", "*.xlsx;*.xls;*.csv"), ("All Files", "*.*")])
    if not data_path or not os.path.exists(data_path):
        print("[!] No valid data file selected. Aborting.")
        return
        
    sheets = DataLoader.get_sheet_names(data_path)
    selected_sheet = sheets[0] if sheets else None
    if len(sheets) > 1:
        print(f"\nFound sheets: {sheets}")
        choice = input(f"Enter sheet name or index [Default: {sheets[0]}]: ").strip()
        if choice in sheets:
            selected_sheet = choice
        elif choice.isdigit() and 1 <= int(choice) <= len(sheets):
            selected_sheet = sheets[int(choice) - 1]
            
    df = DataLoader.load(data_path, selected_sheet)
    print(f"[OK] Loaded {len(df)} rows. Columns: {list(df.columns)}")

    # 2. Template
    pptx_path = DynamicPicker.select_file("Select PowerPoint Template (.pptx)", [("PowerPoint Presentation", "*.pptx")])
    if not pptx_path or not os.path.exists(pptx_path):
        print("[!] No valid template selected. Aborting.")
        return

    prs_inspect = Presentation(pptx_path)

    # 3. Filename Columns
    print(f"\nColumns available: {list(df.columns)}")
    cols_input = input("Enter column name(s) for output filename (comma-separated, Enter for row_1): ").strip()
    id_cols = [c.strip() for c in cols_input.split(",") if c.strip() in df.columns]
    
    # 4. Optional Dynamic QR Code & Image / Photo Configuration (GUI Dialog)
    qr_config, img_config = prompt_qr_and_image_config(df, prs_inspect, pptx_path)

    # 6. Confirmation & Output
    export_dir = FileHelper.get_export_dir(pptx_path)
    print(f"\n[DIR] Files will be saved in: {export_dir}")
    proceed = input(f"Generate {len(df)} individual PDF files? (y/n) [Default: y]: ").strip().lower()
    if proceed not in ["", "y", "yes"]:
        print("Cancelled.")
        return

    num_workers = min(8, max(1, len(df)))
    success_count = 0
    start_time = time.time()
    print(f"\n[*] Processing {len(df)} documents with {num_workers} concurrent workers...")

    items = []
    for idx, row in df.iterrows():
        items.append({
            "idx": idx,
            "row_dict": row.to_dict(),
            "pptx_path": pptx_path,
            "export_dir": export_dir,
            "id_cols": id_cols,
            "qr_config": qr_config,
            "img_config": img_config
        })

    converter = PPTXToPDFConverter()
    if not converter.start():
        print("[!] PowerPoint COM engine unavailable; PDF generation aborted.")
        return

    try:
        with concurrent.futures.ThreadPoolExecutor(max_workers=num_workers) as executor:
            futures = [executor.submit(process_single_individual_pptx, item) for item in items]
            for fut in tqdm(concurrent.futures.as_completed(futures), total=len(items), desc=f"Generating PDFs ({num_workers} Workers)"):
                ok, idx, temp_pptx, pdf_path, err = fut.result()
                if ok and temp_pptx:
                    if converter.convert(temp_pptx, pdf_path):
                        success_count += 1
                    FileHelper.safe_delete(temp_pptx)
                elif err:
                    print(f"\n    [ERR] Row {idx+1} failed: {err}")
    finally:
        converter.stop()

    elapsed = round(time.time() - start_time, 2)
    print("\n" + "=" * 60)
    print(f"  [COMPLETED] {success_count}/{len(df)} PDFs successfully generated in {elapsed}s")
    print(f"  [SAVED IN]  {export_dir}")
    print("=" * 60)


def run_mode_2_merged():
    print("\n" + "=" * 60)
    print("  MODE 2: Merged Master Document Generator (All Rows in 1 File)")
    print("=" * 60)
    
    data_path = DynamicPicker.select_file("Select Excel (.xlsx) or CSV file", [("Spreadsheets", "*.xlsx;*.xls;*.csv"), ("All Files", "*.*")])
    if not data_path or not os.path.exists(data_path):
        return
        
    sheets = DataLoader.get_sheet_names(data_path)
    selected_sheet = sheets[0] if sheets else None
    if len(sheets) > 1:
        print(f"\nFound sheets: {sheets}")
        choice = input(f"Enter sheet name [Default: {sheets[0]}]: ").strip()
        if choice in sheets:
            selected_sheet = choice
            
    df = DataLoader.load(data_path, selected_sheet)
    print(f"[OK] Loaded {len(df)} rows.")

    pptx_path = DynamicPicker.select_file("Select PowerPoint Template (.pptx)", [("PowerPoint Presentation", "*.pptx")])
    if not pptx_path or not os.path.exists(pptx_path):
        return

    prs_inspect = Presentation(pptx_path)

    default_name = f"{os.path.splitext(os.path.basename(pptx_path))[0]}_merged"
    output_name = input(f"\nEnter final output file name [Default: {default_name}]: ").strip()
    if not output_name:
        output_name = default_name
    output_name = FileHelper.sanitize_filename(output_name)

    qr_config, img_config = prompt_qr_and_image_config(df, prs_inspect, pptx_path)

    export_pdf_choice = input("\nExport as PDF in addition to PPTX? (y/n) [Default: y]: ").strip().lower()
    export_pdf = export_pdf_choice in ["", "y", "yes"]

    export_dir = FileHelper.get_export_dir(pptx_path)
    prs = Presentation(pptx_path)
    slides_per_row = len(prs.slides)
    num_rows = len(df)

    print(f"\n[*] Duplicating template slides for {num_rows} recipients ({num_rows * slides_per_row} slides total)...")
    template_slides = [prs.slides[i] for i in range(slides_per_row)]
    
    for r in tqdm(range(1, num_rows), desc="Cloning Slides"):
        for t_slide in template_slides:
            SlideDuplicator.duplicate_slide(prs, t_slide)

    print("[*] Populating placeholders, QR codes & photos in merged presentation...")
    for loop_idx, (idx, row) in enumerate(tqdm(df.iterrows(), total=num_rows, desc="Filling Data")):
        mapping = PlaceholderEngine.build_mapping(row)
        for offset in range(slides_per_row):
            target_slide = prs.slides[loop_idx * slides_per_row + offset]
            PlaceholderEngine.process_slide(target_slide, mapping)
            MultiAssetEngine.embed_all_assets(target_slide, prs, row, mapping, img_config, qr_config, export_dir)

    out_pptx = os.path.join(export_dir, f"{output_name}.pptx")
    prs.save(out_pptx)
    print(f"  [OK] Saved merged PowerPoint: {output_name}.pptx")

    if export_pdf:
        out_pdf = os.path.join(export_dir, f"{output_name}.pdf")
        converter = PPTXToPDFConverter()
        if converter.start():
            print("  [*] Converting merged presentation to master PDF...")
            converted = converter.convert(out_pptx, out_pdf)
            converter.stop()
            if converted:
                print(f"  [OK] Saved master PDF: {output_name}.pdf")
            else:
                print("  [!] PDF conversion failed.")
        else:
            print("  [!] PowerPoint COM unavailable for PDF conversion.")

    print("\n" + "=" * 60)
    print(f"  [COMPLETED] Master documents saved in: {export_dir}")
    print("=" * 60)


def run_mode_3_batch_directory():
    print("\n" + "=" * 60)
    print("  MODE 3: Batch Directory Auto-Processor (Paired XLSX + PPTX)")
    print("=" * 60)
    
    target_dir = DynamicPicker.select_directory("Select directory containing matching XLSX and PPTX files")
    if not target_dir or not os.path.isdir(target_dir):
        print("[!] No directory selected. Aborting.")
        return

    export_dir = os.path.join(target_dir, "output")
    os.makedirs(export_dir, exist_ok=True)

    files = os.listdir(target_dir)
    xlsx_files = [f for f in files if f.endswith((".xlsx", ".xls")) and not f.startswith("~") and not f.startswith("all_")]

    pairs = []
    for xf in xlsx_files:
        base = os.path.splitext(xf)[0]
        pf = f"{base}.pptx"
        if os.path.exists(os.path.join(target_dir, pf)):
            pairs.append((xf, pf, base))

    if not pairs:
        print(f"[!] No matching .xlsx and .pptx pairs found in {target_dir}")
        return

    print(f"\nFound {len(pairs)} matching pair(s):")
    for xf, pf, b in pairs:
        print(f"  - {xf} <--> {pf}")

    # Inspect first batch pair to guide QR and Image options
    sample_df = DataLoader.load(os.path.join(target_dir, pairs[0][0]))
    sample_prs = Presentation(os.path.join(target_dir, pairs[0][1]))

    qr_cfg, img_cfg = prompt_qr_and_image_config(sample_df, sample_prs, os.path.join(target_dir, pairs[0][1]))

    export_pdf_choice = input("\nAlso convert each batch to PDF? (y/n) [Default: y]: ").strip().lower()
    export_pdf = export_pdf_choice in ["", "y", "yes"]

    converter = PPTXToPDFConverter() if export_pdf else None
    if converter and not converter.start():
        print("[WARN] PowerPoint COM unavailable; proceeding with PPTX-only.")
        converter = None

    try:
        for xf, pf, base_name in tqdm(pairs, desc="Processing Batches"):
            xlsx_path = os.path.join(target_dir, xf)
            pptx_path = os.path.join(target_dir, pf)
            
            df = DataLoader.load(xlsx_path)
            if len(df) == 0:
                continue

            prs = Presentation(pptx_path)
            slides_per_row = len(prs.slides)
            template_slides = [prs.slides[i] for i in range(slides_per_row)]

            for r in range(1, len(df)):
                for t_slide in template_slides:
                    SlideDuplicator.duplicate_slide(prs, t_slide)

            for loop_idx, (idx, row) in enumerate(df.iterrows()):
                mapping = PlaceholderEngine.build_mapping(row)
                for offset in range(slides_per_row):
                    target_slide = prs.slides[loop_idx * slides_per_row + offset]
                    PlaceholderEngine.process_slide(target_slide, mapping)
                    MultiAssetEngine.embed_all_assets(target_slide, prs, row, mapping, img_cfg, qr_cfg, export_dir)

            out_pptx = os.path.join(export_dir, f"{base_name}_merged.pptx")
            prs.save(out_pptx)
            
            if converter:
                out_pdf = os.path.join(export_dir, f"{base_name}_merged.pdf")
                converter.convert(out_pptx, out_pdf)
    finally:
        if converter:
            converter.stop()

    print("\n" + "=" * 60)
    print(f"  [COMPLETED] All batch files saved in: {export_dir}")
    print("=" * 60)


def run_mode_4_pdf_merger():
    print("\n" + "=" * 60)
    print("  MODE 4: Standalone PDF Merger Utility")
    print("=" * 60)
    
    folder_path = DynamicPicker.select_directory("Select folder containing PDF files to merge")
    if not folder_path or not os.path.isdir(folder_path):
        print("[!] Invalid folder path.")
        return

    pdf_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(".pdf")]
    pdf_files.sort(key=lambda x: [int(t) if t.isdigit() else t.lower() for t in re.split(r'(\d+)', os.path.basename(x))])

    if not pdf_files:
        print("[!] No PDF files found in the selected folder.")
        return

    print(f"\nFound {len(pdf_files)} PDF file(s).")
    default_out = "Merged_Document.pdf"
    out_name = input(f"Enter merged output filename [Default: {default_out}]: ").strip()
    if not out_name:
        out_name = default_out
    if not out_name.lower().endswith(".pdf"):
        out_name += ".pdf"

    output_path = os.path.join(folder_path, out_name)
    merger = PdfWriter()

    print("[*] Merging PDF files...")
    for pdf_path in tqdm(pdf_files, desc="Merging PDFs"):
        if os.path.abspath(pdf_path) == os.path.abspath(output_path):
            continue
        merger.append(pdf_path)

    with open(output_path, "wb") as f_out:
        merger.write(f_out)
    merger.close()

    print("\n" + "=" * 60)
    print(f"  [OK] Successfully merged {len(pdf_files)} PDFs!")
    print(f"  [SAVED AS] {output_path}")
    print("=" * 60)


def run_mode_5_validator():
    print("\n" + "=" * 60)
    print("  MODE 5: Pre-Flight Template & Data Validator")
    print("=" * 60)
    
    data_path = DynamicPicker.select_file("Select Excel or CSV Data File", [("Spreadsheets", "*.xlsx;*.xls;*.csv")])
    if not data_path: return
    pptx_path = DynamicPicker.select_file("Select PowerPoint Template (.pptx)", [("PowerPoint", "*.pptx")])
    if not pptx_path: return

    sheets = DataLoader.get_sheet_names(data_path)
    selected_sheet = sheets[0] if sheets else None
    if len(sheets) > 1:
        print(f"\nFound sheets: {sheets}")
        choice = input(f"Enter sheet name or index [Default: {sheets[0]}]: ").strip()
        if choice in sheets:
            selected_sheet = choice
        elif choice.isdigit() and 1 <= int(choice) <= len(sheets):
            selected_sheet = sheets[int(choice) - 1]

    df = DataLoader.load(data_path, selected_sheet)
    prs = Presentation(pptx_path)

    template_tokens = PlaceholderEngine.extract_placeholders(prs)
    excel_cols = list(df.columns)
    excel_norm = {PlaceholderEngine.normalize_key(c): c for c in excel_cols}

    print("\n" + "-" * 50)
    print(f"  Spreadsheet Columns ({len(excel_cols)}): {excel_cols}")
    print(f"  Template Placeholders ({len(template_tokens)}): {template_tokens}")
    print("-" * 50)

    matched = []
    missing_in_data = []

    for token in template_tokens:
        tok_low = token.lower()
        if tok_low in ["qr", "image", "photo", "picture"]:
            continue
        norm_t = PlaceholderEngine.normalize_key(token)
        if token in df.columns:
            matched.append((token, token, "Exact Match"))
        elif norm_t in excel_norm:
            matched.append((token, excel_norm[norm_t], "Fuzzy Normalized Match"))
        else:
            missing_in_data.append(token)

    print(f"\n[+] Successfully Matched ({len(matched)}):")
    for t, col, match_type in matched:
        print(f"    - '{{{{{t}}}}}' --> Column '{col}' ({match_type})")

    tok_set = [t.lower() for t in template_tokens]
    if "qr" in tok_set:
        print("    - '{{qr}}' --> Dynamic QR Code Placeholder Shape (Detected!)")
    if any(t in tok_set for t in ["image", "photo", "picture"]):
        print("    - '{{image}}' / '{{photo}}' --> Dynamic Photo Placeholder Shape (Detected!)")

    if missing_in_data:
        print(f"\n[!] Missing in Spreadsheet ({len(missing_in_data)}):")
        for m in missing_in_data:
            print(f"    - '{{{{{m}}}}}' (No matching column in dataset)")
    else:
        print("\n[+] All template placeholders are 100% covered by the dataset!")


# Main Dashboard (Runs Once & Stops Immediately)
def main_dashboard_cli():
    print("\n" + "=" * 65)
    print("  *** AutoCrat Pro -- Universal Document Automation Suite ***")
    print("=" * 65)
    print("  [1] Mode 1: Individual Documents (1 PDF per Row, with QR & Photo)")
    print("  [2] Mode 2: Merged Master Document (All Rows in 1 PPTX & PDF)")
    print("  [3] Mode 3: Batch Directory Auto-Processor (Paired XLSX + PPTX)")
    print("  [4] Mode 4: Standalone PDF Merger Utility")
    print("  [5] Mode 5: Pre-Flight Template & Data Validator (Dry Run)")
    print("  [0] Exit")
    print("=" * 65)
    
    choice = input("Select an option (0-5): ").strip()
    if choice == "1":
        run_mode_1_individual()
    elif choice == "2":
        run_mode_2_merged()
    elif choice == "3":
        run_mode_3_batch_directory()
    elif choice == "4":
        run_mode_4_pdf_merger()
    elif choice == "5":
        run_mode_5_validator()
    elif choice in ["0", "q", "exit"]:
        print("\nExiting AutoCrat Pro. Have a productive day!")
    else:
        print("[!] Invalid option. Exiting.")
    return



import os
import sys
import copy
import time
import re
import threading
import concurrent.futures
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
import tkinter.font as tkfont
from typing import Optional, List, Dict, Any, Tuple
import pandas as pd
from pptx import Presentation
from PIL import Image, ImageTk

# Determine best modern font family
FONT_FAMILY = "Segoe UI"
FONT_TITLE = (FONT_FAMILY, 12, "bold")
FONT_HEADING = (FONT_FAMILY, 10, "bold")
FONT_REGULAR = (FONT_FAMILY, 9)
FONT_BOLD = (FONT_FAMILY, 9, "bold")
FONT_SMALL = (FONT_FAMILY, 8)
FONT_MONO = ("Consolas", 9)

# Google Material Icon Manager
class MaterialIconManager:
    """
    Loads, tints, and caches Google Material Icons with automatic
    Pillow vector fallback rendering.
    """
    _cache = {}
    _icons_dir = r"d:\New folder\PY\scripts\.assets\icons"
    if not os.path.isdir(_icons_dir):
        _icons_dir = r"d:\New folder\PY\.assets\icons"

    @classmethod
    def get(cls, name: str, size=(18, 18), color_hex="#FFFFFF") -> ImageTk.PhotoImage:
        key = (name, size, color_hex)
        if key in cls._cache:
            return cls._cache[key]

        icon_path = os.path.join(cls._icons_dir, f"{name}.png")
        r = int(color_hex[1:3], 16)
        g = int(color_hex[3:5], 16)
        b = int(color_hex[5:7], 16)

        if os.path.isfile(icon_path):
            try:
                base = Image.open(icon_path).convert("RGBA")
                base = base.resize(size, Image.Resampling.LANCZOS)
                alpha = base.split()[3]
                solid = Image.new("RGBA", size, (r, g, b, 255))
                solid.putalpha(alpha)
                photo = ImageTk.PhotoImage(solid)
                cls._cache[key] = photo
                return photo
            except Exception:
                pass

        # Fallback simple vector icon
        img = Image.new("RGBA", size, (0, 0, 0, 0))
        from PIL import ImageDraw
        draw = ImageDraw.Draw(img)
        draw.rectangle([2, 2, size[0] - 3, size[1] - 3], outline=(r, g, b, 200), width=2)
        photo = ImageTk.PhotoImage(img)
        cls._cache[key] = photo
        return photo


class ScrollableFrame(tk.Frame):
    """Frosted Glass Scrollable Frame for dynamic lists and tables."""
    def __init__(self, parent, bg="#162032", height=180, **kwargs):
        super().__init__(parent, bg=bg, **kwargs)
        self.canvas = tk.Canvas(self, bg=bg, bd=0, highlightthickness=0, height=height)
        self.scrollbar = tk.Scrollbar(self, orient="vertical", command=self.canvas.yview)
        self.scrollable_content = tk.Frame(self.canvas, bg=bg)

        self.scrollable_content.bind(
            "<Configure>",
            lambda e: self.canvas.configure(scrollregion=self.canvas.bbox("all"))
        )

        self.canvas_window = self.canvas.create_window((0, 0), window=self.scrollable_content, anchor="nw")
        self.canvas.configure(yscrollcommand=self.scrollbar.set)
        self.canvas.bind("<Configure>", lambda e: self.canvas.itemconfig(self.canvas_window, width=e.width))

        self.canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        self.scrollbar.pack(side=tk.RIGHT, fill=tk.Y)


class AutoCratProDashboardGUI:
    """
    Ultra-Modern Glassmorphism Desktop GUI Dashboard for AutoCrat Pro.
    Features:
      - Interactive Placeholder Mapping: Map ANY template {{placeholder}} to ANY Excel column!
      - Translucent frosted glass aesthetics with dark obsidian background
      - Modern typography hierarchy using Windows Fluent / Segoe UI Variable Display
      - Integrated Google Material Icons on cards, buttons, badges, and headers
      - Stepper navigation: Mode Selection -> Files & Data -> Columns & Layout -> Live Execution
      - Guaranteed top-level Browse dialogs that never open behind other windows
      - Interactive micro-animations with hover lighting
      - Full background multithreading with COM CoInitialize for PowerPoint PDF export
    """
    def __init__(self, initial_mode: str = "1", parent=None):
        self.root = tk.Tk() if parent is None else tk.Toplevel(parent)
        self.root.title("AutoCrat Pro -- Enterprise Document Automation Suite (by mohatamim)")
        self.root.geometry("1160x740")
        self.root.minsize(1020, 660)
        self.root.configure(bg="#0B1120")

        # Core State
        self.current_step = 1
        self.mode_var = tk.StringVar(value=str(initial_mode))
        self.data_path_var = tk.StringVar(value="")
        self.sheet_var = tk.StringVar(value="")
        self.pptx_path_var = tk.StringVar(value="")
        self.target_dir_var = tk.StringVar(value="")
        self.export_dir_var = tk.StringVar(value="")
        self.merged_name_var = tk.StringVar(value="")
        self.export_pdf_var = tk.BooleanVar(value=True)
        self.keep_pptx_var = tk.BooleanVar(value=True)

        # Batch Directory Pairs (Mode 3)
        self.batch_pairs: List[Tuple[str, str, str]] = []

        # Data, Template & Mapping State
        self.df: Optional[pd.DataFrame] = None
        self.prs: Optional[Presentation] = None
        self.template_placeholders: List[str] = []
        self.placeholder_mapping: Dict[str, str] = {}
        self.mapping_combos: Dict[str, ttk.Combobox] = {}

        self.qr_config = None
        self.img_config = None
        self.row_sel_mode_var = tk.StringVar(value="all")
        self.row_range_var = tk.StringVar(value="")
        self.selected_id_cols: List[str] = []
        self.is_running = False
        self.abort_requested = False

        self._icons = {}

        self._build_ui()
        self._set_step(1)

        # Force window to the front foreground
        try:
            self.root.lift()
            self.root.attributes("-topmost", True)
            self.root.after_idle(self.root.attributes, "-topmost", False)
            self.root.focus_force()
        except Exception:
            pass

        # Proactive preview caching for sub-50ms instant modal loading
        default_pptx = r"D:\New folder (3)\template.pptx"
        if os.path.isfile(default_pptx):
            threading.Thread(target=VisualDragDropCanvas.precache_slide_preview, args=(default_pptx,), daemon=True).start()

        # Trace text changes for immediate auto-loading when pasting file paths
        self.data_path_var.trace_add("write", lambda *args: self._on_data_path_changed())
        self.pptx_path_var.trace_add("write", lambda *args: self._on_pptx_path_changed())
        self.target_dir_var.trace_add("write", lambda *args: self._on_target_dir_changed())

    # -------------------------------------------------------------------------
    # Glassmorphism UI Architecture
    # -------------------------------------------------------------------------
    def _build_ui(self):
        # 1. Header Bar with Frosted Glass Styling
        self.hdr = tk.Frame(self.root, bg="#131C2E", padx=18, pady=12, highlightthickness=1, highlightbackground="#24344D")
        self.hdr.pack(fill=tk.X, side=tk.TOP)

        top_row = tk.Frame(self.hdr, bg="#131C2E")
        top_row.pack(fill=tk.X)

        self._icons["app_logo"] = MaterialIconManager.get("dashboard", size=(22, 22), color_hex="#38BDF8")
        tk.Label(
            top_row,
            image=self._icons["app_logo"],
            bg="#131C2E"
        ).pack(side=tk.LEFT, padx=(0, 8))

        tk.Label(
            top_row,
            text="AutoCrat Pro",
            font=(FONT_FAMILY, 13, "bold"),
            fg="#F8FAFC",
            bg="#131C2E"
        ).pack(side=tk.LEFT)

        tk.Label(
            top_row,
            text="Developed by mohatamim | Enterprise Document Automation Suite",
            font=(FONT_FAMILY, 9),
            fg="#38BDF8",
            bg="#131C2E"
        ).pack(side=tk.LEFT, padx=(12, 0))

        # Modern Stepper Breadcrumb Bar
        self.step_bar = tk.Frame(self.hdr, bg="#131C2E", pady=6)
        self.step_bar.pack(fill=tk.X, pady=(8, 0))

        self.step_pills = []
        step_definitions = [
            ("1", "Select Mode", "tune"),
            ("2", "Files & Data", "table_chart"),
            ("3", "Columns & Mapping", "layers"),
            ("4", "Live Execution", "play_arrow")
        ]

        for s_idx, (num, name, icon_name) in enumerate(step_definitions, start=1):
            pill = tk.Frame(self.step_bar, bg="#1E293B", padx=10, pady=3, highlightthickness=1, highlightbackground="#334155")
            pill.pack(side=tk.LEFT, padx=(0, 6))

            ic = MaterialIconManager.get(icon_name, size=(14, 14), color_hex="#94A3B8")
            self._icons[f"step_icon_{s_idx}"] = ic
            lbl_ic = tk.Label(pill, image=ic, bg="#1E293B")
            lbl_ic.pack(side=tk.LEFT, padx=(0, 5))

            lbl_txt = tk.Label(
                pill,
                text=f"{num}. {name}",
                font=(FONT_FAMILY, 9, "bold"),
                fg="#94A3B8",
                bg="#1E293B"
            )
            lbl_txt.pack(side=tk.LEFT)

            self.step_pills.append((pill, lbl_ic, lbl_txt, icon_name))

            if s_idx < 4:
                tk.Label(
                    self.step_bar,
                    text=">",
                    font=(FONT_FAMILY, 9, "bold"),
                    fg="#475569",
                    bg="#131C2E"
                ).pack(side=tk.LEFT, padx=(0, 6))

        # 2. Fixed Navigation Footer Bar (Frosted Glass)
        self.ftr = tk.Frame(self.root, bg="#131C2E", padx=18, pady=10, highlightthickness=1, highlightbackground="#24344D")
        self.ftr.pack(fill=tk.X, side=tk.BOTTOM)

        # Left status in footer
        self.lbl_footer_status = tk.Label(
            self.ftr,
            text="Ready. Select an operation mode to begin.",
            font=(FONT_FAMILY, 9),
            fg="#94A3B8",
            bg="#131C2E"
        )
        self.lbl_footer_status.pack(side=tk.LEFT)

        self.lbl_credit = tk.Label(
            self.ftr,
            text="Crafted by mohatamim",
            font=(FONT_FAMILY, 8, "italic"),
            fg="#64748B",
            bg="#131C2E"
        )
        self.lbl_credit.pack(side=tk.LEFT, padx=(16, 0))

        # Right Navigation Buttons
        self._icons["btn_cancel"] = MaterialIconManager.get("close", size=(15, 15), color_hex="#F8FAFC")
        self.btn_cancel = self._create_glass_button(
            self.ftr,
            text="Cancel",
            command=self._on_cancel_clicked,
            icon=self._icons["btn_cancel"],
            bg="#334155",
            hover_bg="#475569",
            fg="#F8FAFC"
        )
        self.btn_cancel.pack(side=tk.RIGHT, padx=(6, 0))

        self._icons["btn_next"] = MaterialIconManager.get("arrow_forward", size=(15, 15), color_hex="#FFFFFF")
        self.btn_next = self._create_glass_button(
            self.ftr,
            text="Next >",
            command=self._on_next_clicked,
            icon=self._icons["btn_next"],
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold"
        )
        self.btn_next.pack(side=tk.RIGHT, padx=(6, 0))

        self._icons["btn_back"] = MaterialIconManager.get("arrow_back", size=(15, 15), color_hex="#E2E8F0")
        self.btn_back = self._create_glass_button(
            self.ftr,
            text="< Back",
            command=self._on_back_clicked,
            icon=self._icons["btn_back"],
            bg="#1E293B",
            hover_bg="#334155",
            fg="#E2E8F0"
        )
        self.btn_back.pack(side=tk.RIGHT, padx=(6, 0))

        self._icons["btn_folder"] = MaterialIconManager.get("folder", size=(15, 15), color_hex="#FFFFFF")
        self.btn_open_folder = self._create_glass_button(
            self.ftr,
            text="Open Output Folder",
            command=self._open_output_folder,
            icon=self._icons["btn_folder"],
            bg="#0D9488",
            hover_bg="#14B8A6",
            fg="#FFFFFF"
        )
        self.btn_open_folder.pack(side=tk.RIGHT, padx=(0, 12))
        self.btn_open_folder.configure(state="disabled")

        # 3. Central Glass Container
        self.container = tk.Frame(self.root, bg="#0B1120", padx=18, pady=14)
        self.container.pack(fill=tk.BOTH, expand=True)

        self._build_step_1_page()
        self._build_step_2_page()
        self._build_step_3_page()
        self._build_step_4_page()

    # -------------------------------------------------------------------------
    # Glass Button Factory with Micro-Hover Animations
    # -------------------------------------------------------------------------
    def _create_glass_button(self, parent, text: str, command, icon=None, bg="#1E293B", hover_bg="#334155", fg="#F8FAFC", font_weight="normal", padx=14, pady=6):
        btn = tk.Button(
            parent,
            text=f"  {text}" if icon else text,
            image=icon,
            compound=tk.LEFT if icon else tk.NONE,
            command=command,
            font=(FONT_FAMILY, 9, font_weight),
            bg=bg,
            fg=fg,
            activebackground=hover_bg,
            activeforeground=fg,
            relief="flat",
            bd=0,
            padx=padx,
            pady=pady,
            cursor="hand2"
        )
        def on_enter(e):
            if btn["state"] != "disabled":
                btn.configure(bg=hover_bg)
        def on_leave(e):
            if btn["state"] != "disabled":
                btn.configure(bg=bg)
        btn.bind("<Enter>", on_enter)
        btn.bind("<Leave>", on_leave)
        btn._default_bg = bg
        btn._hover_bg = hover_bg
        return btn

    # -------------------------------------------------------------------------
    # Step 1 Page: Choose Operation Mode (Glassmorphism Cards)
    # -------------------------------------------------------------------------
    def _build_step_1_page(self):
        self.page1 = tk.Frame(self.container, bg="#0B1120")

        tk.Label(
            self.page1,
            text="Step 1: Choose Operation Mode",
            font=(FONT_FAMILY, 12, "bold"),
            fg="#38BDF8",
            bg="#0B1120"
        ).pack(anchor="w", pady=(0, 10))

        # Frosted Glass Card Panel
        modes_card = tk.Frame(
            self.page1,
            bg="#162032",
            padx=16,
            pady=12,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        modes_card.pack(fill=tk.X, pady=(0, 12))

        self.mode_data = [
            ("1", "Mode 1: Individual Documents (1 File per Row)", "Generates an individual personalized PowerPoint and PDF file for each row in your spreadsheet. Custom naming columns, flexible placeholder mapping, QR codes and photos.", "description", "#38BDF8"),
            ("2", "Mode 2: Single Merged Master Document (All Rows Combined)", "Creates a single comprehensive PowerPoint presentation and consolidated master PDF containing duplicated personalized slides for every recipient in your spreadsheet. Ideal for bulk printing or unified deck review.", "layers", "#818CF8"),
            ("3", "Mode 3: Batch Directory Auto-Processor (Paired XLSX + PPTX)", "Scans a folder for matching pairs of spreadsheet files (.xlsx) and templates (.pptx). Processes each batch into merged presentations and PDFs automatically in one pass.", "folder_copy", "#34D399"),
            ("4", "Mode 4: Standalone PDF Merger Tool", "Consolidates all existing PDF documents in a target folder into a single PDF with natural numerical sorting.", "picture_as_pdf", "#F87171"),
            ("5", "Mode 5: Pre-Flight Template & Data Validator (Dry Run)", "Verifies that all placeholder tags in the PowerPoint template (e.g. {{Name}}, {{ID}}) match columns in your spreadsheet.", "verified", "#FBBF24")
        ]

        self.mode_radio_frames = []
        for val, title, desc, icon_name, icon_tint in self.mode_data:
            r_frame = tk.Frame(modes_card, bg="#162032", pady=5)
            r_frame.pack(fill=tk.X, anchor="w")
            self.mode_radio_frames.append(r_frame)

            ic = MaterialIconManager.get(icon_name, size=(18, 18), color_hex=icon_tint)
            self._icons[f"mode_icon_{val}"] = ic

            header_row = tk.Frame(r_frame, bg="#162032")
            header_row.pack(anchor="w", fill=tk.X)

            tk.Label(header_row, image=ic, bg="#162032").pack(side=tk.LEFT, padx=(0, 6))

            rb = tk.Radiobutton(
                header_row,
                text=title,
                variable=self.mode_var,
                value=val,
                command=self._on_mode_radio_changed,
                font=(FONT_FAMILY, 9, "bold"),
                fg="#F8FAFC",
                bg="#162032",
                selectcolor="#0B1120",
                activebackground="#162032",
                activeforeground="#38BDF8"
            )
            rb.pack(side=tk.LEFT)

            tk.Label(
                r_frame,
                text=desc,
                font=(FONT_FAMILY, 9),
                fg="#94A3B8",
                bg="#162032",
                wraplength=940,
                justify="left"
            ).pack(anchor="w", padx=(24, 0))

        # Selected Mode Telemetry Card
        self.mode_info_card = tk.LabelFrame(
            self.page1,
            text="Selected Mode Specifications",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#38BDF8",
            bg="#162032",
            padx=14,
            pady=10,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.mode_info_card.pack(fill=tk.BOTH, expand=True)

        self.lbl_mode_details = tk.Label(
            self.mode_info_card,
            text="",
            font=(FONT_FAMILY, 9),
            fg="#E2E8F0",
            bg="#162032",
            justify="left",
            wraplength=940
        )
        self.lbl_mode_details.pack(anchor="w")

    def _on_mode_radio_changed(self):
        m = self.mode_var.get()
        info_map = {
            "1": "Inputs: Excel/CSV spreadsheet (.xlsx/.csv), PowerPoint template (.pptx).\nOutput: Individual PDFs and editable PPTX files named by chosen columns.\nFeatures: Concurrency worker engine, interactive placeholder mapping, optional QR codes, photo badge placement with custom borders.",
            "2": "Inputs: Excel/CSV spreadsheet (.xlsx/.csv), PowerPoint template (.pptx).\nOutput: One consolidated master presentation (.pptx) and master PDF (.pdf) containing all recipients.\nFeatures: Fast slide duplication, vector PDF conversion, interactive placeholder mapping, QR codes and photos embedded per slide.",
            "3": "Inputs: Target directory containing paired files (e.g. EventA.xlsx and EventA.pptx).\nOutput: Output folder with merged presentations and vector PDFs for each matching pair.\nFeatures: Automatic batch directory scanning, bulk pair execution, sample preview canvas and placeholder mapping.",
            "4": "Inputs: Target directory containing existing PDF documents.\nOutput: Single consolidated PDF merged in natural numerical order (1, 2, ... 10, 11).\nFeatures: Non-destructive merging, lossless bookmarks, fast rendering.",
            "5": "Inputs: Excel/CSV spreadsheet (.xlsx/.csv), PowerPoint template (.pptx).\nOutput: Comprehensive pre-flight verification audit report.\nFeatures: Instant tag discovery ({{tags}}), match verification, missing column warning before production."
        }
        for item in self.mode_data:
            if item[0] == m:
                title, desc = item[1], item[2]
                self.lbl_mode_details.configure(text=f"{title}\n\n{desc}\n\n{info_map.get(m, '')}")
                break

    # -------------------------------------------------------------------------
    # Step 2 Page: Files & Data Selection (Robust Top-Level Pickers)
    # -------------------------------------------------------------------------
    def _build_step_2_page(self):
        self.page2 = tk.Frame(self.container, bg="#0B1120")

        tk.Label(
            self.page2,
            text="Step 2: File & Template Selection",
            font=(FONT_FAMILY, 12, "bold"),
            fg="#38BDF8",
            bg="#0B1120"
        ).pack(anchor="w", pady=(0, 10))

        # Frosted Inputs Card
        self.files_box = tk.Frame(
            self.page2,
            bg="#162032",
            padx=16,
            pady=14,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.files_box.pack(fill=tk.X, pady=(0, 12))

        # --- A. Spreadsheet Row ---
        self.row_excel = tk.Frame(self.files_box, bg="#162032")
        self._icons["lbl_excel"] = MaterialIconManager.get("table_chart", size=(16, 16), color_hex="#38BDF8")
        tk.Label(self.row_excel, image=self._icons["lbl_excel"], bg="#162032").pack(side=tk.LEFT, padx=(0, 6))
        tk.Label(self.row_excel, text="Spreadsheet (Data):", width=18, anchor="w", fg="#F8FAFC", bg="#162032", font=(FONT_FAMILY, 9, "bold")).pack(side=tk.LEFT)

        self.ent_excel = tk.Entry(self.row_excel, textvariable=self.data_path_var, font=(FONT_FAMILY, 9), bg="#0B1120", fg="#F8FAFC", insertbackground="#F8FAFC", relief="flat", highlightthickness=1, highlightbackground="#334155")
        self.ent_excel.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=6, ipady=3)

        self._icons["btn_browse_excel"] = MaterialIconManager.get("folder_open", size=(15, 15), color_hex="#FFFFFF")
        btn_browse_data = self._create_glass_button(
            self.row_excel,
            text="Browse Data...",
            command=self._browse_data_file,
            icon=self._icons["btn_browse_excel"],
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold",
            padx=10,
            pady=4
        )
        btn_browse_data.pack(side=tk.LEFT)

        self._icons["btn_quick_load"] = MaterialIconManager.get("refresh", size=(15, 15), color_hex="#E2E8F0")
        btn_reload_data = self._create_glass_button(
            self.row_excel,
            text="Load",
            command=self._load_spreadsheet,
            icon=self._icons["btn_quick_load"],
            bg="#334155",
            hover_bg="#475569",
            fg="#E2E8F0",
            padx=8,
            pady=4
        )
        btn_reload_data.pack(side=tk.LEFT, padx=(4, 0))

        btn_inspect_data = self._create_glass_button(
            self.row_excel,
            text="Preview Table...",
            command=self._open_data_preview_modal,
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold",
            padx=10,
            pady=4
        )
        btn_inspect_data.pack(side=tk.LEFT, padx=(4, 0))

        # --- B. Worksheet Row ---
        self.row_sheet = tk.Frame(self.files_box, bg="#162032")
        tk.Label(self.row_sheet, text="", width=3, bg="#162032").pack(side=tk.LEFT)
        tk.Label(self.row_sheet, text="Worksheet:", width=18, anchor="w", fg="#F8FAFC", bg="#162032", font=(FONT_FAMILY, 9, "bold")).pack(side=tk.LEFT)
        self.cb_sheet = ttk.Combobox(self.row_sheet, textvariable=self.sheet_var, font=(FONT_FAMILY, 9), state="readonly")
        self.cb_sheet.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=6)
        self.cb_sheet.bind("<<ComboboxSelected>>", self._on_sheet_selected)

        # Data summary badge
        self.lbl_data_badge = tk.Label(self.files_box, text="[No spreadsheet loaded]", font=(FONT_FAMILY, 9), fg="#94A3B8", bg="#162032")

        # --- C. PPTX Template Row ---
        self.row_pptx = tk.Frame(self.files_box, bg="#162032")
        self._icons["lbl_pptx"] = MaterialIconManager.get("slideshow", size=(16, 16), color_hex="#F97316")
        tk.Label(self.row_pptx, image=self._icons["lbl_pptx"], bg="#162032").pack(side=tk.LEFT, padx=(0, 6))
        tk.Label(self.row_pptx, text="PPTX Template:", width=18, anchor="w", fg="#F8FAFC", bg="#162032", font=(FONT_FAMILY, 9, "bold")).pack(side=tk.LEFT)

        self.ent_pptx = tk.Entry(self.row_pptx, textvariable=self.pptx_path_var, font=(FONT_FAMILY, 9), bg="#0B1120", fg="#F8FAFC", insertbackground="#F8FAFC", relief="flat", highlightthickness=1, highlightbackground="#334155")
        self.ent_pptx.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=6, ipady=3)

        btn_browse_pptx = self._create_glass_button(
            self.row_pptx,
            text="Browse PPTX...",
            command=self._browse_pptx_file,
            icon=self._icons["btn_browse_excel"],
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold",
            padx=10,
            pady=4
        )
        btn_browse_pptx.pack(side=tk.LEFT)

        btn_reload_pptx = self._create_glass_button(
            self.row_pptx,
            text="Load",
            command=self._load_pptx_template,
            icon=self._icons["btn_quick_load"],
            bg="#334155",
            hover_bg="#475569",
            fg="#E2E8F0",
            padx=8,
            pady=4
        )
        btn_reload_pptx.pack(side=tk.LEFT, padx=(4, 0))

        # PPTX summary badge
        self.lbl_pptx_badge = tk.Label(self.files_box, text="[No template loaded]", font=(FONT_FAMILY, 9), fg="#94A3B8", bg="#162032")

        # --- D. Batch Folder Row (Mode 3 & Mode 4) ---
        self.row_folder = tk.Frame(self.files_box, bg="#162032")
        self._icons["lbl_folder"] = MaterialIconManager.get("folder_copy", size=(16, 16), color_hex="#34D399")
        tk.Label(self.row_folder, image=self._icons["lbl_folder"], bg="#162032").pack(side=tk.LEFT, padx=(0, 6))
        self.lbl_folder_tag = tk.Label(self.row_folder, text="Target Directory:", width=18, anchor="w", fg="#F8FAFC", bg="#162032", font=(FONT_FAMILY, 9, "bold"))
        self.lbl_folder_tag.pack(side=tk.LEFT)

        self.ent_folder = tk.Entry(self.row_folder, textvariable=self.target_dir_var, font=(FONT_FAMILY, 9), bg="#0B1120", fg="#F8FAFC", insertbackground="#F8FAFC", relief="flat", highlightthickness=1, highlightbackground="#334155")
        self.ent_folder.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=6, ipady=3)

        btn_browse_fld = self._create_glass_button(
            self.row_folder,
            text="Browse Folder...",
            command=self._browse_target_folder,
            icon=self._icons["btn_browse_excel"],
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold",
            padx=10,
            pady=4
        )
        btn_browse_fld.pack(side=tk.LEFT)

        btn_reload_fld = self._create_glass_button(
            self.row_folder,
            text="Scan",
            command=lambda: self._on_target_dir_changed(),
            icon=self._icons["btn_quick_load"],
            bg="#334155",
            hover_bg="#475569",
            fg="#E2E8F0",
            padx=8,
            pady=4
        )
        btn_reload_fld.pack(side=tk.LEFT, padx=(4, 0))

        # Batch Directory Pairs List / Telemetry Card
        self.batch_list_card = tk.LabelFrame(
            self.page2,
            text="Batch Directory Pairs Detected",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#38BDF8",
            bg="#162032",
            padx=12,
            pady=10,
            highlightthickness=1,
            highlightbackground="#24344D"
        )

        self.txt_batch_info = tk.Text(
            self.batch_list_card,
            font=("Consolas", 8),
            bg="#0B1120",
            fg="#E2E8F0",
            height=7,
            relief="flat",
            wrap="word",
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.txt_batch_info.pack(fill=tk.BOTH, expand=True)

    def _refresh_step_2_layout(self):
        m = self.mode_var.get()
        if m in ["1", "2", "5"]:
            self.row_excel.pack(fill=tk.X, pady=(0, 6))
            self.row_sheet.pack(fill=tk.X, pady=(0, 6))
            self.lbl_data_badge.pack(anchor="w", pady=(0, 6))
            self.row_pptx.pack(fill=tk.X, pady=(0, 6))
            self.lbl_pptx_badge.pack(anchor="w", pady=(0, 4))
            self.row_folder.pack_forget()
            self.batch_list_card.pack_forget()
        elif m == "3":
            self.row_excel.pack_forget()
            self.row_sheet.pack_forget()
            self.lbl_data_badge.pack_forget()
            self.row_pptx.pack_forget()
            self.lbl_pptx_badge.pack_forget()
            self.lbl_folder_tag.configure(text="Batch Directory:")
            self.row_folder.pack(fill=tk.X, pady=(0, 6))
            self.batch_list_card.pack(fill=tk.BOTH, expand=True, pady=(6, 0))
            if self.target_dir_var.get():
                self._scan_batch_directory(self.target_dir_var.get())
        elif m == "4":
            self.row_excel.pack_forget()
            self.row_sheet.pack_forget()
            self.lbl_data_badge.pack_forget()
            self.row_pptx.pack_forget()
            self.lbl_pptx_badge.pack_forget()
            self.lbl_folder_tag.configure(text="PDF Source Folder:")
            self.row_folder.pack(fill=tk.X, pady=(0, 6))
            self.batch_list_card.pack(fill=tk.BOTH, expand=True, pady=(6, 0))
            if self.target_dir_var.get():
                self._scan_pdf_folder(self.target_dir_var.get())

    # -------------------------------------------------------------------------
    # Safe Top-Level File Browsing Handlers (Guaranteed Front Window)
    # -------------------------------------------------------------------------
    def _browse_data_file(self):
        try:
            self.root.update_idletasks()
            self.root.lift()
            self.root.attributes("-topmost", True)
            f = filedialog.askopenfilename(
                parent=self.root,
                title="Select Excel or CSV Spreadsheet",
                filetypes=[
                    ("All Supported Spreadsheets (*.xlsx, *.xls, *.csv)", ("*.xlsx", "*.xls", "*.csv")),
                    ("Excel Workbooks (*.xlsx)", "*.xlsx"),
                    ("Legacy Excel (*.xls)", "*.xls"),
                    ("CSV Comma Delimited (*.csv)", "*.csv"),
                    ("All Files (*.*)", "*.*")
                ]
            )
            self.root.attributes("-topmost", False)
            self.root.lift()
            self.root.focus_force()
            if f:
                self.data_path_var.set(f)
                self._load_spreadsheet()
        except Exception as e:
            messagebox.showerror("File Dialog Error", f"Could not open file picker dialog:\n{e}")

    def _browse_pptx_file(self):
        try:
            self.root.update_idletasks()
            self.root.lift()
            self.root.attributes("-topmost", True)
            f = filedialog.askopenfilename(
                parent=self.root,
                title="Select PowerPoint Template",
                filetypes=[
                    ("PowerPoint Presentation (*.pptx)", "*.pptx"),
                    ("All Files (*.*)", "*.*")
                ]
            )
            self.root.attributes("-topmost", False)
            self.root.lift()
            self.root.focus_force()
            if f:
                self.pptx_path_var.set(f)
                self._load_pptx_template()
        except Exception as e:
            messagebox.showerror("File Dialog Error", f"Could not open template picker dialog:\n{e}")

    def _browse_target_folder(self):
        try:
            m = self.mode_var.get()
            title = "Select Batch Directory (with matching XLSX & PPTX)" if m == "3" else "Select PDF Folder"
            self.root.update_idletasks()
            self.root.lift()
            self.root.attributes("-topmost", True)
            f = filedialog.askdirectory(parent=self.root, title=title)
            self.root.attributes("-topmost", False)
            self.root.lift()
            self.root.focus_force()
            if f:
                self.target_dir_var.set(f)
                if m == "3":
                    self._scan_batch_directory(f)
                elif m == "4":
                    self._scan_pdf_folder(f)
        except Exception as e:
            messagebox.showerror("Folder Dialog Error", f"Could not open folder picker dialog:\n{e}")

    def _on_data_path_changed(self):
        p = self.data_path_var.get().strip()
        if p and os.path.isfile(p):
            self._load_spreadsheet()

    def _on_pptx_path_changed(self):
        p = self.pptx_path_var.get().strip()
        if p and os.path.isfile(p):
            self._load_pptx_template()

    def _on_target_dir_changed(self):
        p = self.target_dir_var.get().strip()
        if p and os.path.isdir(p):
            m = self.mode_var.get()
            if m == "3":
                self._scan_batch_directory(p)
            elif m == "4":
                self._scan_pdf_folder(p)

    def _load_spreadsheet(self):
        path = self.data_path_var.get().strip()
        if not path or not os.path.isfile(path):
            return
        try:
            if "DataLoader" in globals():
                sheets = DataLoader.get_sheet_names(path)
            else:
                import openpyxl
                wb = openpyxl.load_workbook(path, read_only=True)
                sheets = wb.sheetnames
                wb.close()

            self.cb_sheet["values"] = sheets
            if sheets:
                self.sheet_var.set(sheets[0])
                self._load_sheet_data(path, sheets[0])
        except Exception as e:
            self.lbl_data_badge.configure(text=f"[Error reading spreadsheet: {e}]", fg="#F87171")

    def _on_sheet_selected(self, event=None):
        path = self.data_path_var.get().strip()
        sh = self.sheet_var.get().strip()
        if path and sh:
            self._load_sheet_data(path, sh)

    def _load_sheet_data(self, path: str, sheet: str):
        try:
            if "DataLoader" in globals():
                self.df = DataLoader.load(path, sheet)
            elif path.endswith(".csv"):
                self.df = pd.read_csv(path)
            else:
                self.df = pd.read_excel(path, sheet_name=sheet)

            rows = len(self.df)
            cols = list(self.df.columns)
            self.lbl_data_badge.configure(
                text=f"[OK] Loaded {rows} row(s) | {len(cols)} column(s): {', '.join(cols[:5])}{'...' if len(cols) > 5 else ''}",
                fg="#34D399"
            )
            self._populate_column_listbox()
            self._auto_match_all_placeholders()
        except Exception as e:
            self.lbl_data_badge.configure(text=f"[Failed reading sheet: {e}]", fg="#F87171")

    def _load_pptx_template(self):
        path = self.pptx_path_var.get().strip()
        if not path or not os.path.isfile(path):
            return
        try:
            self.prs = Presentation(path)
            slides_cnt = len(self.prs.slides)
            if "PlaceholderEngine" in globals():
                self.template_placeholders = sorted(list(PlaceholderEngine.extract_placeholders(self.prs)))
            else:
                tags = set()
                for s in self.prs.slides:
                    for shp in s.shapes:
                        if shp.has_text_frame:
                            found = re.findall(r"\{\{([^{}]+)\}\}", shp.text_frame.text)
                            for t in found: tags.add(t.strip())
                self.template_placeholders = sorted(list(tags))

            tag_str = f" | Tags: {', '.join(self.template_placeholders)}" if self.template_placeholders else " | No {{tags}} found"
            self.lbl_pptx_badge.configure(
                text=f"[OK] Template loaded: {slides_cnt} slide(s){tag_str}",
                fg="#34D399"
            )
            self._auto_match_all_placeholders()
            self._init_default_asset_configs()
            threading.Thread(target=VisualDragDropCanvas.precache_slide_preview, args=(path,), daemon=True).start()
        except Exception as e:
            self.lbl_pptx_badge.configure(text=f"[Error loading template: {e}]", fg="#F87171")

    def _scan_batch_directory(self, folder: str):
        self.txt_batch_info.delete("1.0", tk.END)
        self.batch_pairs = []
        if not os.path.isdir(folder):
            return

        files = os.listdir(folder)
        xlsx_files = [f for f in files if f.endswith((".xlsx", ".xls")) and not f.startswith("~") and not f.startswith("all_")]

        for xf in xlsx_files:
            base = os.path.splitext(xf)[0]
            pf = f"{base}.pptx"
            if os.path.exists(os.path.join(folder, pf)):
                self.batch_pairs.append((xf, pf, base))

        if not self.batch_pairs:
            self.txt_batch_info.insert(tk.END, f"[!] No matching .xlsx and .pptx pairs found in: {folder}\n")
            self.txt_batch_info.insert(tk.END, "Ensure your spreadsheets and templates have matching base names (e.g. Event1.xlsx and Event1.pptx).\n")
            self.prs = None
            self.df = None
            return

        self.txt_batch_info.insert(tk.END, f"[OK] Found {len(self.batch_pairs)} matching batch pair(s):\n")
        for xf, pf, b in self.batch_pairs:
            self.txt_batch_info.insert(tk.END, f"  * Batch '{b}': {xf} <--> {pf}\n")

        # Automatically load the first batch pair as sample for visual preview and canvas designer!
        first_x, first_p, _ = self.batch_pairs[0]
        try:
            sample_df_path = os.path.join(folder, first_x)
            sample_pptx_path = os.path.join(folder, first_p)
            if "DataLoader" in globals():
                self.df = DataLoader.load(sample_df_path)
            else:
                self.df = pd.read_excel(sample_df_path)

            self.prs = Presentation(sample_pptx_path)
            self.pptx_path_var.set(sample_pptx_path)
            if "PlaceholderEngine" in globals():
                self.template_placeholders = sorted(list(PlaceholderEngine.extract_placeholders(self.prs)))
            self._auto_match_all_placeholders()
            self.txt_batch_info.insert(tk.END, f"\n[OK] Loaded sample template '{first_p}' ({len(self.df)} rows) for Visual Canvas & Mapping.\n")
        except Exception as e:
            self.txt_batch_info.insert(tk.END, f"\n[WARN] Could not load sample pair: {e}\n")

    def _scan_pdf_folder(self, folder: str):
        self.txt_batch_info.delete("1.0", tk.END)
        if not os.path.isdir(folder):
            return
        pdfs = [f for f in os.listdir(folder) if f.lower().endswith(".pdf")]
        self.txt_batch_info.insert(tk.END, f"[OK] Found {len(pdfs)} PDF file(s) in: {folder}\n\n")
        for p in pdfs[:15]:
            self.txt_batch_info.insert(tk.END, f"  * {p}\n")
        if len(pdfs) > 15:
            self.txt_batch_info.insert(tk.END, f"  ... and {len(pdfs) - 15} more files.\n")

    # -------------------------------------------------------------------------
    # Smart Fuzzy Placeholder Matching Algorithm
    # -------------------------------------------------------------------------
    def fuzzy_match_column(self, tag: str, columns: List[str]) -> str:
        """Smart fuzzy matching: exact -> case-insensitive -> normalized -> word token -> substring."""
        if not columns:
            return ""

        norm_tag = re.sub(r"[\s_]+", "", tag).lower()

        # 1. Exact match
        if tag in columns:
            return tag

        # 2. Case-insensitive exact match
        for col in columns:
            if col.lower() == tag.lower():
                return col

        # 3. Normalized exact match (ignoring whitespace and underscores)
        for col in columns:
            if re.sub(r"[\s_]+", "", col).lower() == norm_tag:
                return col

        # 4. Whole-word / token match (e.g. 'ID' in 'student_id')
        for col in columns:
            tokens = [t.lower() for t in re.split(r"[\s_]+", col)]
            if tag.lower() in tokens:
                return col

        # 5. Substring match only if length >= 4
        if len(tag) >= 4:
            for col in columns:
                if tag.lower() in col.lower() or col.lower() in tag.lower():
                    return col

        return ""

    def _auto_match_all_placeholders(self):
        """Automatically matches all template placeholders to the best spreadsheet column."""
        cols = list(self.df.columns) if self.df is not None else []
        for tag in self.template_placeholders:
            best = self.fuzzy_match_column(tag, cols)
            self.placeholder_mapping[tag] = best
            if tag in self.mapping_combos:
                self.mapping_combos[tag].set(best or "(None / Skip)")
        self._update_mapping_status_badge()

    def _clear_all_placeholder_mappings(self):
        """Clears all placeholder mappings."""
        for tag in self.template_placeholders:
            self.placeholder_mapping[tag] = ""
            if tag in self.mapping_combos:
                self.mapping_combos[tag].set("(None / Skip)")
        self._update_mapping_status_badge()

    def _on_mapping_combo_changed(self, tag: str, event=None):
        """Handles user selecting a different column from the mapping dropdown."""
        cb = self.mapping_combos.get(tag)
        if cb:
            val = cb.get()
            if val in ["(None / Skip)", "", "-- Select Column --"]:
                self.placeholder_mapping[tag] = ""
            else:
                self.placeholder_mapping[tag] = val
        self._update_mapping_status_badge()

    def _update_mapping_status_badge(self):
        total = len(self.template_placeholders)
        if total == 0:
            self.lbl_mapping_summary.configure(text="No {{placeholders}} detected in template.", fg="#94A3B8")
            return

        mapped_count = sum(1 for tag in self.template_placeholders if self.placeholder_mapping.get(tag))
        if mapped_count == total:
            self.lbl_mapping_summary.configure(
                text=f"[OK] All {total} placeholder(s) mapped to spreadsheet columns.",
                fg="#34D399"
            )
        else:
            self.lbl_mapping_summary.configure(
                text=f"[{mapped_count} of {total} placeholders mapped ({total - mapped_count} unmapped)]",
                fg="#FBBF24"
            )

    # -------------------------------------------------------------------------
    # Step 3 Page: Column Selection, Output Options & Placeholder Mapping
    # -------------------------------------------------------------------------
    def _build_step_3_page(self):
        self.page3 = tk.Frame(self.container, bg="#0B1120")

        tk.Label(
            self.page3,
            text="Step 3: Column Selection, Placeholder Mapping & Output Options",
            font=(FONT_FAMILY, 12, "bold"),
            fg="#38BDF8",
            bg="#0B1120"
        ).pack(anchor="w", pady=(0, 10))

        content_frame = tk.Frame(self.page3, bg="#0B1120")
        content_frame.pack(fill=tk.BOTH, expand=True)

        # Left Column Card: Mode Settings & Output Naming
        self.step3_left = tk.Frame(
            content_frame,
            bg="#162032",
            padx=16,
            pady=12,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.step3_left.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=(0, 10))

        # --- Mode 1 Specific: Column Multi-Select for Filename ---
        self.frame_mode1_cols = tk.Frame(self.step3_left, bg="#162032")
        tk.Label(
            self.frame_mode1_cols,
            text="Select Column(s) to Name Output Files:",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        ).pack(anchor="w", pady=(0, 4))

        btn_row_cols = tk.Frame(self.frame_mode1_cols, bg="#162032")
        btn_row_cols.pack(fill=tk.X, pady=(0, 4))

        self._create_glass_button(
            btn_row_cols,
            text="Select All",
            command=self._select_all_cols,
            bg="#334155",
            hover_bg="#475569",
            fg="#F8FAFC",
            padx=8,
            pady=3
        ).pack(side=tk.LEFT)

        self._create_glass_button(
            btn_row_cols,
            text="Clear",
            command=self._clear_all_cols,
            bg="#334155",
            hover_bg="#475569",
            fg="#F8FAFC",
            padx=8,
            pady=3
        ).pack(side=tk.LEFT, padx=4)

        list_scroll = tk.Frame(self.frame_mode1_cols, bg="#162032")
        list_scroll.pack(fill=tk.BOTH, expand=True)
        self.lb_cols = tk.Listbox(
            list_scroll,
            selectmode=tk.MULTIPLE,
            font=(FONT_FAMILY, 9),
            bg="#0B1120",
            fg="#F8FAFC",
            selectbackground="#0284C7",
            selectforeground="#FFFFFF",
            relief="flat",
            highlightthickness=1,
            highlightbackground="#334155",
            height=6
        )
        sb_cols = tk.Scrollbar(list_scroll, orient=tk.VERTICAL, command=self.lb_cols.yview)
        self.lb_cols.configure(yscrollcommand=sb_cols.set)
        self.lb_cols.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        sb_cols.pack(side=tk.RIGHT, fill=tk.Y)
        self.lb_cols.bind("<<ListboxSelect>>", self._on_col_selected)

        self.lbl_filename_preview = tk.Label(
            self.frame_mode1_cols,
            text="Filename Preview: [Row Index].pdf",
            font=(FONT_FAMILY, 8, "italic"),
            fg="#38BDF8",
            bg="#162032"
        )
        self.lbl_filename_preview.pack(anchor="w", pady=(4, 0))

        # --- Mode 2 Specific: Merged Output Filename ---
        self.frame_mode2_opts = tk.Frame(self.step3_left, bg="#162032")
        tk.Label(
            self.frame_mode2_opts,
            text="Merged Master Presentation File Name:",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        ).pack(anchor="w", pady=(0, 4))

        self.ent_merged_name = tk.Entry(
            self.frame_mode2_opts,
            textvariable=self.merged_name_var,
            font=(FONT_FAMILY, 9),
            bg="#0B1120",
            fg="#F8FAFC",
            insertbackground="#F8FAFC",
            relief="flat",
            highlightthickness=1,
            highlightbackground="#334155"
        )
        self.ent_merged_name.pack(fill=tk.X, pady=(0, 6), ipady=3)

        # --- Mode 4 Specific: Consolidated PDF Filename ---
        self.frame_mode4_opts = tk.Frame(self.step3_left, bg="#162032")
        tk.Label(
            self.frame_mode4_opts,
            text="Consolidated PDF Filename:",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        ).pack(anchor="w", pady=(0, 4))

        self.ent_pdf_name = tk.Entry(
            self.frame_mode4_opts,
            textvariable=self.merged_name_var,
            font=(FONT_FAMILY, 9),
            bg="#0B1120",
            fg="#F8FAFC",
            insertbackground="#F8FAFC",
            relief="flat",
            highlightthickness=1,
            highlightbackground="#334155"
        )
        self.ent_pdf_name.pack(fill=tk.X, pady=(0, 6), ipady=3)

        # --- Data Row Selection Card ---
        self.card_row_selection = tk.Frame(
            self.step3_left,
            bg="#162032",
            padx=10,
            pady=8,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.card_row_selection.pack(fill=tk.X, pady=(0, 8))

        row_sel_hdr = tk.Frame(self.card_row_selection, bg="#162032")
        row_sel_hdr.pack(fill=tk.X, pady=(0, 4))
        self._icons["row_sel_icon"] = MaterialIconManager.get("table_rows", size=(15, 15), color_hex="#38BDF8")
        tk.Label(row_sel_hdr, image=self._icons["row_sel_icon"], bg="#162032").pack(side=tk.LEFT, padx=(0, 4))
        tk.Label(
            row_sel_hdr,
            text="Data Row Selection",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        ).pack(side=tk.LEFT)

        row_radio_bar = tk.Frame(self.card_row_selection, bg="#162032")
        row_radio_bar.pack(fill=tk.X, pady=(2, 4))

        self.rb_row_all = tk.Radiobutton(
            row_radio_bar,
            text="All Rows",
            variable=self.row_sel_mode_var,
            value="all",
            command=self._on_row_mode_changed,
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032",
            selectcolor="#0B1120",
            activebackground="#162032"
        )
        self.rb_row_all.pack(side=tk.LEFT, padx=(0, 8))

        self.rb_row_spec = tk.Radiobutton(
            row_radio_bar,
            text="Specific Rows / Range",
            variable=self.row_sel_mode_var,
            value="specific",
            command=self._on_row_mode_changed,
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032",
            selectcolor="#0B1120",
            activebackground="#162032"
        )
        self.rb_row_spec.pack(side=tk.LEFT)

        self.frame_row_input = tk.Frame(self.card_row_selection, bg="#162032")
        self.frame_row_input.pack(fill=tk.X, pady=(2, 2))

        self.ent_row_range = tk.Entry(
            self.frame_row_input,
            textvariable=self.row_range_var,
            font=(FONT_FAMILY, 9),
            bg="#0B1120",
            fg="#F8FAFC",
            insertbackground="#F8FAFC",
            relief="flat",
            highlightthickness=1,
            highlightbackground="#334155"
        )
        self.ent_row_range.pack(side=tk.LEFT, fill=tk.X, expand=True, ipady=2)
        self.ent_row_range.configure(state="disabled")
        self.row_range_var.trace_add("write", lambda *args: self._update_row_selection_summary())

        self.lbl_row_summary = tk.Label(
            self.card_row_selection,
            text="All rows selected.",
            font=(FONT_FAMILY, 9),
            fg="#34D399",
            bg="#162032"
        )
        self.lbl_row_summary.pack(anchor="w", pady=(2, 0))

        tk.Label(
            self.card_row_selection,
            text="Example: 1-5, 8, 10-12 (1-based row numbers)",
            font=(FONT_FAMILY, 8),
            fg="#64748B",
            bg="#162032"
        ).pack(anchor="w")

        # --- Common Output Format Checkboxes ---
        self.frame_format_opts = tk.Frame(self.step3_left, bg="#162032")
        self.frame_format_opts.pack(fill=tk.X, pady=(8, 0))

        self.chk_pdf = tk.Checkbutton(
            self.frame_format_opts,
            text="Export High-Quality Vector PDF (via PowerPoint COM)",
            variable=self.export_pdf_var,
            font=(FONT_FAMILY, 9),
            fg="#F8FAFC",
            bg="#162032",
            selectcolor="#0B1120",
            activebackground="#162032"
        )
        self.chk_pdf.pack(anchor="w")

        self.chk_keep_pptx = tk.Checkbutton(
            self.frame_format_opts,
            text="Keep Editable PowerPoint (.pptx) Presentations",
            variable=self.keep_pptx_var,
            font=(FONT_FAMILY, 9),
            fg="#F8FAFC",
            bg="#162032",
            selectcolor="#0B1120",
            activebackground="#162032"
        )
        self.chk_keep_pptx.pack(anchor="w")

        # Right Column Card: Placeholder Mapping & Visual Designer
        self.step3_right = tk.Frame(
            content_frame,
            bg="#162032",
            padx=16,
            pady=12,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        self.step3_right.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)

        # --- A. INTERACTIVE PLACEHOLDER MAPPING CARD ---
        self.card_placeholder_mapping = tk.Frame(self.step3_right, bg="#162032")
        self.card_placeholder_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 10))

        map_hdr = tk.Frame(self.card_placeholder_mapping, bg="#162032")
        map_hdr.pack(fill=tk.X, pady=(0, 4))

        self._icons["map_title_icon"] = MaterialIconManager.get("layers", size=(16, 16), color_hex="#38BDF8")
        tk.Label(map_hdr, image=self._icons["map_title_icon"], bg="#162032").pack(side=tk.LEFT, padx=(0, 6))

        tk.Label(
            map_hdr,
            text="Interactive Placeholder Mapping",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        ).pack(side=tk.LEFT)

        self._create_glass_button(
            map_hdr,
            text="Auto-Match All",
            command=self._auto_match_all_placeholders,
            bg="#334155",
            hover_bg="#475569",
            fg="#F8FAFC",
            padx=8,
            pady=2
        ).pack(side=tk.RIGHT)

        self._create_glass_button(
            map_hdr,
            text="Clear",
            command=self._clear_all_placeholder_mappings,
            bg="#334155",
            hover_bg="#475569",
            fg="#F8FAFC",
            padx=8,
            pady=2
        ).pack(side=tk.RIGHT, padx=4)

        tk.Label(
            self.card_placeholder_mapping,
            text="Map any template {{tag}} to any spreadsheet column. Exact name match is NOT required!",
            font=(FONT_FAMILY, 9),
            fg="#94A3B8",
            bg="#162032"
        ).pack(anchor="w", pady=(0, 6))

        # Scrollable mapping rows container
        self.scroll_mapping = ScrollableFrame(self.card_placeholder_mapping, bg="#0B1120", height=160)
        self.scroll_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 4))

        self.lbl_mapping_summary = tk.Label(
            self.card_placeholder_mapping,
            text="No placeholders detected.",
            font=(FONT_FAMILY, 9),
            fg="#94A3B8",
            bg="#162032"
        )
        self.lbl_mapping_summary.pack(anchor="w")

        # --- B. VISUAL DESIGNER BUTTON & LAYOUT SUMMARY ---
        self.frame_designer_section = tk.Frame(self.step3_right, bg="#162032")
        self.frame_designer_section.pack(fill=tk.X, pady=(6, 0))

        self._icons["btn_canvas"] = MaterialIconManager.get("tune", size=(16, 16), color_hex="#FFFFFF")
        self.btn_canvas = self._create_glass_button(
            self.frame_designer_section,
            text="Open Visual Layout & Style Designer...",
            command=self._launch_canvas_editor,
            icon=self._icons["btn_canvas"],
            bg="#0284C7",
            hover_bg="#0369A1",
            fg="#FFFFFF",
            font_weight="bold",
            pady=6
        )
        self.btn_canvas.pack(fill=tk.X, pady=(0, 6))

        # Layout Summary Badges
        layout_badges_frame = tk.Frame(self.frame_designer_section, bg="#162032")
        layout_badges_frame.pack(fill=tk.X)

        self._icons["badge_qr"] = MaterialIconManager.get("qr_code", size=(14, 14), color_hex="#94A3B8")
        row_qr = tk.Frame(layout_badges_frame, bg="#162032")
        row_qr.pack(side=tk.LEFT, fill=tk.X, expand=True)
        tk.Label(row_qr, image=self._icons["badge_qr"], bg="#162032").pack(side=tk.LEFT, padx=(0, 4))
        self.lbl_qr_summary = tk.Label(row_qr, text="QR Code: Inactive", font=(FONT_FAMILY, 9), fg="#94A3B8", bg="#162032")
        self.lbl_qr_summary.pack(side=tk.LEFT)

        self._icons["badge_img"] = MaterialIconManager.get("image", size=(14, 14), color_hex="#94A3B8")
        row_img = tk.Frame(layout_badges_frame, bg="#162032")
        row_img.pack(side=tk.LEFT, fill=tk.X, expand=True)
        tk.Label(row_img, image=self._icons["badge_img"], bg="#162032").pack(side=tk.LEFT, padx=(0, 4))
        self.lbl_img_summary = tk.Label(row_img, text="Image Asset: Inactive", font=(FONT_FAMILY, 9), fg="#94A3B8", bg="#162032")
        self.lbl_img_summary.pack(side=tk.LEFT)

    def _refresh_step_3_layout(self):
        m = self.mode_var.get()
        if m == "1":
            self.frame_mode1_cols.pack(fill=tk.BOTH, expand=True)
            self.frame_mode2_opts.pack_forget()
            self.frame_mode4_opts.pack_forget()
            self.chk_keep_pptx.pack(anchor="w")
            self.chk_pdf.pack(anchor="w")
            self.card_placeholder_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 10))
            self.btn_canvas.configure(state="normal")
            self._populate_column_listbox()
            self._rebuild_placeholder_mapping_rows()
        elif m == "2":
            self.frame_mode1_cols.pack_forget()
            self.frame_mode2_opts.pack(fill=tk.X)
            self.frame_mode4_opts.pack_forget()
            self.chk_keep_pptx.pack_forget()
            self.chk_pdf.pack(anchor="w")
            self.card_placeholder_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 10))
            self.btn_canvas.configure(state="normal")
            if not self.merged_name_var.get().strip():
                pptx_p = self.pptx_path_var.get().strip()
                if pptx_p:
                    base = os.path.splitext(os.path.basename(pptx_p))[0]
                    self.merged_name_var.set(f"{base}_merged")
                else:
                    self.merged_name_var.set("Master_Merged_Presentation")
            self._rebuild_placeholder_mapping_rows()
        elif m == "3":
            self.frame_mode1_cols.pack_forget()
            self.frame_mode2_opts.pack_forget()
            self.frame_mode4_opts.pack_forget()
            self.chk_keep_pptx.pack_forget()
            self.chk_pdf.pack(anchor="w")
            self.card_placeholder_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 10))
            self.btn_canvas.configure(state="normal" if self.prs is not None else "disabled")
            self._rebuild_placeholder_mapping_rows()
        elif m == "4":
            self.frame_mode1_cols.pack_forget()
            self.frame_mode2_opts.pack_forget()
            self.frame_mode4_opts.pack(fill=tk.X)
            self.chk_keep_pptx.pack_forget()
            self.chk_pdf.pack_forget()
            self.card_placeholder_mapping.pack_forget()
            self.btn_canvas.configure(state="disabled")
            if not self.merged_name_var.get().strip():
                self.merged_name_var.set("Consolidated_Master.pdf")
        elif m == "5":
            self.frame_mode1_cols.pack_forget()
            self.frame_mode2_opts.pack_forget()
            self.frame_mode4_opts.pack_forget()
            self.chk_keep_pptx.pack_forget()
            self.chk_pdf.pack_forget()
            self.card_placeholder_mapping.pack(fill=tk.BOTH, expand=True, pady=(0, 10))
            self.btn_canvas.configure(state="disabled")
            self._rebuild_placeholder_mapping_rows()

    def _rebuild_placeholder_mapping_rows(self):
        """Builds interactive dropdown mapping rows for every detected placeholder."""
        for child in self.scroll_mapping.scrollable_content.winfo_children():
            child.destroy()
        self.mapping_combos.clear()

        cols = list(self.df.columns) if self.df is not None else []
        options = ["(None / Skip)"] + cols

        if not self.template_placeholders:
            tk.Label(
                self.scroll_mapping.scrollable_content,
                text="No {{placeholders}} found in PowerPoint template.\nStandard column names will be used.",
                font=(FONT_FAMILY, 9),
                fg="#94A3B8",
                bg="#0B1120",
                pady=10
            ).pack(anchor="w")
            self._update_mapping_status_badge()
            return

        for tag in self.template_placeholders:
            row_frame = tk.Frame(self.scroll_mapping.scrollable_content, bg="#0B1120", pady=3)
            row_frame.pack(fill=tk.X, padx=4)

            # Left tag badge
            tk.Label(
                row_frame,
                text=f"{{{{{tag}}}}}",
                font=(FONT_FAMILY, 9, "bold"),
                fg="#38BDF8",
                bg="#162032",
                padx=6,
                pady=2,
                width=18,
                anchor="w"
            ).pack(side=tk.LEFT)

            tk.Label(row_frame, text="-->", fg="#64748B", bg="#0B1120", font=(FONT_FAMILY, 9, "bold")).pack(side=tk.LEFT, padx=4)

            cb = ttk.Combobox(row_frame, values=options, state="readonly", font=(FONT_FAMILY, 9))
            current_mapped = self.placeholder_mapping.get(tag, "")
            if not current_mapped:
                current_mapped = self.fuzzy_match_column(tag, cols)
                self.placeholder_mapping[tag] = current_mapped

            cb.set(current_mapped or "(None / Skip)")
            cb.pack(side=tk.LEFT, fill=tk.X, expand=True, padx=(2, 0))

            cb.bind("<<ComboboxSelected>>", lambda e, t=tag: self._on_mapping_combo_changed(t, e))
            self.mapping_combos[tag] = cb

        self._update_mapping_status_badge()

    def _populate_column_listbox(self):
        self.lb_cols.delete(0, tk.END)
        if self.df is not None:
            cols = list(self.df.columns)
            for c in cols:
                self.lb_cols.insert(tk.END, c)
            for idx, c in enumerate(cols):
                if c.lower() in ["id", "serial", "roll", "name", "full_name", "student_name", "candidate"]:
                    self.lb_cols.selection_set(idx)
            self._on_col_selected()

    def _select_all_cols(self):
        self.lb_cols.selection_set(0, tk.END)
        self._on_col_selected()

    def _clear_all_cols(self):
        self.lb_cols.selection_clear(0, tk.END)
        self._on_col_selected()

    def _on_col_selected(self, event=None):
        sel_indices = self.lb_cols.curselection()
        if self.df is not None and sel_indices:
            cols = [self.lb_cols.get(i) for i in sel_indices]
            self.selected_id_cols = cols
            first_vals = [str(self.df.iloc[0][c]) for c in cols if c in self.df.columns]
            preview = "_".join(first_vals) + ".pdf"
            self.lbl_filename_preview.configure(text=f"Filename Preview: {preview}")
        else:
            self.selected_id_cols = []
            self.lbl_filename_preview.configure(text="Filename Preview: [Row_Index].pdf")

    def _on_row_mode_changed(self):
        mode = self.row_sel_mode_var.get()
        if mode == "all":
            self.ent_row_range.configure(state="disabled")
        else:
            self.ent_row_range.configure(state="normal")
            self.ent_row_range.focus_set()
        self._update_row_selection_summary()

    def _update_row_selection_summary(self):
        if self.df is None or len(self.df) == 0:
            self.lbl_row_summary.configure(text="No data loaded yet.", fg="#94A3B8")
            return
        total = len(self.df)
        mode = self.row_sel_mode_var.get()
        if mode == "all":
            self.lbl_row_summary.configure(text=f"All {total} rows selected.", fg="#34D399")
        else:
            raw = self.row_range_var.get().strip()
            if not raw:
                self.lbl_row_summary.configure(text="Enter row numbers (e.g. 1-3, 5)", fg="#F59E0B")
            else:
                indices = parse_row_selection(raw, total)
                disp_rows = [i + 1 for i in indices]
                if len(disp_rows) <= 6:
                    disp_str = ", ".join(map(str, disp_rows))
                else:
                    disp_str = f"{', '.join(map(str, disp_rows[:5]))}... ({len(disp_rows)} rows)"
                self.lbl_row_summary.configure(text=f"Selected: {len(disp_rows)} of {total} rows ({disp_str})", fg="#38BDF8")

    def _get_active_df(self) -> pd.DataFrame:
        if self.df is None or len(self.df) == 0:
            return self.df
        if self.row_sel_mode_var.get() == "specific":
            raw = self.row_range_var.get().strip()
            indices = parse_row_selection(raw, len(self.df))
            return self.df.iloc[indices].copy()
        return self.df

    def _init_default_asset_configs(self):
        """
        Scans presentation and dataset to initialize active QR and Image configurations
        so that QR and Photo insertion are enabled by default whenever placeholders or assets exist.
        """
        if self.prs is None:
            return

        cols = list(self.df.columns) if self.df is not None else []
        id_col = self.selected_id_cols[0] if self.selected_id_cols else ("Serial" if "Serial" in cols else (cols[0] if cols else ""))

        # Scan slide placeholders
        discovered = MultiAssetEngine.scan_placeholders(self.prs) if "MultiAssetEngine" in globals() else {"images": [], "qrs": []}
        
        # 1. QR Configuration
        if self.qr_config is None:
            qr_tags = discovered["qrs"] if discovered["qrs"] else ["{{qr}}"]
            multi_qrs = {}
            for idx, tag in enumerate(qr_tags):
                multi_qrs[tag] = QRElementConfig(
                    tag=tag,
                    enabled=True,
                    payload_mode="column" if id_col else "static",
                    data_column=id_col,
                    static_payload="https://carnival2026.org/verify",
                    module_drawer="rounded",
                    preset_theme="corporate_blue" if idx == 0 else "classic_bw",
                    position_mode="custom",
                    custom_left_in=max(0.5, (self.prs.slide_width / 914400.0) - 2.4 - (idx * 2.0)),
                    custom_top_in=max(0.5, (self.prs.slide_height / 914400.0) - 2.4),
                    custom_size_in=1.6
                )
            primary_qr = list(multi_qrs.values())[0]
            self.qr_config = primary_qr.to_dict()
            self.qr_config["enabled"] = True
            self.qr_config["multi_qrs"] = multi_qrs

        # 2. Image Configuration
        if self.img_config is None:
            img_tags = discovered["images"] if discovered["images"] else ["{{photo}}"]
            default_folder = r"D:\New folder (3)\images" if os.path.isdir(r"D:\New folder (3)\images") else ""
            multi_images = {}
            for idx, tag in enumerate(img_tags):
                multi_images[tag] = ImageElementConfig(
                    tag=tag,
                    enabled=True,
                    source_mode="folder" if default_folder else "column",
                    image_folder=default_folder,
                    id_column=id_col,
                    data_column=cols[0] if cols else "",
                    shape_mode="rounded",
                    corner_radius=25,
                    border_width=3,
                    border_color="#1E3A8A",
                    position_mode="custom",
                    custom_left_in=max(0.5, (self.prs.slide_width / 914400.0) - 2.8 - (idx * 0.4)),
                    custom_top_in=max(0.5, 0.8 + (idx * 0.4)),
                    custom_width_in=1.8,
                    custom_height_in=2.2
                )
            primary_img = list(multi_images.values())[0]
            self.img_config = primary_img.to_dict()
            self.img_config["enabled"] = True
            self.img_config["multi_images"] = multi_images

        # Update Summary Badges
        if hasattr(self, "lbl_qr_summary"):
            qr_cnt = len(self.qr_config.get("multi_qrs", {}))
            self.lbl_qr_summary.configure(text=f"QR Code: Active ({qr_cnt})", fg="#34D399")
        if hasattr(self, "lbl_img_summary"):
            img_cnt = len(self.img_config.get("multi_images", {}))
            self.lbl_img_summary.configure(text=f"Image Asset: Active ({img_cnt})", fg="#34D399")


    def _launch_canvas_editor(self):
        if self.df is None or self.prs is None:
            messagebox.showinfo("Need Data & Template", "Please select both a data spreadsheet and PowerPoint template before opening the designer.")
            return

        pptx_p = self.pptx_path_var.get().strip() or None
        editor = VisualDragDropCanvas(self.df, self.prs, pptx_p, parent=self.root)
        raw_qr, raw_img = editor.show()

        if raw_qr is not None and raw_img is not None:
            qr_enabled = raw_qr.get("enabled", False)
            if qr_enabled:
                col = raw_qr.get("data_column") or raw_qr.get("data_template")
                self.lbl_qr_summary.configure(text=f"QR Code: Active ({col})", fg="#34D399")
            else:
                self.lbl_qr_summary.configure(text="QR Code: Inactive", fg="#94A3B8")

            img_enabled = raw_img.get("enabled", False)
            if img_enabled:
                src = raw_img.get("static_image_path") or raw_img.get("image_folder") or raw_img.get("data_column")
                self.lbl_img_summary.configure(text=f"Image Asset: Active ({raw_img.get('shape_mode')})", fg="#34D399")
            else:
                self.lbl_img_summary.configure(text="Image Asset: Inactive", fg="#94A3B8")

            self.qr_config = raw_qr
            self.img_config = raw_img
            messagebox.showinfo("Layout Applied", "Visual layout & styling applied successfully.")

    # -------------------------------------------------------------------------
    # Step 4 Page: Live Generation & Telemetry
    # -------------------------------------------------------------------------
    def _build_step_4_page(self):
        self.page4 = tk.Frame(self.container, bg="#0B1120")

        tk.Label(
            self.page4,
            text="Step 4: Live Generation & Telemetry",
            font=(FONT_FAMILY, 12, "bold"),
            fg="#38BDF8",
            bg="#0B1120"
        ).pack(anchor="w", pady=(0, 10))

        # Progress Status Card
        prog_card = tk.Frame(
            self.page4,
            bg="#162032",
            padx=16,
            pady=12,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        prog_card.pack(fill=tk.X, pady=(0, 10))

        self.lbl_prog_title = tk.Label(
            prog_card,
            text="Executing Automation Pipeline...",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#F8FAFC",
            bg="#162032"
        )
        self.lbl_prog_title.pack(anchor="w", pady=(0, 4))

        self.progress_bar = ttk.Progressbar(prog_card, orient=tk.HORIZONTAL, mode="determinate")
        self.progress_bar.pack(fill=tk.X, pady=(0, 6))

        self.lbl_prog_counter = tk.Label(
            prog_card,
            text="Processed: 0 / 0 items (0%)",
            font=(FONT_FAMILY, 9),
            fg="#38BDF8",
            bg="#162032"
        )
        self.lbl_prog_counter.pack(anchor="w")

        # Live Execution Log Box
        log_frame = tk.LabelFrame(
            self.page4,
            text="Execution Activity Log",
            font=(FONT_FAMILY, 9, "bold"),
            fg="#38BDF8",
            bg="#162032",
            padx=12,
            pady=8,
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        log_frame.pack(fill=tk.BOTH, expand=True)

        self.log_txt = tk.Text(
            log_frame,
            font=("Consolas", 8),
            bg="#0B1120",
            fg="#E2E8F0",
            relief="flat",
            wrap="word",
            highlightthickness=1,
            highlightbackground="#24344D"
        )
        log_sb = tk.Scrollbar(log_frame, orient=tk.VERTICAL, command=self.log_txt.yview)
        self.log_txt.configure(yscrollcommand=log_sb.set)
        self.log_txt.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        log_sb.pack(side=tk.RIGHT, fill=tk.Y)

    # -------------------------------------------------------------------------
    # Stepper Navigation & Button Control
    # -------------------------------------------------------------------------
    def _set_step(self, step: int):
        self.current_step = step

        # Update Stepper Pills Visuals
        for idx, (pill, lbl_ic, lbl_txt, icon_name) in enumerate(self.step_pills, start=1):
            if idx == step:
                pill.configure(bg="#0369A1", highlightbackground="#38BDF8")
                lbl_ic.configure(bg="#0369A1", image=MaterialIconManager.get(icon_name, size=(14, 14), color_hex="#FFFFFF"))
                lbl_txt.configure(bg="#0369A1", fg="#FFFFFF")
            elif idx < step:
                pill.configure(bg="#065F46", highlightbackground="#10B981")
                lbl_ic.configure(bg="#065F46", image=MaterialIconManager.get("check_circle", size=(14, 14), color_hex="#34D399"))
                lbl_txt.configure(bg="#065F46", fg="#34D399")
            else:
                pill.configure(bg="#1E293B", highlightbackground="#334155")
                lbl_ic.configure(bg="#1E293B", image=MaterialIconManager.get(icon_name, size=(14, 14), color_hex="#94A3B8"))
                lbl_txt.configure(bg="#1E293B", fg="#94A3B8")

        for p in [self.page1, self.page2, self.page3, self.page4]:
            p.pack_forget()

        if step == 1:
            self.page1.pack(fill=tk.BOTH, expand=True)
            self._on_mode_radio_changed()
            self.btn_back.configure(state="disabled")
            self.btn_cancel.configure(text="  Cancel", state="normal", bg="#334155", fg="#F8FAFC")
            self.btn_next.configure(text="  Next >", state="normal", bg="#0284C7", fg="#FFFFFF")
            self.btn_open_folder.configure(state="disabled")
            self.lbl_footer_status.configure(text="Select an operation mode and click 'Next >'.")

        elif step == 2:
            self.page2.pack(fill=tk.BOTH, expand=True)
            self._refresh_step_2_layout()
            self.btn_back.configure(state="normal")
            self.btn_cancel.configure(text="  Cancel", state="normal", bg="#334155", fg="#F8FAFC")
            self.btn_next.configure(text="  Next >", state="normal", bg="#0284C7", fg="#FFFFFF")
            self.btn_open_folder.configure(state="disabled")
            self.lbl_footer_status.configure(text="Choose your files or target folder, then click 'Next >'.")

        elif step == 3:
            self.page3.pack(fill=tk.BOTH, expand=True)
            self._refresh_step_3_layout()
            self.btn_back.configure(state="normal")
            self.btn_cancel.configure(text="  Cancel", state="normal", bg="#334155", fg="#F8FAFC")
            self._icons["btn_start"] = MaterialIconManager.get("play_arrow", size=(15, 15), color_hex="#FFFFFF")
            self.btn_next.configure(
                text="  [OK] Start Generation",
                image=self._icons["btn_start"],
                state="normal",
                bg="#10B981",
                fg="#FFFFFF"
            )
            self.btn_open_folder.configure(state="disabled")
            self.lbl_footer_status.configure(text="Configure output options & placeholder mappings, then click '[OK] Start Generation'.")

        elif step == 4:
            self.page4.pack(fill=tk.BOTH, expand=True)
            self.btn_back.configure(state="disabled")
            self.btn_next.configure(text="  [OK] Done", state="disabled", bg="#0284C7", fg="#FFFFFF")
            self.btn_cancel.configure(text="  Abort / Stop", state="normal", bg="#DC2626", fg="#FFFFFF")
            self.lbl_footer_status.configure(text="Automation pipeline running...")

    def _on_next_clicked(self):
        if self.current_step == 1:
            self._set_step(2)
        elif self.current_step == 2:
            m = self.mode_var.get()
            if m in ["1", "2", "5"]:
                if self.df is None or len(self.df) == 0:
                    messagebox.showwarning("Missing Data", "Please select a valid Excel or CSV spreadsheet before proceeding.")
                    return
                pptx_p = self.pptx_path_var.get().strip()
                if not pptx_p or not os.path.isfile(pptx_p):
                    messagebox.showwarning("Missing Template", "Please select a valid PowerPoint template (.pptx) before proceeding.")
                    return
            elif m in ["3", "4"]:
                tgt = self.target_dir_var.get().strip()
                if not tgt or not os.path.isdir(tgt):
                    messagebox.showwarning("Missing Folder", "Please select a valid target directory.")
                    return
                if m == "3" and len(self.batch_pairs) == 0:
                    messagebox.showwarning("No Pairs Found", "No matching .xlsx and .pptx pairs found in the selected folder.")
                    return

            self._set_step(3)

        elif self.current_step == 3:
            self._on_start_generation_clicked()

        elif self.current_step == 4:
            self._set_step(1)

    def _on_back_clicked(self):
        if self.current_step > 1 and not self.is_running:
            self._set_step(self.current_step - 1)

    def _on_cancel_clicked(self):
        if self.is_running:
            if messagebox.askyesno("Abort Operation", "Automation pipeline is running. Do you want to stop execution?"):
                self.abort_requested = True
                self._log("[!] Abort requested by user. Terminating worker...")
        else:
            if self.current_step > 1:
                if messagebox.askyesno("Exit to Mode Selection", "Do you want to return to Step 1 (Mode Selection)?"):
                    self._set_step(1)
            else:
                self.root.destroy()

    # -------------------------------------------------------------------------
    # Execution & Background Worker Logic
    # -------------------------------------------------------------------------
    def _log(self, msg: str):
        self.log_txt.insert(tk.END, msg + "\n")
        self.log_txt.see(tk.END)

    def _update_progress(self, current: int, total: int, status_text: str):
        pct = int((current / max(1, total)) * 100)
        self.progress_bar["value"] = pct
        self.lbl_prog_counter.configure(text=f"Processed: {current} / {total} items ({pct}%)")
        self.lbl_prog_title.configure(text=status_text)

    def _on_start_generation_clicked(self):
        if self.is_running:
            return

        self._set_step(4)
        self.is_running = True
        self.abort_requested = False
        self.progress_bar["value"] = 0
        self.log_txt.delete("1.0", tk.END)
        self._log("=" * 60)
        self._log("  AutoCrat Pro Generation Pipeline Started")
        self._log("=" * 60)

        t = threading.Thread(target=self._execute_worker, daemon=True)
        t.start()

    def _execute_worker(self):
        try:
            import pythoncom
            pythoncom.CoInitialize()
        except Exception:
            pass

        try:
            mode = self.mode_var.get()
            if mode == "1":
                self._run_mode_1_worker()
            elif mode == "2":
                self._run_mode_2_worker()
            elif mode == "3":
                self._run_mode_3_worker()
            elif mode == "4":
                self._run_mode_4_worker()
            elif mode == "5":
                self._run_mode_5_worker()
        except Exception as e:
            self._log(f"\n[FATAL ERROR] Worker terminated with exception: {e}")
            import traceback
            self._log(traceback.format_exc())
        finally:
            try:
                import pythoncom
                pythoncom.CoUninitialize()
            except Exception:
                pass

            self.is_running = False
            self.root.after(0, self._on_worker_finished)

    def _on_worker_finished(self):
        self.btn_cancel.configure(text="  Cancel", bg="#334155", fg="#F8FAFC")
        self._icons["btn_restart"] = MaterialIconManager.get("refresh", size=(15, 15), color_hex="#FFFFFF")
        self.btn_next.configure(
            text="  [OK] Start Another",
            image=self._icons["btn_restart"],
            state="normal",
            bg="#0284C7",
            fg="#FFFFFF"
        )
        self.btn_open_folder.configure(state="normal" if os.path.isdir(self.export_dir_var.get()) else "disabled")
        self.lbl_prog_title.configure(text="Automation Pipeline Completed!")
        self.lbl_footer_status.configure(text="Completed. Click '[OK] Start Another' or 'Open Output Folder'.")
        messagebox.showinfo("Execution Complete", f"AutoCrat Pro pipeline finished successfully!\nOutput folder: {self.export_dir_var.get()}")

    def _open_output_folder(self):
        out_f = self.export_dir_var.get().strip()
        if out_f and os.path.isdir(out_f):
            os.startfile(out_f)
        else:
            messagebox.showwarning("Folder Not Found", "Output folder does not exist yet.")

    # -------------------------------------------------------------------------
    # Mode Worker Implementations
    # -------------------------------------------------------------------------
    def _run_mode_1_worker(self):
        pptx_path = self.pptx_path_var.get().strip()
        export_dir = FileHelper.get_export_dir(pptx_path)
        self.export_dir_var.set(export_dir)

        qr_cfg = self.qr_config if self.qr_config is not None else {"enabled": False}
        img_cfg = self.img_config if self.img_config is not None else {"enabled": False}

        df_active = self._get_active_df()
        total_rows = len(df_active)
        self._log(f"\n[MODE 1] Preparing {total_rows} personalized presentations (from {len(self.df)} total dataset rows)...")
        self._log(f"Custom Placeholder Mappings Active: {len(self.placeholder_mapping)} tags")

        items = []
        for idx, row in df_active.iterrows():
            if self.selected_id_cols:
                vals = [str(row[c]).strip() for c in self.selected_id_cols if c in self.df.columns]
                fn_base = "_".join(vals)
            else:
                fn_base = f"document_{idx + 1}"
            fn_base = FileHelper.sanitize_filename(fn_base)

            items.append({
                "idx": idx,
                "row_dict": row.to_dict(),
                "fn_base": fn_base,
                "pptx_path": pptx_path,
                "export_dir": export_dir,
                "id_cols": self.selected_id_cols,
                "custom_mapping": self.placeholder_mapping,
                "qr_config": qr_cfg,
                "img_config": img_cfg,
                "total": total_rows
            })

        max_workers = min(os.cpu_count() or 4, 8)
        self._log(f"[*] Concurrency Engine: Using {max_workers} worker threads for PPTX generation...")

        generated_pptx_files = []
        completed_count = 0

        with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
            futures = [executor.submit(process_single_individual_pptx, item) for item in items]
            for fut in concurrent.futures.as_completed(futures):
                if self.abort_requested:
                    self._log("[!] Generation aborted by user.")
                    return
                res = fut.result()
                completed_count += 1
                if isinstance(res, tuple):
                    success, r_idx, temp_pptx, pdf_path, err = res
                    if success:
                        generated_pptx_files.append(temp_pptx)
                    else:
                        self._log(f"  [ERROR] Row {r_idx + 1}: {err}")
                elif isinstance(res, dict):
                    if res.get("status") == "ok" or res.get("success"):
                        generated_pptx_files.append(res.get("pptx_out") or res.get("temp_pptx"))
                    else:
                        self._log(f"  [ERROR] Row {res.get('idx', '?')}: {res.get('error')}")

                curr = completed_count
                self.root.after(0, lambda c=curr, t=total_rows: self._update_progress(c, t, f"Generated PPTX {c} of {t}"))

        self._log(f"[OK] Generated {len(generated_pptx_files)} presentation(s).")

        if self.export_pdf_var.get() and generated_pptx_files:
            self._log("\n[*] Initializing PowerPoint COM Engine for PDF export...")
            converter = PPTXToPDFConverter()
            if converter.start():
                self._log("[OK] PowerPoint COM engine active. Rendering PDFs...")
                pdf_count = 0
                total_pdf = len(generated_pptx_files)
                for p_file in generated_pptx_files:
                    if self.abort_requested:
                        self._log("[!] PDF conversion aborted by user.")
                        break
                    pdf_file = os.path.splitext(p_file)[0] + ".pdf"
                    if converter.convert(p_file, pdf_file):
                        pdf_count += 1
                        if not self.keep_pptx_var.get():
                            FileHelper.safe_delete(p_file)

                    curr_p = pdf_count
                    self.root.after(0, lambda c=curr_p, t=total_pdf: self._update_progress(c, t, f"Rendered PDF {c} of {t}"))
                converter.stop()
                self._log(f"[OK] Successfully rendered {pdf_count} vector PDF(s).")
            else:
                self._log("[WARN] PowerPoint COM unavailable; kept presentations as PPTX.")

        self._log(f"\n[OK] Mode 1 Complete! Output directory: {export_dir}")

    def _run_mode_2_worker(self):
        pptx_path = self.pptx_path_var.get().strip()
        export_dir = FileHelper.get_export_dir(pptx_path)
        self.export_dir_var.set(export_dir)

        output_name = self.merged_name_var.get().strip() or f"{os.path.splitext(os.path.basename(pptx_path))[0]}_merged"
        output_name = FileHelper.sanitize_filename(output_name)

        qr_cfg = self.qr_config if self.qr_config is not None else {"enabled": False}
        img_cfg = self.img_config if self.img_config is not None else {"enabled": False}

        prs = Presentation(pptx_path)
        slides_per_row = len(prs.slides)
        df_active = self._get_active_df()
        num_rows = len(df_active)

        self._log(f"\n[MODE 2] Creating Merged Master Document for {num_rows} recipients...")
        self._log(f"Custom Placeholder Mappings Active: {len(self.placeholder_mapping)} tags")
        self._log(f"Cloning slides ({num_rows * slides_per_row} total slides)...")

        template_slides = [prs.slides[i] for i in range(slides_per_row)]
        for r in range(1, num_rows):
            if self.abort_requested:
                self._log("[!] Merging stopped by user.")
                return
            for t_slide in template_slides:
                SlideDuplicator.duplicate_slide(prs, t_slide)
            self.root.after(0, lambda c=r, t=num_rows: self._update_progress(c, t, f"Cloning slides: {c}/{t}"))

        self._log("[*] Populating placeholders, QR codes & photos in presentation...")
        for loop_idx, (idx, row) in enumerate(df_active.iterrows()):
            if self.abort_requested:
                self._log("[!] Processing stopped by user.")
                return
            mapping = PlaceholderEngine.build_mapping(row, self.placeholder_mapping)

            for offset in range(slides_per_row):
                target_slide = prs.slides[loop_idx * slides_per_row + offset]
                PlaceholderEngine.process_slide(target_slide, mapping)
                MultiAssetEngine.embed_all_assets(target_slide, prs, row, mapping, img_cfg, qr_cfg, export_dir)

            curr = loop_idx + 1
            self.root.after(0, lambda c=curr, t=num_rows: self._update_progress(c, t, f"Populating data: {c}/{t}"))

        out_pptx = os.path.join(export_dir, f"{output_name}.pptx")
        prs.save(out_pptx)
        self._log(f"[OK] Saved merged PowerPoint: {output_name}.pptx")

        if self.export_pdf_var.get():
            out_pdf = os.path.join(export_dir, f"{output_name}.pdf")
            self._log("[*] Converting merged presentation to master PDF...")
            converter = PPTXToPDFConverter()
            if converter.start():
                converter.convert(out_pptx, out_pdf)
                converter.stop()
                self._log(f"[OK] Saved merged PDF: {output_name}.pdf")
            else:
                self._log("[WARN] PowerPoint COM unavailable; PDF conversion skipped.")

        self._log(f"\n[OK] Mode 2 Complete! Merged output in: {export_dir}")

    def _run_mode_3_worker(self):
        target_dir = self.target_dir_var.get().strip()
        export_dir = os.path.join(target_dir, "output")
        os.makedirs(export_dir, exist_ok=True)
        self.export_dir_var.set(export_dir)

        if not self.batch_pairs:
            self._scan_batch_directory(target_dir)

        if not self.batch_pairs:
            self._log(f"[!] No matching .xlsx and .pptx pairs found in {target_dir}")
            return

        self._log(f"\n[MODE 3] Processing {len(self.batch_pairs)} matching batch pair(s)...")

        qr_cfg = self.qr_config if self.qr_config is not None else {"enabled": False}
        img_cfg = self.img_config if self.img_config is not None else {"enabled": False}

        converter = PPTXToPDFConverter() if self.export_pdf_var.get() else None
        if converter and not converter.start():
            self._log("[WARN] PowerPoint COM unavailable; will produce PPTX only.")
            converter = None

        total_batches = len(self.batch_pairs)
        try:
            for p_idx, (xf, pf, base_name) in enumerate(self.batch_pairs):
                if self.abort_requested:
                    self._log("[!] Batch processing aborted by user.")
                    break

                xlsx_path = os.path.join(target_dir, xf)
                pptx_path = os.path.join(target_dir, pf)
                df = DataLoader.load(xlsx_path)
                if len(df) == 0:
                    self._log(f"[WARN] Batch '{base_name}' spreadsheet is empty. Skipping.")
                    continue

                prs = Presentation(pptx_path)
                slides_per_row = len(prs.slides)
                template_slides = [prs.slides[i] for i in range(slides_per_row)]

                self._log(f"\nProcessing Batch {p_idx + 1}/{total_batches}: '{base_name}' ({len(df)} rows)...")

                for r in range(1, len(df)):
                    for t_slide in template_slides:
                        SlideDuplicator.duplicate_slide(prs, t_slide)

                for loop_idx, (idx, row) in enumerate(df.iterrows()):
                    mapping = PlaceholderEngine.build_mapping(row, self.placeholder_mapping)

                    for offset in range(slides_per_row):
                        target_slide = prs.slides[loop_idx * slides_per_row + offset]
                        PlaceholderEngine.process_slide(target_slide, mapping)
                        MultiAssetEngine.embed_all_assets(target_slide, prs, row, mapping, img_cfg, qr_cfg, export_dir)

                out_pptx = os.path.join(export_dir, f"{base_name}_merged.pptx")
                prs.save(out_pptx)
                self._log(f"  [OK] Saved presentation: {base_name}_merged.pptx")

                if converter:
                    out_pdf = os.path.join(export_dir, f"{base_name}_merged.pdf")
                    converter.convert(out_pptx, out_pdf)
                    self._log(f"  [OK] Saved PDF: {base_name}_merged.pdf")

                curr_b = p_idx + 1
                self.root.after(0, lambda c=curr_b, t=total_batches: self._update_progress(c, t, f"Processed batch {c} of {t}"))

        finally:
            if converter:
                converter.stop()

        self._log(f"\n[OK] Mode 3 Complete! All batches saved in: {export_dir}")

    def _run_mode_4_worker(self):
        pdf_folder = self.target_dir_var.get().strip()
        self.export_dir_var.set(pdf_folder)
        output_name = self.merged_name_var.get().strip() or "Consolidated_Certificates.pdf"
        if not output_name.lower().endswith(".pdf"):
            output_name += ".pdf"

        def natural_sort_key(s):
            return [int(text) if text.isdigit() else text.lower() for text in re.split(r'(\d+)', s)]

        files = [f for f in os.listdir(pdf_folder) if f.lower().endswith(".pdf") and f != output_name]
        files.sort(key=natural_sort_key)

        if not files:
            self._log(f"[!] No PDF files found to merge in: {pdf_folder}")
            return

        self._log(f"\n[MODE 4] Merging {len(files)} PDF file(s) into: {output_name}...")
        try:
            import pypdf
            merger = PdfWriter()
        except Exception:
            from pypdf import PdfWriter
            merger = PdfWriter()
        total_f = len(files)

        for idx, f in enumerate(files):
            if self.abort_requested:
                self._log("[!] Merge aborted by user.")
                return
            p = os.path.join(pdf_folder, f)
            merger.append(p)
            curr = idx + 1
            self.root.after(0, lambda c=curr, t=total_f: self._update_progress(c, t, f"Merging PDF {c} of {t}"))

        out_path = os.path.join(pdf_folder, output_name)
        merger.write(out_path)
        merger.close()
        self._log(f"[OK] Saved consolidated PDF: {out_path}")
        self._log(f"\n[OK] Mode 4 Complete! Output in: {pdf_folder}")

    def _run_mode_5_worker(self):
        self._log("\n[MODE 5] Pre-Flight Template & Data Validator running...")
        pptx_path = self.pptx_path_var.get().strip()
        self.export_dir_var.set(os.path.dirname(pptx_path))

        prs = Presentation(pptx_path)
        tags = set(self.template_placeholders) if self.template_placeholders else set(PlaceholderEngine.extract_placeholders(prs))

        df_cols = set(self.df.columns)

        self._log(f"\n=== PRE-FLIGHT VALIDATION & MAPPING AUDIT ===")
        self._log(f"Template Slide Count    : {len(prs.slides)}")
        self._log(f"Spreadsheet Rows Loaded : {len(self.df)}")
        self._log(f"Total Template Tags     : {len(tags)}")
        self._log(f"Total Spreadsheet Cols  : {len(df_cols)}")

        self._log(f"\n[+] Active Placeholder Mappings:")
        for t in sorted(tags):
            mapped_col = self.placeholder_mapping.get(t)
            if mapped_col and mapped_col in df_cols:
                self._log(f"    * {{{{{t}}}}} --> Column '{mapped_col}' [MAPPED OK]")
            elif t in df_cols:
                self._log(f"    * {{{{{t}}}}} --> Column '{t}' [AUTO-MATCHED OK]")
            else:
                self._log(f"    * {{{{{t}}}}} --> [UNMAPPED / SKIPPED]")

        self.root.after(0, lambda: self._update_progress(100, 100, "Validation complete!"))
        self._log("\n[OK] Mode 5 Pre-Flight audit finished.")



    def _reset_all_mappings(self):
        """Clears all placeholder mappings to None/Skip."""
        for tag, cb in self.mapping_combos.items():
            cb.set("(None / Skip)")
            self.placeholder_mapping[tag] = ""
        self._update_mapping_status_badge()
        self._log("[MAPPER] Reset all placeholder mappings.")

    def _auto_map_all_placeholders(self):
        """Automatically fuzzy-matches all template tags to spreadsheet columns."""
        if not self.template_placeholders:
            return
        cols = list(self.df.columns) if self.df is not None else []
        mapped_c = 0
        for tag in self.template_placeholders:
            match = self.fuzzy_match_column(tag, cols)
            if match:
                self.placeholder_mapping[tag] = match
                if tag in self.mapping_combos:
                    self.mapping_combos[tag].set(match)
                mapped_c += 1
        self._update_mapping_status_badge()
        self._log(f"[AUTO-MAPPER] Successfully matched {mapped_c}/{len(self.template_placeholders)} placeholders to data columns.")

    def _open_data_preview_modal(self):
        """Opens an interactive modal to inspect loaded spreadsheet data rows and columns."""
        if self.df is None or len(self.df) == 0:
            messagebox.showinfo("No Data Loaded", "Please select and load a spreadsheet first.", parent=self.root)
            return

        modal = tk.Toplevel(self.root)
        modal.title(f"Spreadsheet Data Preview -- {len(self.df)} Rows, {len(self.df.columns)} Columns")
        modal.geometry("860x520")
        modal.minsize(680, 420)
        modal.configure(bg="#0B1120")
        modal.transient(self.root)

        # Header
        hdr = tk.Frame(modal, bg="#162032", padx=16, pady=10, relief="solid", bd=1)
        hdr.pack(fill=tk.X, padx=10, pady=(10, 4))

        tk.Label(
            hdr,
            text=f"Data Inspector: {self.data_path_var.get()}",
            font=("Segoe UI", 11, "bold"),
            fg="#38BDF8",
            bg="#162032"
        ).pack(anchor="w")

        info_txt = f"Total Records: {len(self.df)}  |  Total Columns: {len(self.df.columns)}  |  Sheet: {self.sheet_var.get()}"
        tk.Label(hdr, text=info_txt, font=("Segoe UI", 8), fg="#94A3B8", bg="#162032").pack(anchor="w")

        # Treeview Table
        table_frame = tk.Frame(modal, bg="#0F172A", relief="solid", bd=1)
        table_frame.pack(fill=tk.BOTH, expand=True, padx=10, pady=4)

        cols = ["#"] + list(self.df.columns)[:25]  # Display up to first 25 columns
        tree = ttk.Treeview(table_frame, columns=cols, show="headings", height=15)

        for col in cols:
            tree.heading(col, text=str(col))
            tree.column(col, width=110, minwidth=60, anchor="w")
        tree.column("#", width=50, stretch=False, anchor="center")

        tree_y = ttk.Scrollbar(table_frame, orient="vertical", command=tree.yview)
        tree_x = ttk.Scrollbar(table_frame, orient="horizontal", command=tree.xview)
        tree.configure(yscrollcommand=tree_y.set, xscrollcommand=tree_x.set)

        tree.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)
        tree_y.pack(side=tk.RIGHT, fill=tk.Y)
        tree_x.pack(side=tk.BOTTOM, fill=tk.X)

        # Populate rows
        for idx, row in self.df.iterrows():
            vals = [f"{idx+1:02d}"] + [str(row[c]) if pd.notna(row[c]) else "" for c in cols[1:]]
            tree.insert("", "end", values=vals)

        # Footer
        ftr = tk.Frame(modal, bg="#162032", padx=12, pady=8, relief="solid", bd=1)
        ftr.pack(fill=tk.X, padx=10, pady=(4, 10))

        tk.Button(
            ftr,
            text="Close Inspector",
            command=modal.destroy,
            font=("Segoe UI", 9, "bold"),
            bg="#334155",
            fg="#F8FAFC",
            activebackground="#475569",
            padx=14,
            pady=4,
            relief="flat",
            cursor="hand2"
        ).pack(side=tk.RIGHT)


def main_dashboard(initial_mode: str = "1"):
    """
    Main entry point: Launches the Interactive Desktop GUI Dashboard with Interactive Placeholder Mapping.
    Provides visual mode selection, Excel file/sheet picking, PowerPoint template scanning,
    custom placeholder mapping dropdowns, QR & photo drag-and-drop designer, and live progress bars.
    Gracefully falls back to interactive console menu if running in headless environments.
    """
    try:
        app = AutoCratProDashboardGUI(initial_mode=initial_mode)
        app.root.mainloop()
    except Exception as e:
        print(f"[NOTE] GUI Dashboard could not be launched ({e}). Falling back to interactive console menu.")
        main_dashboard_cli()


# Launch the Interactive Dashboard immediately upon running this cell
main_dashboard()
````
